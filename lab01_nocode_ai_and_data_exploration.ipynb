```python
# @title ⚙️ การตั้งค่า — โปรดรันเซลล์นี้ก่อน (NAME และ JOIN_CODE ใช้เฉพาะโจทย์ท้าทายบนตารางอันดับ ซึ่งไม่บังคับ)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้เป็นการ "เตรียมความพร้อม" ก่อนเริ่มปฏิบัติการ โดยทำ 3 อย่าง ได้แก่
#   1) ติดตั้งเครื่องมือ (ไลบรารี) เพิ่มเติมที่จำเป็นต้องใช้ใน Google Colab
#   2) กำหนดค่าพื้นฐาน เช่น รหัสปฏิบัติการ ชื่อที่จะแสดง และรหัสเข้าร่วมตารางอันดับ
#   3) สร้าง "ฟังก์ชัน" (ชุดคำสั่งสำเร็จรูปที่เรียกใช้ซ้ำได้) 2 ตัว คือ
#        - lab_data(...)              : ดาวน์โหลดไฟล์ข้อมูลของรายวิชา
#        - submit_to_leaderboard(...) : ส่งผลการทำนายไปยังตารางอันดับ (ไม่บังคับ)
#   ผู้เรียนเพียงกดรันเซลล์นี้ 1 ครั้ง ไม่จำเป็นต้องแก้ไขโค้ด
#   (ยกเว้นต้องการเข้าร่วมตารางอันดับ ให้กรอก NAME และ JOIN_CODE ด้านล่าง)
# ============================================================================

# ติดตั้งไลบรารีเพิ่มเติมลงในเครื่อง Colab
#   - เครื่องหมาย ! ด้านหน้า หมายถึงสั่งงานระบบปฏิบัติการ (ไม่ใช่คำสั่งภาษา Python)
#   - pip install : คำสั่งติดตั้งไลบรารีของ Python
#   - -q (quiet)  : ให้แสดงข้อความระหว่างติดตั้งให้น้อยที่สุด
#   - gradio_client   : ไลบรารีสำหรับติดต่อกับเว็บแอปตารางอันดับ (Hugging Face Space)
#   - huggingface_hub : ไลบรารีสำหรับดาวน์โหลดไฟล์จากเว็บไซต์ Hugging Face
!pip -q install gradio_client huggingface_hub

# สร้างตัวแปร (ช่องเก็บค่า) HF_OWNER แล้วใส่ข้อความ "wdittaya" ไว้
#   เครื่องหมาย = หมายถึง "นำค่าทางขวา ไปเก็บไว้ในชื่อทางซ้าย"
#   ข้อความในภาษา Python ต้องอยู่ในเครื่องหมายคำพูด " "
HF_OWNER  = "wdittaya"     # บัญชี Hugging Face ที่ใช้โฮสต์ตารางอันดับของรายวิชา (ผู้สอนเป็นผู้กำหนด)
# รหัสของปฏิบัติการนี้ ใช้ระบุว่าผลที่ส่งเป็นของปฏิบัติการใด (ไม่ต้องแก้ไข)
LAB_ID    = "lab01"
# ชื่อที่จะแสดงบนตารางอันดับ: พิมพ์ชื่อไว้ระหว่างเครื่องหมาย " " เช่น NAME = "หมอเอ"
NAME      = ""            # ไม่บังคับ: ชื่อที่จะแสดงบนตารางอันดับแบบสาธารณะ (สามารถใช้นามแฝงได้)
# รหัสเข้าร่วมตารางอันดับ: พิมพ์รหัสที่ผู้สอนแจ้งไว้ระหว่างเครื่องหมาย " "
JOIN_CODE = ""            # ไม่บังคับ: รหัสเข้าร่วมซึ่งผู้สอนประกาศในการเรียนการสอนผ่าน Zoom

# สร้างชื่อเว็บแอปตารางอันดับ โดยนำ HF_OWNER มาต่อกับข้อความ
#   - f"..." (f-string) คือข้อความที่แทรกค่าตัวแปรได้ โดยเขียนชื่อตัวแปรไว้ใน { }
#   - ผลลัพธ์คือ "wdittaya/aidh-leaderboard"
SPACE_ID  = f"{HF_OWNER}/aidh-leaderboard"
# สร้างชื่อคลังข้อมูล (dataset) ของรายวิชาในลักษณะเดียวกัน ผลลัพธ์คือ "wdittaya/aidh-lab-data"
DATA_REPO = f"{HF_OWNER}/aidh-lab-data"

# นำเข้า (import) ฟังก์ชัน hf_hub_download จากไลบรารี huggingface_hub
#   ฟังก์ชันนี้ใช้ดาวน์โหลดไฟล์จากเว็บไซต์ Hugging Face มาเก็บไว้ในเครื่อง
from huggingface_hub import hf_hub_download

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน lab_data สำหรับดาวน์โหลดไฟล์ข้อมูลของรายวิชา
#   - def คือคำสั่งสร้างฟังก์ชันใหม่ ตามด้วยชื่อฟังก์ชันและพารามิเตอร์ในวงเล็บ
#   - path : พารามิเตอร์ (ค่าที่ผู้ใช้ส่งเข้ามา) คือชื่อไฟล์ที่ต้องการ เช่น "pima/test_ids.csv"
#   - บรรทัดที่ย่อหน้าเข้าไปด้านล่าง คือคำสั่งที่อยู่ภายในฟังก์ชัน
# ----------------------------------------------------------------------------
def lab_data(path):
    """ดาวน์โหลดไฟล์ข้อมูลทดสอบสาธารณะของรายวิชา 1 ไฟล์ (เช่น 'pima/test_ids.csv') และคืนค่าตำแหน่งไฟล์ในเครื่อง"""
    # เรียก hf_hub_download(...) เพื่อดาวน์โหลดไฟล์ แล้วส่งตำแหน่งไฟล์ในเครื่องกลับไป (return)
    #   - DATA_REPO           : ชื่อคลังข้อมูลบน Hugging Face ที่เก็บไฟล์
    #   - path                : ชื่อไฟล์ที่ต้องการดาวน์โหลดภายในคลังข้อมูลนั้น
    #   - repo_type="dataset" : ระบุว่าคลังนี้เป็นประเภท "ชุดข้อมูล" (ไม่ใช่โมเดล)
    return hf_hub_download(DATA_REPO, path, repo_type="dataset")

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน submit_to_leaderboard สำหรับส่งไฟล์ผลการทำนายไปยังตารางอันดับ
#   - path   : ชื่อไฟล์ผลการทำนายที่ต้องการส่ง
#   - lab=LAB_ID : รหัสปฏิบัติการ หากไม่ระบุ จะใช้ค่า LAB_ID ที่กำหนดไว้ด้านบนโดยอัตโนมัติ
# ----------------------------------------------------------------------------
def submit_to_leaderboard(path, lab=LAB_ID):
    """ส่งไฟล์ผลการทำนายไปยังตารางอันดับของรายวิชา และแสดงคะแนนพร้อมอันดับ"""
    # นำเข้าเครื่องมือ 2 ตัวจากไลบรารี gradio_client
    #   - Client      : ใช้เชื่อมต่อกับเว็บแอปตารางอันดับ
    #   - handle_file : ใช้เตรียมไฟล์ในเครื่องให้อยู่ในรูปแบบที่ส่งขึ้นเว็บแอปได้
    from gradio_client import Client, handle_file
    # assert คือการตรวจสอบเงื่อนไข หากเงื่อนไขเป็นเท็จ โปรแกรมจะหยุดและแสดงข้อความเตือน
    #   - NAME and JOIN_CODE : เงื่อนไขว่าต้องกรอกทั้ง NAME และ JOIN_CODE แล้ว (ไม่เป็นข้อความว่าง)
    #   - ข้อความหลังเครื่องหมายจุลภาค คือข้อความเตือนที่จะแสดงเมื่อยังไม่ได้กรอก
    assert NAME and JOIN_CODE, "โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่าก่อน (จำเป็นเฉพาะการส่งผลไปยังตารางอันดับ)"
    # เชื่อมต่อกับเว็บแอปตารางอันดับ แล้วเก็บการเชื่อมต่อไว้ในตัวแปร client
    #   - SPACE_ID      : ชื่อเว็บแอปตารางอันดับที่กำหนดไว้ด้านบน
    #   - verbose=False : ไม่ต้องแสดงข้อความรายละเอียดระหว่างเชื่อมต่อ
    client = Client(SPACE_ID, verbose=False)
    # ส่งข้อมูลไปยังเว็บแอปด้วย client.predict(...) แล้วแสดงผลลัพธ์ (คะแนนและอันดับ) ด้วย print(...)
    #   - lab=lab                 : รหัสปฏิบัติการ
    #   - name=NAME               : ชื่อที่จะแสดงบนตารางอันดับ
    #   - code=JOIN_CODE          : รหัสเข้าร่วม
    #   - file=handle_file(path)  : ไฟล์ผลการทำนายที่เตรียมพร้อมสำหรับส่ง
    #   - api_name="/submit"      : ชื่อช่องทางรับข้อมูลของเว็บแอป (ส่วน "ส่งผล")
    print(client.predict(lab=lab, name=NAME, code=JOIN_CODE, file=handle_file(path), api_name="/submit"))

# แสดงข้อความยืนยันว่าการตั้งค่าเสร็จแล้ว ด้วยฟังก์ชัน print(...) (พิมพ์ข้อความออกทางหน้าจอ)
#   - ข้อความเป็น f-string จึงแทรกค่า SPACE_ID, LAB_ID และ NAME ลงในข้อความได้
#   - NAME or '(ยังไม่ได้ระบุ ...)' : หาก NAME ว่าง จะแสดงข้อความในวงเล็บแทน
print(f"Leaderboard: https://huggingface.co/spaces/{SPACE_ID}  |  lab={LAB_ID}  name={NAME or '(ยังไม่ได้ระบุ — ตารางอันดับไม่บังคับ)'}")
```

ส่วนนี้ไม่ต้องเขียนโปรแกรม ขั้นตอนทั้งหมดดำเนินการบนเว็บเบราว์เซอร์

1. เปิด [Teachable Machine](https://teachablemachine.withgoogle.com/) → **Get Started → Image Project → Standard image model**
2. สร้างคลาส (class) 2–3 คลาส โดยเปลี่ยนชื่อ "Class 1", "Class 2", … เช่น *แบมือ*, *กำมือ*, *ไม่มีมือ* แล้วเพิ่มภาพด้วย **Webcam** (กดปุ่มค้างไว้เพื่อบันทึกภาพต่อเนื่อง) หรือ **Upload** (หากไม่มีกล้องเว็บแคม สามารถใช้ภาพจากโทรศัพท์หรือคอมพิวเตอร์ได้)
   สำหรับโจทย์ทางการแพทย์ (ไม่บังคับ): ดาวน์โหลดภาพรอยโรคผิวหนังสาธารณะจาก [ISIC Archive](https://gallery.isic-archive.com/) เพื่อสร้างคลาส *benign (ไม่ร้ายแรง) และ malignant (มะเร็ง)*
3. คลิก **Train Model** แล้วทดสอบในแผง **Preview** ด้วยภาพ *ใหม่* ที่โมเดลไม่เคยใช้ฝึก พร้อมบันทึกข้อสังเกต ดังนี้
   - เริ่มจากภาพจำนวนน้อย (ประมาณ 5 ภาพต่อคลาส) และตรวจสอบว่าโมเดลทำงานได้หรือไม่ จากนั้นเพิ่มภาพและฝึกใหม่
   - ทดสอบโดยเปลี่ยนฉากหลังหรือแสง (เช่น เปิดหรือปิดไฟ หรือเปลี่ยนตำแหน่ง)
4. คลิก **Export Model** เพื่อพิจารณาว่าโมเดลที่ฝึกแล้วสามารถนำไปใช้ในเว็บไซต์หรือแอปพลิเคชันได้อย่างไร ขั้นตอนนี้คือ "การนำไปใช้งานจริง (deployment)" (ไม่จำเป็นต้องดาวน์โหลดไฟล์ใด)

> 💡 **คำศัพท์สำคัญของรายวิชา:** *คลาส (class)* คือหมวดหมู่ที่โมเดลต้องเลือก; *การฝึกสอน (training)* คือกระบวนการที่โมเดลเรียนรู้รูปแบบจากตัวอย่าง; *การเปลี่ยนแปลงของการกระจายข้อมูล (distribution shift)* คือสภาวะที่ข้อมูลใหม่มีลักษณะแตกต่างจากข้อมูลที่ใช้ฝึก (เช่น สภาพแสงต่างกัน เครื่องสแกนของโรงพยาบาลอื่น หรือกลุ่มผู้ป่วยต่างกลุ่ม) ซึ่งเป็นปัญหาสำคัญที่สุดประการหนึ่งของปัญญาประดิษฐ์ทางการแพทย์

## ส่วน B — การเขียนโปรแกรม: การสำรวจชุดข้อมูลเบาหวาน Pima (~40 นาที)

ปฏิบัติการนี้ใช้ชุดข้อมูลสาธารณะ **Pima Indians Diabetes** ซึ่งประกอบด้วยข้อมูลของสตรีเชื้อสายชนพื้นเมืองอเมริกันเผ่า Pima อายุ 21 ปีขึ้นไป จำนวน 768 ราย จากการศึกษาของ US National Institute of Diabetes and Digestive and Kidney Diseases (NIDDK) โดย `Outcome` = 1 หมายถึงเป็นโรคเบาหวาน (ตามเกณฑ์ขององค์การอนามัยโลก (WHO)) และ 0 หมายถึงไม่เป็นโรคเบาหวาน

| คอลัมน์ | ความหมาย |
|---|---|
| `Pregnancies` | จำนวนครั้งของการตั้งครรภ์ |
| `Glucose` | ระดับน้ำตาลในพลาสมา (mg/dL) **ที่ 2 ชั่วโมงหลังดื่มกลูโคส 75 g ในการทดสอบความทนต่อกลูโคส (oral glucose tolerance test, OGTT)** — *ไม่ใช่* fasting glucose |
| `BloodPressure` | ความดันโลหิตขณะหัวใจคลายตัว (diastolic, mmHg) |
| `SkinThickness` | ความหนาของชั้นผิวหนังบริเวณกล้ามเนื้อไตรเซปส์ (triceps skin-fold, mm) |
| `Insulin` | ระดับอินซูลินในซีรัมที่ 2 ชั่วโมง (µU/mL) |
| `BMI` | ดัชนีมวลกาย (kg/m²) |
| `DiabetesPedigreeFunction` | คะแนนที่สะท้อนประวัติโรคเบาหวานในครอบครัว |
| `Age` | อายุ (ปี) |
| `Outcome` | 1 = เป็นโรคเบาหวาน, 0 = ไม่เป็นโรคเบาหวาน |

ส่วนนี้ใช้ **pandas** ซึ่งเป็นไลบรารีมาตรฐานของภาษา Python สำหรับจัดการข้อมูลตาราง (เทียบได้กับ Excel ในรูปแบบโปรแกรม) ตารางใน pandas เรียกว่า *DataFrame* และในปฏิบัติการนี้จัดเก็บไว้ในตัวแปร `df` ให้รันเซลล์ด้านล่างตามลำดับ

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้เป็นจุดเริ่มต้นของส่วน B โดยทำ 2 อย่าง ได้แก่
#   1) นำเข้า (import) ไลบรารีหลัก 4 ตัวที่ใช้ตลอดปฏิบัติการ
#        - pandas     : จัดการข้อมูลตาราง (เทียบได้กับ Excel ในรูปแบบโปรแกรม)
#        - numpy      : คำนวณตัวเลข และมีค่าพิเศษ NaN สำหรับแทน "ค่าว่าง"
#        - matplotlib : วาดกราฟพื้นฐาน
#        - seaborn    : วาดกราฟทางสถิติที่สวยงามขึ้น (สร้างต่อยอดจาก matplotlib)
#   2) โหลดชุดข้อมูลเบาหวาน Pima (ผู้ป่วย 768 ราย) จากอินเทอร์เน็ตเข้ามาเป็นตาราง
#      ชื่อ df (ย่อมาจาก DataFrame = ตารางข้อมูลของ pandas)
#   สิ่งที่ผู้เรียนจะเห็น: ขนาดของตาราง (จำนวนแถว, จำนวนคอลัมน์) และข้อมูลผู้ป่วย 5 รายแรก
#   ซึ่งเป็นการ "ดูหน้าตาข้อมูล" ครั้งแรก ก่อนเริ่มสำรวจและทำความสะอาดข้อมูล
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: นำเข้าไลบรารี
#   - import ... as ... คือการนำเครื่องมือ (ไลบรารี) เข้ามาใช้งาน
#     และตั้ง "ชื่อเล่น" ที่สั้นลงหลังคำว่า as เพื่อให้พิมพ์สะดวก
#   - ชื่อเล่น pd, np, plt, sns เป็นธรรมเนียมที่นักวิเคราะห์ข้อมูลทั่วโลกใช้กัน
# ----------------------------------------------------------------------------
# นำเข้าไลบรารี pandas โดยตั้งชื่อเล่นว่า pd (เรียกใช้ภายหลังเป็น pd.xxx เช่น pd.read_csv)
import pandas as pd
# นำเข้าไลบรารี numpy (Numerical Python) สำหรับงานคำนวณตัวเลข โดยตั้งชื่อเล่นว่า np
#   ในปฏิบัติการนี้ใช้ np.nan ซึ่งเป็นค่าพิเศษที่หมายถึง "ค่าว่าง/ไม่มีข้อมูล"
import numpy as np
# นำเข้าโมดูลย่อย pyplot ของไลบรารี matplotlib สำหรับวาดกราฟ โดยตั้งชื่อเล่นว่า plt
#   - จุด (.) ใน matplotlib.pyplot หมายถึง "ส่วนย่อย pyplot ที่อยู่ภายใน matplotlib"
import matplotlib.pyplot as plt
# นำเข้าไลบรารี seaborn สำหรับวาดกราฟทางสถิติ (เช่น แผนภาพความร้อน) โดยตั้งชื่อเล่นว่า sns
import seaborn as sns

# ----------------------------------------------------------------------------
# ส่วนที่ 2: โหลดชุดข้อมูลและดูภาพรวมเบื้องต้น
# ----------------------------------------------------------------------------
# สร้างตัวแปร URL เก็บที่อยู่เว็บ (ลิงก์) ของไฟล์ข้อมูล diabetes.csv
#   - ตัวแปร คือ "ช่องเก็บค่า" ที่ตั้งชื่อไว้ เครื่องหมาย = หมายถึงนำค่าทางขวาไปเก็บในชื่อทางซ้าย
#   - ข้อความ (string) ต้องอยู่ในเครื่องหมายคำพูด " "
#   - ไฟล์ .csv (comma-separated values) คือไฟล์ตารางที่คั่นแต่ละช่องด้วยเครื่องหมายจุลภาค (,)
#     เปิดด้วย Excel ได้
URL = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"
# อ่านไฟล์ CSV จากลิงก์ด้วยฟังก์ชัน pd.read_csv(...) ของ pandas แล้วเก็บตารางที่ได้ไว้ในตัวแปร df
#   - URL : ตำแหน่งของไฟล์ที่ต้องการอ่าน (เป็นลิงก์เว็บหรือชื่อไฟล์ในเครื่องก็ได้)
#   - ผลลัพธ์เป็น DataFrame (ตารางของ pandas): 1 แถว = ผู้ป่วย 1 ราย, 1 คอลัมน์ = ตัวแปรที่วัด 1 ตัว
df = pd.read_csv(URL)
# แสดงขนาดของตารางออกทางหน้าจอด้วย print(...) (ฟังก์ชันพื้นฐานของ Python สำหรับพิมพ์ผล)
#   - df.shape : คุณสมบัติ (attribute) ของตาราง บอก (จำนวนแถว, จำนวนคอลัมน์) ไม่ต้องมีวงเล็บ ()
#     เพราะเป็นค่าที่เก็บไว้ ไม่ใช่คำสั่งที่ต้องเรียกทำงาน — ผลที่ควรเห็นคือ (768, 9)
#   - ข้อความหลังเครื่องหมาย # คือคอมเมนต์ (comment) ซึ่ง Python จะไม่นำไปประมวลผล
print(df.shape)   # (จำนวนแถว, จำนวนคอลัมน์) = (จำนวนผู้ป่วย, จำนวนตัวแปรที่วัด)
# แสดงข้อมูล 5 แถวแรกของตารางด้วยเมธอด (method = คำสั่งที่ติดมากับตาราง) .head()
#   - ไม่ระบุค่าในวงเล็บ จึงใช้ค่าเริ่มต้น 5 แถว (หากต้องการ 10 แถว เขียน df.head(10))
#   - ใน Jupyter/Colab คำสั่งบรรทัดสุดท้ายของเซลล์จะถูกแสดงผลบนหน้าจอโดยอัตโนมัติ
#     (ไม่ต้องใช้ print) และแสดงเป็นตารางที่อ่านง่าย
df.head()         # แสดงข้อมูลผู้ป่วย 5 รายแรก
```

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   ใช้คำสั่งพื้นฐานของ pandas 2 คำสั่ง เพื่อ "ตรวจสุขภาพ" ของชุดข้อมูลเบื้องต้น
#   1) df.info()     : ดูโครงสร้างตาราง ได้แก่ ชื่อคอลัมน์ ชนิดข้อมูล และจำนวนค่าที่ไม่ว่าง
#   2) df.describe() : ดูสถิติสรุป (descriptive statistics) ของแต่ละคอลัมน์ที่เป็นตัวเลข
#   สิ่งที่ผู้เรียนจะเห็น: รายการคอลัมน์ทั้ง 9 คอลัมน์ และตารางสถิติ (count, mean, std,
#   min, 25%, 50%, 75%, max) ซึ่งจะช่วยให้สังเกตเห็นความผิดปกติ เช่น ค่าต่ำสุดเป็น 0
#   ในตัวแปรที่เป็น 0 ไม่ได้ทางสรีรวิทยา (เช่น ระดับน้ำตาล ความดันโลหิต)
# ============================================================================

# คำสั่ง pandas พื้นฐาน 2 คำสั่งที่ใช้บ่อย

# เรียกเมธอด .info() ของตาราง df เพื่อพิมพ์สรุปโครงสร้างตาราง ได้แก่
#   - จำนวนแถวทั้งหมด (RangeIndex) และจำนวนคอลัมน์
#   - ชื่อแต่ละคอลัมน์ (Column)
#   - จำนวนค่าที่ไม่ว่าง (Non-Null Count) — หากน้อยกว่าจำนวนแถว แปลว่ามีค่าว่าง
#   - ชนิดข้อมูล (Dtype) เช่น int64 = จำนวนเต็ม, float64 = ทศนิยม
#   - ไม่ต้องใส่ค่าใดในวงเล็บ (ใช้ค่าเริ่มต้นทั้งหมด) และคำสั่งนี้พิมพ์ผลออกหน้าจอเอง
df.info()         # ชื่อคอลัมน์ ชนิดข้อมูล และจำนวนค่าที่ไม่ว่าง
# เรียกเมธอด .describe() ของตาราง df เพื่อคำนวณสถิติสรุปของทุกคอลัมน์ที่เป็นตัวเลข
#   - count : จำนวนค่าที่ไม่ว่าง
#   - mean  : ค่าเฉลี่ย, std : ส่วนเบี่ยงเบนมาตรฐาน (standard deviation)
#   - min / max : ค่าต่ำสุด / ค่าสูงสุด
#   - 25% / 50% / 75% : ควอร์ไทล์ (quartile) โดย 50% คือค่ามัธยฐาน (median)
#   - เป็นบรรทัดสุดท้ายของเซลล์ Jupyter จึงแสดงผลเป็นตารางโดยอัตโนมัติ
df.describe()     # สถิติสรุปของแต่ละคอลัมน์: mean, std, min, ควอร์ไทล์, max
```

### ค่าสูญหายที่แฝงอยู่ (hidden missing values)

จากแถว `min` ข้างต้น ค่าต่ำสุดของ `Glucose`, `BloodPressure`, `BMI` … มีค่าเป็น **0** ซึ่งเป็นไปไม่ได้ทางสรีรวิทยา
ในข้อมูลทางคลินิกจริง ค่าสูญหาย (missing values) มักถูกบันทึกแฝงไว้ในรูปของเลข 0 หรือรหัสพิเศษ (เช่น 999) เซลล์ถัดไปจะแปลงเลข 0 ดังกล่าวให้เป็นค่าว่าง (`NaN` = *not a number*)

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   แปลง "ค่าสูญหายที่แฝงอยู่ (hidden missing values)" ให้เป็นค่าว่างที่แท้จริง แล้วนับจำนวน
#   - ในชุดข้อมูลนี้ ค่า 0 ในคอลัมน์ Glucose, BloodPressure, SkinThickness, Insulin และ BMI
#     เป็นไปไม่ได้ทางสรีรวิทยา จึงหมายถึง "ไม่ได้ตรวจวัด/ไม่ได้บันทึก"
#   - เซลล์นี้เปลี่ยนเลข 0 เหล่านั้นเป็น NaN (not a number = ค่าว่าง) ที่ pandas รู้จัก
#   - เก็บสำเนาข้อมูลที่ยังมีค่าว่างไว้ในชื่อ df_with_missing เพื่อใช้ในแบบฝึกหัดภายหลัง
#   - นับจำนวนค่าสูญหายในแต่ละคอลัมน์ และวาดกราฟแท่งแนวนอนเปรียบเทียบ
#   สิ่งที่ผู้เรียนจะเห็น: กราฟแท่งจำนวนค่าสูญหายต่อคอลัมน์ และตารางตัวเลขจำนวนค่าสูญหาย
#   (ใช้ตอบคำถาม B1 ว่าคอลัมน์ใดมีค่าสูญหายมากที่สุด)
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: เปลี่ยนเลข 0 ที่เป็นไปไม่ได้ทางคลินิกให้เป็นค่าว่าง (NaN)
# ----------------------------------------------------------------------------
# สร้างตัวแปร zero_as_missing เก็บ "รายชื่อคอลัมน์" ที่เลข 0 หมายถึงค่าสูญหาย
#   - [ ... ] คือ list (รายการ) ของ Python ใช้เก็บหลายค่าเรียงกัน คั่นด้วยเครื่องหมายจุลภาค
#   - แต่ละรายการเป็นข้อความ (ชื่อคอลัมน์) จึงอยู่ในเครื่องหมาย " "
zero_as_missing = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
# แทนเลข 0 ด้วยค่าว่าง (NaN) ในคอลัมน์ทั้ง 5 แล้วเขียนทับคอลัมน์เดิมในตาราง df
#   - df[zero_as_missing] : เลือกเฉพาะคอลัมน์ที่อยู่ใน list (ได้ตารางย่อย 5 คอลัมน์)
#   - .replace(0, np.nan) : เมธอดของ pandas สำหรับ "ค้นหาและแทนที่" ค่า
#       - 0      : ค่าที่ต้องการค้นหา
#       - np.nan : ค่าใหม่ที่ใช้แทน คือ NaN (ค่าว่าง) จากไลบรารี numpy
#   - เครื่องหมาย = ด้านซ้ายมี df[zero_as_missing] หมายถึงนำผลลัพธ์ไปเขียนทับคอลัมน์เหล่านั้น
#   - หมายเหตุ: คอลัมน์อื่น เช่น Pregnancies ไม่ถูกเปลี่ยน เพราะ 0 ครั้งเป็นค่าที่เป็นไปได้จริง
df[zero_as_missing] = df[zero_as_missing].replace(0, np.nan)
# ทำสำเนาตาราง df ทั้งตาราง (ที่ยังมีค่าว่างอยู่) ด้วยเมธอด .copy() แล้วเก็บไว้ในชื่อ df_with_missing
#   - ต้องใช้ .copy() เพื่อให้ได้ตารางแยกต่างหาก การแก้ df ภายหลังจะไม่กระทบสำเนานี้
#     (หากเขียนเพียง df_with_missing = df จะเป็นเพียงชื่อเรียกอีกชื่อของตารางเดียวกัน)
df_with_missing = df.copy()   # เก็บสำเนาที่ยังมีค่าว่าง สำหรับจัดทำรายงานสรุปข้อมูลในภายหลัง

# ----------------------------------------------------------------------------
# ส่วนที่ 2: นับจำนวนค่าสูญหายต่อคอลัมน์ และแสดงเป็นกราฟ
# ----------------------------------------------------------------------------
# นับจำนวนค่าว่างของแต่ละคอลัมน์ แล้วเรียงจากมากไปน้อย เก็บไว้ในตัวแปร missing
#   คำสั่งนี้เป็นการเรียกเมธอดต่อกันเป็นทอด (chained calls) ทำงานจากซ้ายไปขวา ดังนี้
#   1) df.isnull() : สร้างตารางขนาดเท่าเดิม ช่องใดว่างได้ True (จริง) ไม่ว่างได้ False (เท็จ)
#   2) .sum()      : รวมค่าในแต่ละคอลัมน์ (True นับเป็น 1, False นับเป็น 0)
#                    จึงได้ "จำนวนค่าว่างต่อคอลัมน์" ในรูป Series (ข้อมูล 1 คอลัมน์ที่มีชื่อกำกับ)
#   3) .sort_values(ascending=False) : เรียงลำดับค่า
#       - ascending=False : ไม่เรียงจากน้อยไปมาก กล่าวคือเรียงจาก "มากไปน้อย"
missing = df.isnull().sum().sort_values(ascending=False)
# วาดกราฟแท่งแนวนอน (horizontal bar chart) จากข้อมูล missing
#   - .plot.barh(...) : เมธอดวาดกราฟของ pandas (ใช้ matplotlib อยู่เบื้องหลัง) barh = bar แนวนอน
#   - title="Missing values per column" : ชื่อกราฟที่แสดงด้านบน ("จำนวนค่าสูญหายต่อคอลัมน์")
missing.plot.barh(title="Missing values per column")
# สั่งแสดงกราฟที่วาดไว้บนหน้าจอด้วย plt.show() ของ matplotlib (ไม่มีพารามิเตอร์)
plt.show()
# บรรทัดสุดท้ายของเซลล์: พิมพ์ชื่อตัวแปร missing เพื่อให้ Jupyter แสดงตัวเลขจำนวนค่าสูญหาย
# ของแต่ละคอลัมน์โดยอัตโนมัติ
missing
```

**✍️ B1.** โปรดระบุคอลัมน์ที่มีค่าสูญหายแฝงอยู่มากที่สุด และเสนอเหตุผลทางคลินิก 1 ประการที่ทำให้ตัวแปรนี้มักไม่ได้รับการตรวจวัดหรือบันทึก

### การแทนค่าสูญหาย (imputation)

การวิเคราะห์หลายวิธีไม่สามารถใช้ข้อมูลที่มีค่าว่างได้ วิธีแก้ไขเบื้องต้นคือการแทนค่าว่างด้วยค่าตัวแทนของคอลัมน์นั้น (*imputation*) ในที่นี้ใช้ **มัธยฐาน (median)** ซึ่งเป็นค่ากลางของข้อมูล เนื่องจากมัธยฐานไม่ได้รับผลกระทบจากค่าผิดปกติ (outlier) เพียงไม่กี่ค่า ต่างจากค่าเฉลี่ย

```python
# ⭐ Exercise 1 — แบบฝึกหัดที่ 1: การแทนค่าสูญหายด้วยค่ามัธยฐาน
# วัตถุประสงค์: แทนค่าว่างทุกตำแหน่งในคอลัมน์ที่มีค่าสูญหาย ด้วยค่ามัธยฐาน (median) ของคอลัมน์นั้น
# สิ่งที่ต้องแก้ไข: แทน ... ด้วยค่ามัธยฐานของคอลัมน์
# คำใบ้: ค่ามัธยฐานของคอลัมน์เขียนได้เป็น  df_with_missing[col].median()
# (เซลล์นี้เริ่มจากสำเนา `df_with_missing` จึงสามารถรันซ้ำได้โดยไม่กระทบข้อมูล)

# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   ฝึก "การแทนค่าสูญหาย (imputation)" ด้วยค่ามัธยฐาน (median) ของแต่ละคอลัมน์
#   - วนทำงานทีละคอลัมน์ใน zero_as_missing (Glucose, BloodPressure, SkinThickness, Insulin, BMI)
#   - นำข้อมูลจากสำเนา df_with_missing (ที่ยังมีค่าว่าง) มาเติมค่าว่าง แล้วเขียนลงตาราง df
#   - จากนั้นตรวจสอบอัตโนมัติ 2 ข้อ ว่าทำถูกต้องหรือไม่
#   ผู้เรียนต้องแก้ไขเฉพาะเครื่องหมาย ... (ช่องว่างให้เติม) ตามคำใบ้ด้านบน
#   สิ่งที่ผู้เรียนจะเห็น: หากทำถูก จะแสดง "✅ แทนค่าสูญหายครบถ้วนแล้ว"
#   หากยังไม่ถูก จะแสดงข้อความ AssertionError พร้อมคำแนะนำ
#   (ทำไม่ได้ ให้รันเซลล์ ✅ เฉลย ถัดไป)
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: วนเติมค่าว่างทีละคอลัมน์
# ----------------------------------------------------------------------------
# for ... in ... คือ "การวนซ้ำ (loop)": ทำคำสั่งที่ย่อหน้าเข้าไปด้านล่าง ซ้ำ 1 รอบต่อ 1 รายการใน list
#   - zero_as_missing : list รายชื่อ 5 คอลัมน์ที่สร้างไว้ในเซลล์ก่อนหน้า
#   - col : ตัวแปรที่เปลี่ยนค่าไปทุกรอบ รอบแรก col = "Glucose" รอบถัดไป = "BloodPressure" ...
#   - เครื่องหมาย : ท้ายบรรทัด และการย่อหน้า บอกว่าบรรทัดใดอยู่ภายใน loop
for col in zero_as_missing:
    # เติมค่าว่างของคอลัมน์ col แล้วเขียนผลลงคอลัมน์เดียวกันในตาราง df
    #   - df_with_missing[col] : เลือกคอลัมน์ชื่อ col จากสำเนาที่ยังมีค่าว่าง (ใช้วงเล็บเหลี่ยม [ ])
    #   - .fillna(...) : เมธอดของ pandas สำหรับ "เติมค่าว่าง (NaN)" ด้วยค่าที่ระบุในวงเล็บ
    #       ค่าที่ไม่ว่างจะคงเดิม เฉพาะช่องที่ว่างเท่านั้นที่ถูกเติม
    #   - ... : ช่องว่างที่ผู้เรียนต้องเติม — คือ "ค่าที่จะใช้เติม" ให้ดูคำใบ้ด้านบนสุดของเซลล์
    #   - df[col] = ... : นำผลลัพธ์ไปเขียนทับคอลัมน์ col ในตาราง df
    df[col] = df_with_missing[col].fillna(...)   # TODO

# ----------------------------------------------------------------------------
# ส่วนที่ 2: ตรวจสอบคำตอบอัตโนมัติ
#   assert คือการตรวจเงื่อนไข หากเป็นจริงจะผ่านไปเงียบ ๆ หากเป็นเท็จจะหยุดและแสดงข้อความหลังจุลภาค
# ----------------------------------------------------------------------------
# ตรวจข้อ 1: ทุกคอลัมน์ใน zero_as_missing ต้องยังเป็นชนิด "ตัวเลข"
#   (หากยังไม่ได้แก้ ... ค่าว่างจะถูกเติมด้วยสัญลักษณ์ ... ซึ่งไม่ใช่ตัวเลข การตรวจนี้จึงไม่ผ่าน)
#   - pd.api.types.is_numeric_dtype(df[c]) : ฟังก์ชันของ pandas ตอบ True หากคอลัมน์ df[c] เป็นตัวเลข
#   - ... for c in zero_as_missing : เขียน loop แบบย่อในบรรทัดเดียว (generator expression)
#       ตรวจทีละคอลัมน์ c ในรายชื่อ
#   - all(...) : ฟังก์ชันพื้นฐานของ Python ตอบ True เมื่อ "ทุกผล" เป็น True
#   - ข้อความหลังจุลภาค : ข้อความเตือนที่แสดงเมื่อเงื่อนไขไม่ผ่าน
assert all(pd.api.types.is_numeric_dtype(df[c]) for c in zero_as_missing), "โปรดแทน ... ด้วย df_with_missing[col].median()"
# ตรวจข้อ 2: ต้องไม่มีค่าว่างเหลือใน 5 คอลัมน์นี้เลย
#   - df[zero_as_missing].isnull() : ตาราง True/False บอกว่าช่องใดว่าง
#   - .sum() ครั้งแรก : นับจำนวนค่าว่างของแต่ละคอลัมน์
#   - .sum() ครั้งที่สอง : รวมจำนวนจากทุกคอลัมน์ให้เหลือตัวเลขเดียว
#   - == 0 : เปรียบเทียบว่า "เท่ากับ 0 หรือไม่" (== คือเครื่องหมายเปรียบเทียบ ต่างจาก = ที่ใช้เก็บค่า)
assert df[zero_as_missing].isnull().sum().sum() == 0, "ยังมีค่าสูญหายเหลืออยู่"
# หากผ่านการตรวจทั้ง 2 ข้อ โปรแกรมจะทำงานมาถึงบรรทัดนี้ และพิมพ์ข้อความยืนยันด้วย print(...)
print("✅ แทนค่าสูญหายครบถ้วนแล้ว")
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เฉลยของแบบฝึกหัดที่ 1: แทนค่าสูญหาย (imputation) ใน 5 คอลัมน์ด้วยค่ามัธยฐาน (median)
#   ของแต่ละคอลัมน์ ซึ่งเป็นค่ากลางที่ไม่ได้รับผลกระทบจากค่าผิดปกติ (outlier)
#   - บรรทัดแรกของเซลล์ (คำสั่งตั้งหัวข้อ title) ทำให้ Colab แสดงเซลล์นี้เป็นแบบฟอร์ม (ซ่อนโค้ด)
#     { display-mode: "form" } คือการตั้งค่าให้แสดงเฉพาะหัวข้อ ดับเบิลคลิกเพื่อดูโค้ด
#   สิ่งที่ผู้เรียนจะเห็น: ข้อความ "✅ แทนค่าสูญหายครบถ้วนแล้ว"
#   ผลลัพธ์: ตาราง df ไม่มีค่าว่างใน 5 คอลัมน์นี้แล้ว (ผู้ป่วยยังครบ 768 ราย)
# ============================================================================

# วนซ้ำ (for loop) ทีละคอลัมน์ในรายชื่อ zero_as_missing โดยตัวแปร col เก็บชื่อคอลัมน์ของรอบนั้น
for col in zero_as_missing:
    # เติมค่าว่างของคอลัมน์ col ด้วยค่ามัธยฐานของคอลัมน์นั้น แล้วเขียนลงตาราง df
    #   - df_with_missing[col] : คอลัมน์ col จากสำเนาที่ยังมีค่าว่าง
    #   - .median() : เมธอดของ pandas คำนวณค่ามัธยฐาน (ค่ากลางเมื่อเรียงข้อมูล)
    #                 โดยข้ามค่าว่างโดยอัตโนมัติ เช่น Insulin ได้ประมาณ 125 µU/mL
    #   - .fillna(ค่ามัธยฐาน) : เติมเฉพาะช่องที่ว่าง (NaN) ด้วยค่ามัธยฐาน ช่องอื่นคงเดิม
    #   - df[col] = ... : เขียนผลลัพธ์ทับคอลัมน์ col ในตาราง df
    df[col] = df_with_missing[col].fillna(df_with_missing[col].median())

# ตรวจสอบด้วย assert ว่าไม่มีค่าว่างเหลือใน 5 คอลัมน์ หากยังเหลือ จะหยุดและแสดงข้อความเตือน
#   - .isnull()      : ตาราง True/False บอกว่าช่องใดว่าง
#   - .sum().sum()   : รวมจำนวนค่าว่างรายคอลัมน์ แล้วรวมทุกคอลัมน์ให้เป็นตัวเลขเดียว
#   - == 0           : ตรวจว่าเท่ากับศูนย์หรือไม่
#   - "ยังมีค่าสูญหายเหลืออยู่" : ข้อความเตือนเมื่อไม่ผ่าน
assert df[zero_as_missing].isnull().sum().sum() == 0, "ยังมีค่าสูญหายเหลืออยู่"
# พิมพ์ข้อความยืนยันเมื่อผ่านการตรวจสอบ
print("✅ แทนค่าสูญหายครบถ้วนแล้ว")
```

### การเปรียบเทียบผู้ป่วยที่เป็นและไม่เป็นโรคเบาหวาน

กราฟแต่ละรูปด้านล่างคือ *ฮิสโทแกรม (histogram)* ความสูงของแท่งแสดงจำนวนผู้ป่วยที่มีค่าอยู่ในช่วงนั้น สีน้ำเงิน = ไม่เป็นโรคเบาหวาน (Outcome 0) สีส้ม = เป็นโรคเบาหวาน (Outcome 1) ตัวแปรที่การกระจายของสองสีแยกจากกันชัดเจน คือตัวแปรที่ช่วยจำแนกสองกลุ่มได้ดี

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   วาดฮิสโทแกรม (histogram) ของตัวแปรทั้ง 8 ตัว โดยแยกสีตามผลลัพธ์ Outcome
#   (สีน้ำเงิน = ไม่เป็นโรคเบาหวาน (0), สีส้ม = เป็นโรคเบาหวาน (1)) ซ้อนกันในกราฟเดียว
#   - จัดเรียงกราฟย่อย 8 รูปเป็นตาราง 2 แถว x 4 คอลัมน์
#   - ตัวแปรที่การกระจายของสองสีแยกจากกันชัดเจน คือตัวแปรที่ช่วยจำแนกผู้ป่วยสองกลุ่มได้ดี
#   สิ่งที่ผู้เรียนจะเห็น: ภาพกราฟ 8 รูป ใช้ประกอบการตอบคำถาม B3
#   และจะเห็นแท่งสูงผิดปกติใน Insulin/SkinThickness ซึ่งเกิดจากการแทนค่าด้วยมัธยฐาน
# ============================================================================

# การกระจายของแต่ละตัวแปร จำแนกตามผลลัพธ์ (0 = ไม่เป็นโรคเบาหวาน, 1 = เป็นโรคเบาหวาน)
# สร้าง list ชื่อ features เก็บชื่อคอลัมน์ตัวแปรทั้ง 8 ตัวที่จะวาดกราฟ (ไม่รวม Outcome)
#   - list นี้เขียนยาวต่อ 2 บรรทัดได้ เพราะ Python อนุญาตให้ขึ้นบรรทัดใหม่ภายในวงเล็บ [ ]
features = ["Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
            "Insulin", "BMI", "DiabetesPedigreeFunction", "Age"]
# สร้างพื้นที่วาดภาพ (figure) ที่แบ่งเป็นกราฟย่อยหลายช่อง ด้วย plt.subplots(...) ของ matplotlib
#   - 2 : จำนวนแถวของกราฟย่อย
#   - 4 : จำนวนคอลัมน์ของกราฟย่อย (รวม 2 x 4 = 8 ช่อง พอดีกับ 8 ตัวแปร)
#   - figsize=(16, 7) : ขนาดภาพทั้งหมด กว้าง 16 x สูง 7 นิ้ว
#   - ผลลัพธ์มี 2 ค่า จึงรับด้วยตัวแปร 2 ตัวคั่นด้วยจุลภาค
#       fig  = ภาพทั้งภาพ, axes = ตารางของกราฟย่อย (แต่ละช่องเรียกว่า ax/axes)
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
# วนซ้ำเพื่อจับคู่ "กราฟย่อย 1 ช่อง" กับ "ตัวแปร 1 ตัว" ไปพร้อมกัน
#   - axes.flat : เรียงกราฟย่อย 2 x 4 ให้เป็นแถวเดียว 8 ช่อง (ซ้ายไปขวา บนลงล่าง)
#   - zip(a, b) : ฟังก์ชันพื้นฐานของ Python จับคู่สมาชิกลำดับเดียวกันของ a และ b
#   - ax, col   : ตัวแปร 2 ตัวที่รับคู่ในแต่ละรอบ (ax = กราฟย่อย, col = ชื่อตัวแปร)
for ax, col in zip(axes.flat, features):
    # วนซ้ำซ้อนอีกชั้น (nested loop): แบ่งผู้ป่วยตามค่า Outcome แล้ววาดทีละกลุ่ม
    #   - df.groupby("Outcome") : แบ่งตาราง df เป็นกลุ่มตามค่าในคอลัมน์ Outcome (กลุ่ม 0 และกลุ่ม 1)
    #   - outcome : ค่าของกลุ่มในรอบนั้น (0 หรือ 1)
    #   - grp     : ตารางย่อยของผู้ป่วยในกลุ่มนั้น
    for outcome, grp in df.groupby("Outcome"):
        # วาดฮิสโทแกรมของตัวแปร col สำหรับผู้ป่วยกลุ่มนี้ ลงในกราฟย่อย ax ด้วยเมธอด .hist(...)
        #   - grp[col] : ค่าของตัวแปร col ของผู้ป่วยในกลุ่มนี้
        #   - alpha=0.6 : ความทึบของสี 0 (โปร่งใส) ถึง 1 (ทึบ) ใช้ 0.6 เพื่อให้เห็นแท่งที่ซ้อนกัน
        #   - bins=25 : แบ่งช่วงค่าเป็น 25 ช่วง (จำนวนแท่ง)
        #   - label=f"Outcome={outcome}" : ชื่อที่ใช้ในคำอธิบายสัญลักษณ์ (legend)
        #       f-string แทรกค่า outcome ลงในข้อความ เช่น "Outcome=0"
        ax.hist(grp[col], alpha=0.6, bins=25, label=f"Outcome={outcome}")
    # ตั้งชื่อกราฟย่อยเป็นชื่อตัวแปร col ด้วยเมธอด .set_title(...) (บรรทัดนี้อยู่ใน loop ชั้นนอก)
    ax.set_title(col)
# แสดงคำอธิบายสัญลักษณ์ (legend) บอกว่าสีใดคือ Outcome=0/1 เฉพาะในกราฟย่อยช่องแรก
#   - axes.flat[0] : เลือกกราฟย่อยลำดับที่ 0 (Python เริ่มนับจาก 0 ดังนั้น 0 = ช่องแรก)
#   - .legend()    : เมธอดของ matplotlib แสดงกล่องคำอธิบายสัญลักษณ์ (ใช้ค่า label ที่ตั้งไว้)
axes.flat[0].legend()
# จัดระยะห่างระหว่างกราฟย่อยอัตโนมัติด้วย plt.tight_layout() เพื่อไม่ให้ชื่อกราฟหรือแกนซ้อนทับกัน
plt.tight_layout()
# แสดงภาพกราฟทั้งหมดบนหน้าจอ
plt.show()
```

💡 **คำอธิบายผลลัพธ์** แท่งที่สูงผิดปกติใน `Insulin` และ `SkinThickness` คือค่ามัธยฐานที่เพิ่งแทนเข้าไป ซึ่งเป็นผลข้างเคียงจากการแทนค่า (imputation artefact)

*แผนภาพความร้อนแสดงสหสัมพันธ์ (correlation heatmap)* ด้านล่างแสดงระดับความสัมพันธ์ระหว่างสองคอลัมน์: +1 = สัมพันธ์กันในทิศทางเดียวกันอย่างสมบูรณ์, 0 = ไม่สัมพันธ์กัน, −1 = สัมพันธ์กันในทิศทางตรงข้าม ให้พิจารณาที่แถว `Outcome`

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   วาดแผนภาพความร้อนแสดงสหสัมพันธ์ (correlation heatmap) ระหว่างทุกคู่ของคอลัมน์
#   - สหสัมพันธ์ (correlation) มีค่า -1 ถึง +1: +1 = สัมพันธ์ไปทางเดียวกันอย่างสมบูรณ์,
#     0 = ไม่สัมพันธ์กัน (เชิงเส้น), -1 = สัมพันธ์ไปทางตรงข้ามอย่างสมบูรณ์
#   - สีแดงเข้ม = สัมพันธ์บวกสูง, สีน้ำเงินเข้ม = สัมพันธ์ลบสูง, สีขาว = ใกล้ 0
#   สิ่งที่ผู้เรียนจะเห็น: ตารางสีพร้อมตัวเลขทศนิยม 2 ตำแหน่ง ให้พิจารณาแถว Outcome
#   ว่าตัวแปรใดสัมพันธ์กับการเป็นโรคเบาหวานมากที่สุด (ใช้ตอบคำถาม B3)
# ============================================================================

# แผนภาพความร้อนแสดงสหสัมพันธ์ (correlation heatmap)
# สร้างพื้นที่วาดภาพใหม่ด้วย plt.figure(...) ของ matplotlib
#   - figsize=(8, 6) : ขนาดภาพ กว้าง 8 x สูง 6 นิ้ว
plt.figure(figsize=(8, 6))
# คำนวณสหสัมพันธ์ แล้ววาดเป็นแผนภาพความร้อนด้วย sns.heatmap(...) ของไลบรารี seaborn
#   - df.corr(numeric_only=True) : เมธอดของ pandas คำนวณค่าสหสัมพันธ์ของเพียร์สัน (Pearson)
#       ระหว่างทุกคู่คอลัมน์ ได้ตารางสี่เหลี่ยมจัตุรัส
#       numeric_only=True : ใช้เฉพาะคอลัมน์ที่เป็นตัวเลข (ข้ามคอลัมน์ข้อความ/กลุ่ม)
#   - annot=True    : เขียนตัวเลขค่าสหสัมพันธ์ลงในแต่ละช่อง (annotation)
#   - fmt=".2f"     : รูปแบบตัวเลข — ทศนิยม (f = float) 2 ตำแหน่ง เช่น 0.47
#   - cmap="RdBu_r" : ชุดสี (colormap) แดง-น้ำเงิน (Red-Blue) แบบกลับด้าน (_r = reversed)
#                     จึงให้ค่าบวกเป็นสีแดง และค่าลบเป็นสีน้ำเงิน
#   - center=0      : กำหนดให้ค่า 0 อยู่ตรงกลางของชุดสี (สีขาว)
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="RdBu_r", center=0)
# ตั้งชื่อภาพด้วย plt.title(...) — "Feature correlations" = สหสัมพันธ์ระหว่างตัวแปร
plt.title("Feature correlations")
# แสดงภาพบนหน้าจอ
plt.show()
```

```python
# ⭐ Exercise 2 — แบบฝึกหัดที่ 2: สถิติจำแนกตามกลุ่ม
# วัตถุประสงค์: เปรียบเทียบผู้ป่วยที่เป็นและไม่เป็นโรคเบาหวาน และหาอัตราการเป็นโรคเบาหวานตามกลุ่มอายุ
# สิ่งที่ต้องแก้ไข: เติม ... ทั้ง 3 ตำแหน่ง (ตำแหน่งละ 1 คำหรือ 1 รายการ)
#  a) ค่าสถิติที่ต้องการคำนวณ                   → คำใบ้: "mean"
#  b) จุดแบ่งช่วงอายุ 0, 30, 50 และ 120          → คำใบ้: [0, 30, 50, 120]
#  c) คอลัมน์ที่ค่าเฉลี่ยเท่ากับอัตราการเป็นโรคเบาหวาน
#     (Outcome มีค่า 0/1 ค่าเฉลี่ยจึงเท่ากับสัดส่วนผู้ที่เป็นโรคเบาหวาน)   → คำใบ้: "Outcome"

# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   ฝึกคำนวณ "สถิติจำแนกตามกลุ่ม (group statistics)" ด้วย groupby ของ pandas
#   (คล้าย pivot table ใน Excel) โดยทำ 3 ขั้นตอน
#   1) เปรียบเทียบค่าสถิติของ Glucose และ BMI ระหว่างกลุ่ม Outcome 0 และ 1  → group_stats
#   2) สร้างคอลัมน์ใหม่ AgeGroup แบ่งอายุเป็น 3 ช่วง: "<30", "30-50", ">50"
#   3) คำนวณอัตราการเป็นโรคเบาหวานในแต่ละช่วงอายุ                         → rate_by_age
#   ผู้เรียนต้องเติมเครื่องหมาย ... ทั้ง 3 ตำแหน่ง (TODO a, b, c) ตามคำใบ้ด้านบน
#   สิ่งที่ผู้เรียนจะเห็น: ตาราง group_stats และรายการ rate_by_age (ใช้ตอบคำถาม B3)
# ============================================================================

# (a) จัดกลุ่มผู้ป่วยตาม Outcome แล้วคำนวณค่าสถิติของ Glucose และ BMI เก็บไว้ใน group_stats
#   ทำงานต่อกันเป็นทอดจากซ้ายไปขวา
#   1) df.groupby("Outcome")       : แบ่งผู้ป่วยเป็นกลุ่มตามค่าในคอลัมน์ Outcome (0 และ 1)
#   2) [["Glucose", "BMI"]]         : เลือกเฉพาะ 2 คอลัมน์นี้มาคำนวณ
#                                     (วงเล็บเหลี่ยม 2 ชั้น = ส่ง list ของชื่อคอลัมน์ ได้ผลเป็นตาราง)
#   3) .agg(...)                    : เมธอด aggregate คำนวณค่าสถิติสรุปของแต่ละกลุ่ม
#                                     ตามชื่อสถิติที่ระบุในวงเล็บ
#   - ... : ช่องว่างที่ผู้เรียนต้องเติม (a) — ชื่อค่าสถิติที่ต้องการ (ดูคำใบ้ด้านบน)
group_stats = df.groupby("Outcome")[["Glucose", "BMI"]].agg(...)                  # TODO (a)
# (b) สร้างคอลัมน์ใหม่ชื่อ AgeGroup ในตาราง df โดยแบ่งอายุ (Age) เป็นช่วง ๆ ด้วย pd.cut(...) ของ pandas
#   - df["AgeGroup"] = ... : หากยังไม่มีคอลัมน์ชื่อนี้ pandas จะสร้างคอลัมน์ใหม่ให้
#   - df["Age"]  : ข้อมูลที่จะนำมาแบ่งช่วง (อายุของผู้ป่วยแต่ละราย)
#   - bins=...   : "จุดแบ่งช่วง" เป็น list ของตัวเลข — ช่องว่างที่ผู้เรียนต้องเติม (b) ดูคำใบ้ด้านบน
#                  pd.cut ใช้ช่วงแบบ "ไม่รวมขอบซ้าย แต่รวมขอบขวา" เช่น (30, 50] = มากกว่า 30 ถึง 50
#   - labels=["<30", "30-50", ">50"] : ชื่อที่ตั้งให้แต่ละช่วงตามลำดับ (จำนวนชื่อต้องน้อยกว่าจุดแบ่ง 1 ตัว)
#   - ผลลัพธ์เป็นข้อมูลชนิดกลุ่ม (categorical) เช่น ผู้ป่วยอายุ 45 ปี จะได้ "30-50"
df["AgeGroup"] = pd.cut(df["Age"], bins=..., labels=["<30", "30-50", ">50"])      # TODO (b)
# (c) คำนวณค่าเฉลี่ยของคอลัมน์หนึ่งในแต่ละกลุ่มอายุ เก็บไว้ใน rate_by_age
#   - df.groupby("AgeGroup", observed=True) : แบ่งผู้ป่วยตามกลุ่มอายุ
#       observed=True : แสดงเฉพาะกลุ่มที่มีข้อมูลจริง (ใช้กับคอลัมน์ชนิด categorical
#                       และช่วยไม่ให้ pandas แสดงคำเตือน)
#   - [...]   : เลือกคอลัมน์ที่จะนำมาเฉลี่ย — ช่องว่างที่ผู้เรียนต้องเติม (c) ดูคำใบ้ด้านบน
#   - .mean() : เมธอดคำนวณค่าเฉลี่ย (mean) ของแต่ละกลุ่ม
rate_by_age = df.groupby("AgeGroup", observed=True)[...].mean()                   # TODO (c)

# แสดงตาราง group_stats บนหน้าจอด้วย print(...)
print(group_stats)
# แสดงผล rate_by_age บนหน้าจอ (ค่า 0.20 = ร้อยละ 20)
print(rate_by_age)
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เฉลยของแบบฝึกหัดที่ 2: สถิติจำแนกตามกลุ่ม (group statistics)
#   1) group_stats : ค่าเฉลี่ย (mean) ของ Glucose และ BMI ในผู้ป่วยกลุ่ม Outcome 0 และ 1
#   2) AgeGroup    : คอลัมน์ใหม่ แบ่งอายุเป็น "<30" (ไม่เกิน 30 ปี), "30-50" (31–50 ปี), ">50" (มากกว่า 50 ปี)
#   3) rate_by_age : อัตราการเป็นโรคเบาหวานในแต่ละกลุ่มอายุ
#      (Outcome มีค่า 0/1 ค่าเฉลี่ยจึงเท่ากับสัดส่วนผู้ที่เป็นโรค เช่น 0.20 = ร้อยละ 20)
#   สิ่งที่ผู้เรียนจะเห็น: ตาราง group_stats และรายการ rate_by_age
# ============================================================================

# จัดกลุ่มผู้ป่วยตาม Outcome แล้วคำนวณค่าเฉลี่ยของ Glucose และ BMI ในแต่ละกลุ่ม
#   - df.groupby("Outcome")  : แบ่งผู้ป่วยเป็นกลุ่ม 0 (ไม่เป็นโรค) และ 1 (เป็นโรค)
#   - [["Glucose", "BMI"]]    : เลือก 2 คอลัมน์ที่จะคำนวณ (list ของชื่อคอลัมน์)
#   - .agg("mean")            : คำนวณสถิติสรุปตามชื่อที่ระบุ "mean" = ค่าเฉลี่ย
group_stats = df.groupby("Outcome")[["Glucose", "BMI"]].agg("mean")
# สร้างคอลัมน์ใหม่ AgeGroup โดยแบ่งอายุเป็นช่วงด้วย pd.cut(...) ของ pandas
#   - df["Age"]               : ข้อมูลอายุที่จะนำมาแบ่ง
#   - bins=[0, 30, 50, 120]   : จุดแบ่งช่วง ได้ 3 ช่วง คือ (0, 30], (30, 50], (50, 120]
#                               (ไม่รวมขอบซ้าย แต่รวมขอบขวา; 120 เป็นขอบบนที่สูงกว่าอายุทุกราย)
#   - labels=["<30", "30-50", ">50"] : ชื่อของแต่ละช่วงตามลำดับ
df["AgeGroup"] = pd.cut(df["Age"], bins=[0, 30, 50, 120], labels=["<30", "30-50", ">50"])
# คำนวณอัตราการเป็นโรคเบาหวานในแต่ละกลุ่มอายุ
#   - df.groupby("AgeGroup", observed=True) : แบ่งผู้ป่วยตามกลุ่มอายุ
#       observed=True : แสดงเฉพาะกลุ่มที่มีข้อมูลจริง (สำหรับคอลัมน์ชนิด categorical)
#   - ["Outcome"] : เลือกคอลัมน์ Outcome (ค่า 0/1)
#   - .mean()     : ค่าเฉลี่ยของ 0/1 = สัดส่วนผู้ที่เป็นโรคเบาหวานในกลุ่มนั้น
rate_by_age = df.groupby("AgeGroup", observed=True)["Outcome"].mean()

# แสดงตาราง group_stats บนหน้าจอ
print(group_stats)
# แสดงอัตราการเป็นโรคเบาหวานตามกลุ่มอายุบนหน้าจอ
print(rate_by_age)
```

💡 **คำอธิบายผลลัพธ์** `group_stats` แสดงค่าเฉลี่ยของระดับน้ำตาลที่ 2 ชั่วโมงและ BMI ในแต่ละกลุ่มผลลัพธ์ ส่วน `rate_by_age` แสดงสัดส่วนของสตรีที่เป็นโรคเบาหวานในแต่ละกลุ่มอายุ (0.20 = ร้อยละ 20) (กลุ่มอายุ: "<30" = ไม่เกิน 30 ปี, "30-50" = 31–50 ปี, ">50" = มากกว่า 50 ปี)

### รายงานสรุปลักษณะข้อมูลฉบับย่อ (mini profiling report)

*Data profile* คือตารางที่สรุปลักษณะของทุกคอลัมน์ไว้ในตารางเดียว ซึ่งควรจัดทำเป็นขั้นตอนแรกเมื่อได้รับชุดข้อมูลทางคลินิกชุดใหม่ ในที่นี้จัดทำ profile จาก `df_with_missing` (ข้อมูล *ก่อน* การแทนค่า) เพื่อให้ยังคงเห็นค่าสูญหาย

```python
# ⭐ Exercise 3 — แบบฝึกหัดที่ 3: รายงานสรุปลักษณะข้อมูลฉบับย่อ
# วัตถุประสงค์: สร้างตารางที่มี 1 แถวต่อ 1 คอลัมน์ ได้แก่ ชนิดข้อมูล จำนวนและร้อยละของค่าสูญหาย
#               จำนวนค่าที่ไม่ซ้ำกัน ค่าต่ำสุด และค่าสูงสุด
# สิ่งที่ต้องแก้ไข: เติม ... ทั้ง 2 ตำแหน่ง โดยใช้คำสั่ง pandas ต่อไปนี้
#     data[col].isnull().sum()   → นับจำนวนค่าสูญหายในคอลัมน์
#     data[col].nunique()        → นับจำนวนค่าที่ไม่ซ้ำกันในคอลัมน์

# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   สร้าง "รายงานสรุปลักษณะข้อมูลฉบับย่อ (mini data profiling report)" ซึ่งควรทำเป็นขั้นตอนแรก
#   เมื่อได้รับชุดข้อมูลทางคลินิกชุดใหม่
#   - สร้างฟังก์ชัน profile(data) ที่รับตารางเข้าไป แล้วคืนตารางสรุป 1 แถวต่อ 1 คอลัมน์ของข้อมูล
#     ประกอบด้วย: ชื่อคอลัมน์, ชนิดข้อมูล, จำนวน/ร้อยละค่าสูญหาย, จำนวนค่าที่ไม่ซ้ำกัน, ค่าต่ำสุด, ค่าสูงสุด
#   - เรียกใช้กับ df_with_missing (ข้อมูล "ก่อน" การแทนค่า) เพื่อให้ยังเห็นค่าสูญหาย
#   ผู้เรียนต้องเติมเครื่องหมาย ... ทั้ง 2 ตำแหน่ง (TODO) โดยใช้คำสั่งที่ให้ไว้ด้านบน
#   สิ่งที่ผู้เรียนจะเห็น: ตาราง report (ใช้เป็นหลักฐานในการตอบคำถาม B4)
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: สร้างฟังก์ชัน profile
#   - def คือคำสั่งสร้างฟังก์ชัน (ชุดคำสั่งที่ตั้งชื่อไว้และเรียกใช้ซ้ำได้)
#   - data : พารามิเตอร์ = ตาราง (DataFrame) ที่ผู้ใช้ส่งเข้ามาตอนเรียกฟังก์ชัน
#   - คำสั่งที่ย่อหน้าเข้าไปทั้งหมดด้านล่าง คือเนื้อหาภายในฟังก์ชัน
# ----------------------------------------------------------------------------
def profile(data):
    # สร้าง list ว่าง [] ชื่อ rows ไว้สะสมข้อมูลสรุปของแต่ละคอลัมน์ (คอลัมน์ละ 1 รายการ)
    rows = []
    # วนซ้ำทีละคอลัมน์ของตาราง
    #   - data.columns : รายชื่อคอลัมน์ทั้งหมดของตาราง data
    #   - col : ชื่อคอลัมน์ในรอบนั้น
    for col in data.columns:
        # นับจำนวนค่าสูญหาย (ค่าว่าง) ของคอลัมน์ col แล้วเก็บไว้ในตัวแปร n_missing
        #   - ... : ช่องว่างที่ผู้เรียนต้องเติม — ใช้คำสั่งนับค่าสูญหายตามที่ให้ไว้ด้านบนของเซลล์
        n_missing = ...                                  # TODO: จำนวนค่าสูญหาย
        # เพิ่มข้อมูลสรุปของคอลัมน์นี้ 1 รายการต่อท้าย list rows ด้วยเมธอด .append(...)
        #   - สิ่งที่ส่งเข้าไปคือ dict (พจนานุกรม) เขียนในวงเล็บปีกกา { }
        #     ประกอบด้วยคู่ "ชื่อ": ค่า คั่นด้วยจุลภาค — ชื่อแต่ละตัวจะกลายเป็นหัวคอลัมน์ของตารางสรุป
        rows.append({
            # ชื่อคอลัมน์ของข้อมูล
            "column": col,
            # ชนิดข้อมูลของคอลัมน์ (เช่น float64, int64)
            #   - data[col].dtype : คุณสมบัติบอกชนิดข้อมูลของคอลัมน์
            #   - str(...) : ฟังก์ชันพื้นฐานของ Python แปลงค่าให้เป็นข้อความ
            "dtype": str(data[col].dtype),
            # จำนวนค่าสูญหายที่นับไว้ในตัวแปร n_missing
            "n_missing": n_missing,
            # ร้อยละของค่าสูญหาย = 100 x จำนวนค่าสูญหาย / จำนวนแถวทั้งหมด
            #   - len(data) : ฟังก์ชันพื้นฐานของ Python นับจำนวนแถวของตาราง
            #   - round(ค่า, 1) : ปัดเศษให้เหลือทศนิยม 1 ตำแหน่ง
            "pct_missing": round(100 * n_missing / len(data), 1),
            # จำนวนค่าที่ไม่ซ้ำกันในคอลัมน์ (ค่าน้อย เช่น 2 มักบ่งบอกว่าเป็นข้อมูลเชิงกลุ่ม (categorical))
            #   - ... : ช่องว่างที่ผู้เรียนต้องเติม — ใช้คำสั่งนับค่าที่ไม่ซ้ำกันตามที่ให้ไว้ด้านบนของเซลล์
            "n_unique": ...,                             # TODO: จำนวนค่าที่ไม่ซ้ำกัน
            # ค่าต่ำสุดของคอลัมน์ โดยใช้ "if แบบบรรทัดเดียว" (ค่า_ก if เงื่อนไข else ค่า_ข)
            #   - pd.api.types.is_numeric_dtype(data[col]) : ตรวจว่าคอลัมน์เป็นตัวเลขหรือไม่
            #   - ถ้าเป็นตัวเลข  → ใช้ data[col].min() (ค่าต่ำสุด ข้ามค่าว่างโดยอัตโนมัติ)
            #   - ถ้าไม่ใช่ตัวเลข → ใช้ np.nan (ค่าว่าง) เพราะค่าต่ำสุดไม่มีความหมาย
            "min": data[col].min() if pd.api.types.is_numeric_dtype(data[col]) else np.nan,
            # ค่าสูงสุดของคอลัมน์ หลักการเดียวกับบรรทัดก่อน แต่ใช้ .max() (ค่าสูงสุด)
            "max": data[col].max() if pd.api.types.is_numeric_dtype(data[col]) else np.nan,
        })
    # เมื่อวนครบทุกคอลัมน์แล้ว แปลง list ของ dict ให้เป็นตาราง ด้วย pd.DataFrame(rows)
    # แล้วส่งตารางนั้นกลับไปเป็นผลลัพธ์ของฟังก์ชันด้วยคำสั่ง return
    #   (return อยู่นอก loop เพราะย่อหน้าตรงกับ for จึงทำงานครั้งเดียวหลังวนครบ)
    return pd.DataFrame(rows)

# ----------------------------------------------------------------------------
# ส่วนที่ 2: เรียกใช้ฟังก์ชันกับข้อมูลก่อนการแทนค่า และแสดงผล
# ----------------------------------------------------------------------------
# เรียกฟังก์ชัน profile โดยส่ง df_with_missing (ข้อมูลที่ยังมีค่าว่าง) เข้าไป แล้วเก็บตารางผลลัพธ์ไว้ใน report
report = profile(df_with_missing)
# บรรทัดสุดท้ายของเซลล์: Jupyter จะแสดงตาราง report บนหน้าจอโดยอัตโนมัติ
report
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เฉลยของแบบฝึกหัดที่ 3: สร้างรายงานสรุปลักษณะข้อมูลฉบับย่อ (mini data profiling report)
#   - ฟังก์ชัน profile(data) สรุปทุกคอลัมน์ของตาราง: ชนิดข้อมูล จำนวน/ร้อยละค่าสูญหาย
#     จำนวนค่าที่ไม่ซ้ำกัน ค่าต่ำสุด และค่าสูงสุด (1 แถวต่อ 1 คอลัมน์)
#   - เรียกใช้กับ df_with_missing (ข้อมูลก่อนการแทนค่า) เพื่อให้เห็นค่าสูญหายจริง
#   สิ่งที่ผู้เรียนจะเห็น: ตาราง report เช่น Insulin มีค่าสูญหายเกือบครึ่งหนึ่ง
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: สร้างฟังก์ชัน profile (def = สร้างฟังก์ชัน, data = ตารางที่ส่งเข้ามา)
# ----------------------------------------------------------------------------
def profile(data):
    # สร้าง list ว่างชื่อ rows ไว้สะสมข้อมูลสรุปของแต่ละคอลัมน์
    rows = []
    # วนซ้ำทีละคอลัมน์ (col) ในรายชื่อคอลัมน์ทั้งหมด data.columns
    for col in data.columns:
        # นับจำนวนค่าสูญหายของคอลัมน์ col
        #   - data[col].isnull() : True ในช่องที่ว่าง, False ในช่องที่มีค่า
        #   - .sum()             : รวม True (นับเป็น 1) ได้จำนวนค่าว่างทั้งหมดของคอลัมน์
        n_missing = data[col].isnull().sum()
        # เพิ่มข้อมูลสรุปของคอลัมน์นี้ในรูป dict { "ชื่อ": ค่า, ... } ต่อท้าย list rows ด้วย .append(...)
        rows.append({
            # ชื่อคอลัมน์
            "column": col,
            # ชนิดข้อมูล (dtype) แปลงเป็นข้อความด้วย str(...)
            "dtype": str(data[col].dtype),
            # จำนวนค่าสูญหาย
            "n_missing": n_missing,
            # ร้อยละค่าสูญหาย = 100 x n_missing / จำนวนแถว (len(data)) ปัดเป็นทศนิยม 1 ตำแหน่งด้วย round(..., 1)
            "pct_missing": round(100 * n_missing / len(data), 1),
            # จำนวนค่าที่ไม่ซ้ำกันด้วยเมธอด .nunique() ของ pandas (number of unique values; ไม่นับค่าว่าง)
            "n_unique": data[col].nunique(),
            # ค่าต่ำสุด หากคอลัมน์เป็นตัวเลข (pd.api.types.is_numeric_dtype) มิฉะนั้นใช้ค่าว่าง np.nan
            #   (รูปแบบ ค่า_ก if เงื่อนไข else ค่า_ข คือ if แบบบรรทัดเดียว)
            "min": data[col].min() if pd.api.types.is_numeric_dtype(data[col]) else np.nan,
            # ค่าสูงสุด ด้วยหลักการเดียวกัน แต่ใช้ .max()
            "max": data[col].max() if pd.api.types.is_numeric_dtype(data[col]) else np.nan,
        })
    # หลังวนครบทุกคอลัมน์ แปลง list ของ dict เป็นตารางด้วย pd.DataFrame(rows) แล้วส่งกลับ (return)
    return pd.DataFrame(rows)

# ----------------------------------------------------------------------------
# ส่วนที่ 2: เรียกใช้ฟังก์ชันและแสดงผล
# ----------------------------------------------------------------------------
# เรียกฟังก์ชัน profile กับ df_with_missing แล้วเก็บตารางผลลัพธ์ไว้ใน report
report = profile(df_with_missing)
# บรรทัดสุดท้ายของเซลล์: แสดงตาราง report บนหน้าจอโดยอัตโนมัติ
report
```

## 🏆 โจทย์ท้าทายบนตารางอันดับ (Leaderboard) — ไม่บังคับ สำหรับผู้เรียนที่เขียนโปรแกรมได้

> **ไม่บังคับ — ไม่เป็นส่วนหนึ่งของชิ้นงานที่ต้องส่ง** ผู้เรียนทุกคนสามารถส่งผลได้ เนื่องจากกฎพื้นฐาน (baseline) ด้านล่างรันได้ทันที (โปรดระบุ `NAME` และ `JOIN_CODE` ในเซลล์การตั้งค่าและรันเซลล์นั้นใหม่ก่อน) ผู้เรียนที่เขียนโปรแกรมได้สามารถปรับปรุงให้ได้ผลดีกว่า baseline

**โจทย์: เครื่องมือคัดกรองแบบใช้กฎ (rule-based screener) โดยยังไม่ใช้การเรียนรู้ของเครื่อง (machine learning)** ก่อนการใช้การเรียนรู้ของเครื่อง แพทย์ใช้ **เกณฑ์ (rules)** ในการคัดกรองและวินิจฉัย ตัวอย่างเช่น เกณฑ์ของ ADA/WHO สำหรับการทดสอบ OGTT ที่ 2 ชั่วโมง ได้แก่ ระดับน้ำตาลในพลาสมาที่ 2 ชั่วโมง **≥ 200 mg/dL = โรคเบาหวาน** และ **140–199 mg/dL = ภาวะความทนต่อกลูโคสบกพร่อง (impaired glucose tolerance)** หรือภาวะก่อนเบาหวาน (prediabetes) คอลัมน์ `Glucose` ในชุดข้อมูลนี้ *คือ* ค่า OGTT ที่ 2 ชั่วโมงดังกล่าว ส่วนเกณฑ์ "≥ 126 mg/dL" ใช้กับ *fasting glucose* ซึ่งไม่มีในชุดข้อมูลนี้ ให้เขียนกฎทำนายโรคเบาหวานสำหรับ **ผู้ป่วยชุดทดสอบ 154 ราย** ซึ่งผลลัพธ์ถูกซ่อนไว้บนตารางอันดับ

**ตัวชี้วัด: balanced accuracy** = (ความไว (sensitivity) + ความจำเพาะ (specificity)) / 2 ดังนั้น หากทำนายว่าผู้ป่วยทุกราย "ไม่เป็นโรคเบาหวาน" จะได้คะแนน 0.5 พอดี ใน Lab 2 ผู้เรียนจะได้เปรียบเทียบว่าโมเดลที่เรียนรู้จากข้อมูลให้ผลดีกว่ากฎที่เขียนขึ้นเพียงใด

**การปรับกฎต้องใช้เฉพาะข้อมูลผู้ป่วยอีก 614 รายเท่านั้น** ฮิสโทแกรมในส่วน B ใช้ข้อมูลผู้ป่วยทั้ง 768 ราย *ซึ่งรวมผู้ป่วยชุดทดสอบ 154 รายด้วย* หากเลือกเกณฑ์ตัดสิน (threshold) โดยอาศัยผลลัพธ์ของผู้ป่วยชุดทดสอบ คะแนนบนตารางอันดับจะสูงเกินจริง เปรียบเสมือนการดูเฉลยก่อนสอบ และกฎดังกล่าวจะให้ผลด้อยลงเมื่อใช้กับผู้ป่วยรายใหม่ ในงานปัญญาประดิษฐ์ทางคลินิก ข้อผิดพลาดนี้เรียกว่า *การรั่วไหลของข้อมูล (data leakage)* และเป็นสาเหตุที่พบบ่อยซึ่งทำให้โมเดลที่ได้รับการตีพิมพ์ไม่สามารถใช้งานได้จริง ดังนั้น ในส่วนนี้จึงแยก **ชุดข้อมูลฝึก (training set)** ซึ่งไม่รวมผู้ป่วยชุดทดสอบออกมา และใช้ชุดข้อมูลฝึกเท่านั้นในการสร้างฮิสโทแกรมและคำนวณคะแนน

> **หลักความซื่อสัตย์ทางวิชาการ (honour code)** ผลลัพธ์ที่ซ่อนไว้มาจากชุดข้อมูลสาธารณะ การสืบค้นผลลัพธ์ดังกล่าวจึงทำได้โดยง่ายแต่ไม่เกิดประโยชน์ต่อการเรียนรู้ ตารางอันดับมีไว้เพื่อสนับสนุนการเรียนรู้ของผู้เรียน ห้ามแก้ไขผลการทำนายด้วยมือโดยเด็ดขาด

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทาย — ไม่บังคับ)
#   เตรียมทุกอย่างสำหรับโจทย์ "เครื่องมือคัดกรองแบบใช้กฎ (rule-based screener)" โดยทำ 3 อย่าง
#   1) แบ่งผู้ป่วยเป็น 2 ชุด
#        - ชุดข้อมูลฝึก (training set) 614 ราย : มีผลลัพธ์ Outcome ใช้ปรับกฎได้
#        - ชุดทดสอบ (test set) 154 ราย         : ซ่อนผลลัพธ์ไว้บนตารางอันดับ ใช้วัดผลจริง
#      (การแยกชุดทดสอบไว้ต่างหาก ช่วยให้ประเมินได้ว่ากฎใช้ได้กับผู้ป่วยที่ไม่เคยเห็นหรือไม่)
#   2) สร้างฟังก์ชัน evaluate(rule) คำนวณ ความไว (sensitivity), ความจำเพาะ (specificity)
#      และ balanced accuracy = (ความไว + ความจำเพาะ) / 2 ของกฎ จากชุดข้อมูลฝึก
#   3) วาดฮิสโทแกรมของ Glucose, BMI และ Age จากชุดข้อมูลฝึกเท่านั้น เพื่อช่วยเลือกเกณฑ์ตัดสิน
#   สิ่งที่ผู้เรียนจะเห็น: จำนวนผู้ป่วยแต่ละชุด และกราฟ 3 รูป
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: แบ่งข้อมูลเป็นชุดข้อมูลฝึกและชุดทดสอบ
# ----------------------------------------------------------------------------
# แบ่งข้อมูล: ผู้ป่วยชุดทดสอบ (ผลลัพธ์ถูกซ่อนบนตารางอันดับ) และผู้ป่วยชุดข้อมูลฝึก (ใช้ปรับกฎ)
# นำเข้าฟังก์ชันวัดผล 2 ตัวจากโมดูล sklearn.metrics ของไลบรารี scikit-learn (ไลบรารีการเรียนรู้ของเครื่อง)
#   - from ... import ... : นำเข้าเฉพาะฟังก์ชันที่ต้องการจากไลบรารี
#   - balanced_accuracy_score : คำนวณ balanced accuracy (ค่าเฉลี่ยของความไวและความจำเพาะ)
#   - recall_score            : คำนวณ recall = สัดส่วนที่ทำนายถูกในกลุ่มที่ระบุ
#                               (recall ของกลุ่มเป็นโรค = ความไว, ของกลุ่มไม่เป็นโรค = ความจำเพาะ)
from sklearn.metrics import balanced_accuracy_score, recall_score

# อ่านรายการรหัสผู้ป่วยชุดทดสอบ แล้วเก็บไว้ใน test_ids
#   - lab_data("pima/test_ids.csv") : ฟังก์ชันจากเซลล์การตั้งค่า ดาวน์โหลดไฟล์รหัสผู้ป่วยชุดทดสอบ
#                                     แล้วคืนตำแหน่งไฟล์ในเครื่อง
#   - pd.read_csv(...) : อ่านไฟล์ CSV นั้นเป็นตาราง
#   - ["id"] : เลือกเฉพาะคอลัมน์ id (เลขลำดับแถวของผู้ป่วยในตาราง df) ได้เป็น Series รายการรหัส
test_ids = pd.read_csv(lab_data("pima/test_ids.csv"))["id"]
# สร้างชุดข้อมูลฝึก train โดยตัดผู้ป่วยชุดทดสอบออกจาก df
#   - df.drop(...) : เมธอดของ pandas ลบแถวหรือคอลัมน์ (ได้ตารางใหม่ ไม่แก้ df เดิม)
#   - index=test_ids : ลบ "แถว" ที่มีเลขแถว (index) อยู่ในรายการ test_ids
train = df.drop(index=test_ids)                            # ผู้ป่วย 614 ราย - ใช้ปรับกฎจากชุดนี้เท่านั้น
# สร้างชุดทดสอบ test โดยเลือกเฉพาะผู้ป่วยชุดทดสอบ แล้วตัดคอลัมน์ผลลัพธ์ออก (จำลองการซ่อนคำตอบ)
#   - df.loc[test_ids] : .loc[...] เลือกแถวตามเลขแถว (index) ที่อยู่ใน test_ids
#   - .drop(columns=["Outcome"]) : ลบ "คอลัมน์" Outcome ออก
test  = df.loc[test_ids].drop(columns=["Outcome"])         # ผู้ป่วยชุดทดสอบ 154 ราย (ผลลัพธ์ถูกซ่อน)
# พิมพ์จำนวนผู้ป่วยในแต่ละชุด
#   - f"...{...}..." : f-string แทรกค่าที่อยู่ใน { } ลงในข้อความ
#   - len(train) / len(test) : จำนวนแถว (ผู้ป่วย) ของแต่ละตาราง
print(f"ชุดข้อมูลฝึก: {len(train)} ราย | ชุดทดสอบ: {len(test)} ราย")

# ----------------------------------------------------------------------------
# ส่วนที่ 2: สร้างฟังก์ชัน evaluate สำหรับวัดผลกฎคัดกรองบนชุดข้อมูลฝึก
#   - rule : พารามิเตอร์ที่เป็น "ฟังก์ชันกฎ" ซึ่งรับข้อมูลผู้ป่วย 1 ราย แล้วคืน 1 (เป็นโรค) หรือ 0
#   - data=train : ตารางที่ใช้วัดผล หากไม่ระบุ จะใช้ train โดยอัตโนมัติ (ค่าเริ่มต้น)
# ----------------------------------------------------------------------------
def evaluate(rule, data=train):
    """แสดงความไว ความจำเพาะ และ balanced accuracy ของกฎ โดยคำนวณจากผู้ป่วยชุดข้อมูลฝึก"""
    # ใช้กฎกับผู้ป่วยทุกราย แล้วเก็บผลการทำนาย (0/1) ไว้ใน pred
    #   - data.apply(rule, axis=1) : เมธอดของ pandas เรียกฟังก์ชัน rule กับข้อมูลทีละส่วน
    #       - rule   : ฟังก์ชันที่จะเรียกใช้
    #       - axis=1 : ส่งข้อมูล "ทีละแถว" (ผู้ป่วยทีละราย) เข้าไปในฟังก์ชัน (axis=0 จะเป็นทีละคอลัมน์)
    pred = data.apply(rule, axis=1)
    # คำนวณความไว (sensitivity) = สัดส่วนผู้ที่เป็นโรคจริงที่กฎทำนายว่าเป็นโรค
    #   - data["Outcome"] : คำตอบจริง (y_true)
    #   - pred            : ผลการทำนาย (y_pred)
    #   - pos_label=1     : ให้นับกลุ่ม 1 (เป็นโรคเบาหวาน) เป็น "กลุ่มบวก" ที่ต้องการวัด recall
    sens = recall_score(data["Outcome"], pred, pos_label=1)
    # คำนวณความจำเพาะ (specificity) = สัดส่วนผู้ที่ไม่เป็นโรคจริงที่กฎทำนายว่าไม่เป็นโรค
    #   - pos_label=0 : ให้นับกลุ่ม 0 (ไม่เป็นโรค) เป็นกลุ่มที่วัด recall ซึ่งเท่ากับความจำเพาะ
    spec = recall_score(data["Outcome"], pred, pos_label=0)
    # คำนวณ balanced accuracy = (ความไว + ความจำเพาะ) / 2 จากคำตอบจริงและผลการทำนาย
    bacc = balanced_accuracy_score(data["Outcome"], pred)
    # พิมพ์ผลทั้ง 3 ค่า
    #   - :.3f ใน f-string คือรูปแบบตัวเลข: ทศนิยม (f = float) 3 ตำแหน่ง เช่น 0.742
    print(f"ชุดข้อมูลฝึก: sensitivity={sens:.3f}  specificity={spec:.3f}  balanced accuracy={bacc:.3f}")

# ----------------------------------------------------------------------------
# ส่วนที่ 3: วาดฮิสโทแกรมของ 3 ตัวแปร จากชุดข้อมูลฝึกเท่านั้น
# ----------------------------------------------------------------------------
# ฮิสโทแกรมของผู้ป่วยชุดข้อมูลฝึกเท่านั้น
# สร้างพื้นที่วาดภาพ 1 แถว x 3 คอลัมน์ด้วย plt.subplots(...)
#   - 1, 3 : จำนวนแถวและคอลัมน์ของกราฟย่อย
#   - figsize=(15, 3.5) : ขนาดภาพ กว้าง 15 x สูง 3.5 นิ้ว
#   - fig = ภาพทั้งภาพ, axes = รายการกราฟย่อย 3 ช่อง
fig, axes = plt.subplots(1, 3, figsize=(15, 3.5))
# วนจับคู่กราฟย่อย (ax) กับชื่อตัวแปร (col) ทีละคู่ด้วย zip(...)
#   - axes : กราฟย่อย 3 ช่อง
#   - ["Glucose", "BMI", "Age"] : list ของตัวแปร 3 ตัวที่จะวาด
for ax, col in zip(axes, ["Glucose", "BMI", "Age"]):
    # แบ่งผู้ป่วยชุดข้อมูลฝึกตาม Outcome (outcome = 0 หรือ 1, grp = ตารางย่อยของกลุ่มนั้น)
    for outcome, grp in train.groupby("Outcome"):
        # วาดฮิสโทแกรมของตัวแปร col ของกลุ่มนี้ลงในกราฟย่อย ax
        #   - alpha=0.6 : ความทึบของสี (ให้เห็นแท่งที่ซ้อนกัน)
        #   - bins=25   : จำนวนช่วง (แท่ง)
        #   - label=f"Outcome={outcome}" : ชื่อสำหรับคำอธิบายสัญลักษณ์ (legend)
        ax.hist(grp[col], alpha=0.6, bins=25, label=f"Outcome={outcome}")
    # ตั้งชื่อกราฟย่อย เช่น "Glucose (training patients only)" = เฉพาะผู้ป่วยชุดข้อมูลฝึก
    ax.set_title(f"{col} (training patients only)")
# แสดงคำอธิบายสัญลักษณ์ในกราฟย่อยช่องแรก (axes[0] = ลำดับที่ 0 คือช่องแรก)
axes[0].legend()
# จัดระยะห่างระหว่างกราฟย่อยอัตโนมัติไม่ให้ซ้อนทับกัน
plt.tight_layout()
# แสดงภาพบนหน้าจอ
plt.show()
```

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทาย — ไม่บังคับ)
#   1) สร้าง "กฎพื้นฐาน (baseline)" ชื่อ screen: ทำนายว่าเป็นโรคเบาหวาน (1) เมื่อระดับน้ำตาล
#      OGTT ที่ 2 ชั่วโมง (Glucose) >= 140 mg/dL ซึ่งเป็นเกณฑ์ภาวะความทนต่อกลูโคสบกพร่อง
#      (impaired glucose tolerance) ขึ้นไป มิฉะนั้นทำนายว่าไม่เป็น (0)
#   2) วัดผลกฎบนชุดข้อมูลฝึกด้วย evaluate(...) (จากเซลล์ก่อนหน้า)
#   3) สร้างฟังก์ชัน make_submission(...) ใช้กฎกับผู้ป่วยชุดทดสอบ 154 ราย บันทึกเป็นไฟล์ CSV
#      และส่งไปยังตารางอันดับ (เฉพาะเมื่อกรอก NAME และ JOIN_CODE ในเซลล์การตั้งค่าแล้ว)
#   สิ่งที่ผู้เรียนจะเห็น: ค่าความไว ความจำเพาะ balanced accuracy, จำนวนผู้ที่ทำนายเป็น 0/1
#   และผลการส่ง (หรือข้อความแจ้งว่ายังไม่ได้ส่ง)
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: กฎพื้นฐาน (baseline) และการวัดผล
# ----------------------------------------------------------------------------
# 🏆 กฎพื้นฐาน (baseline): คัดกรองผู้ที่มีภาวะความทนต่อกลูโคสบกพร่องขึ้นไป (ระดับน้ำตาล OGTT ที่ 2 ชั่วโมง >= 140 mg/dL)
# สร้างฟังก์ชันกฎ screen
#   - row : พารามิเตอร์ = ข้อมูลผู้ป่วย 1 ราย (1 แถวของตาราง) ที่ apply(..., axis=1) ส่งเข้ามา
def screen(row):
    """คืนค่า 1 หากทำนายว่าเป็นโรคเบาหวาน มิฉะนั้นคืนค่า 0"""
    # ตรวจเงื่อนไข แล้วคืนผลเป็นตัวเลข 0/1
    #   - row["Glucose"] : ค่า Glucose ของผู้ป่วยรายนี้
    #   - >= 140         : เปรียบเทียบ "มากกว่าหรือเท่ากับ 140" ได้ผลเป็น True (จริง) หรือ False (เท็จ)
    #   - int(...)       : ฟังก์ชันพื้นฐานของ Python แปลงเป็นจำนวนเต็ม True → 1, False → 0
    #   - return         : ส่งค่านั้นกลับเป็นผลลัพธ์ของฟังก์ชัน
    return int(row["Glucose"] >= 140)

# วัดผลกฎ screen บนผู้ป่วยชุดข้อมูลฝึกด้วยฟังก์ชัน evaluate (ส่งชื่อฟังก์ชันเข้าไปโดยไม่ใส่วงเล็บ)
#   - ไม่ระบุ data จึงใช้ค่าเริ่มต้น train
evaluate(screen)                                           # คะแนนจากผู้ป่วยชุดข้อมูลฝึก

# ----------------------------------------------------------------------------
# ส่วนที่ 2: สร้างฟังก์ชัน make_submission สำหรับสร้างไฟล์ผลการทำนายและส่งผล
#   - rule : ฟังก์ชันกฎที่จะใช้ทำนาย
#   - path="lab01_submission.csv" : ชื่อไฟล์ที่จะบันทึก (ค่าเริ่มต้น หากไม่ระบุ)
# ----------------------------------------------------------------------------
def make_submission(rule, path="lab01_submission.csv"):
    """ใช้กฎกับผู้ป่วยชุดทดสอบ 154 ราย บันทึกไฟล์ผลการทำนาย และส่งไปยังตารางอันดับ (หากระบุ NAME/JOIN_CODE แล้ว)"""
    # ใช้กฎกับผู้ป่วยชุดทดสอบทีละราย (axis=1 = ทีละแถว) ได้ผลการทำนาย 0/1 เก็บไว้ใน preds
    preds = test.apply(rule, axis=1)
    # สร้างตารางผลการทำนาย sub ที่มี 2 คอลัมน์ ด้วย pd.DataFrame({...})
    #   - ส่ง dict เข้าไป: ชื่อ (key) = ชื่อคอลัมน์, ค่า = ข้อมูลในคอลัมน์
    #   - "id": test_ids.values          : รหัสผู้ป่วยชุดทดสอบ (.values = ดึงเฉพาะค่า ไม่เอาเลขแถวกำกับ)
    #   - "prediction": preds.values     : ผลการทำนาย 0/1 ของผู้ป่วยแต่ละราย เรียงลำดับเดียวกัน
    sub = pd.DataFrame({"id": test_ids.values, "prediction": preds.values})
    # บันทึกตาราง sub เป็นไฟล์ CSV ด้วยเมธอด .to_csv(...) ของ pandas
    #   - path        : ชื่อไฟล์ที่จะบันทึก
    #   - index=False : ไม่ต้องบันทึกเลขแถว (index) ลงในไฟล์
    sub.to_csv(path, index=False)
    # พิมพ์ยืนยันการบันทึกไฟล์ พร้อมจำนวนผู้ที่ทำนายเป็น 0 และ 1
    #   - sub["prediction"].value_counts() : นับจำนวนครั้งของแต่ละค่า (0 กี่ราย, 1 กี่ราย)
    #   - .to_dict() : แปลงผลนับเป็น dict เช่น {0: 100, 1: 54} เพื่อแสดงสั้น ๆ
    #   - print(ข้อความ, ค่า) : พิมพ์หลายค่าต่อกันในบรรทัดเดียว คั่นด้วยช่องว่าง
    print(f"บันทึกไฟล์ {path} เรียบร้อย:", sub["prediction"].value_counts().to_dict())
    # if ... else ... คือ "เงื่อนไข": ทำคำสั่งชุดแรกเมื่อเงื่อนไขเป็นจริง มิฉะนั้นทำชุดหลัง else
    #   - NAME and JOIN_CODE : จริงเมื่อกรอกทั้ง NAME และ JOIN_CODE แล้ว (ไม่เป็นข้อความว่าง)
    if NAME and JOIN_CODE:
        # ส่งไฟล์ผลการทำนายไปยังตารางอันดับด้วยฟังก์ชันจากเซลล์การตั้งค่า
        #   - path : ชื่อไฟล์ที่เพิ่งบันทึก
        submit_to_leaderboard(path)
    # กรณียังไม่ได้กรอก NAME หรือ JOIN_CODE
    else:
        # พิมพ์ข้อความแจ้งว่ายังไม่ได้ส่งผล และวิธีแก้ไข
        print("ยังไม่ได้ส่งผล - โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่า (และรันเซลล์นั้นใหม่) เพื่อส่งผล")

# เรียกใช้ make_submission กับกฎพื้นฐาน screen (ใช้ชื่อไฟล์ค่าเริ่มต้น lab01_submission.csv)
make_submission(screen)
```

### แนวทางการปรับปรุงให้ดีกว่า baseline
- ทดลองเกณฑ์ตัดสิน `Glucose` ค่าอื่น และสังเกตการแลกเปลี่ยน (trade-off) ระหว่างความไวกับความจำเพาะ การตรวจ *คัดกรอง* มักให้ความสำคัญกับความไว (≥ 200 mg/dL คือเกณฑ์ *วินิจฉัย* — โปรดตรวจสอบว่าเกณฑ์นี้ให้ความไวเท่าใดในชุดข้อมูลนี้ คำใบ้: พิจารณาค่าสูงสุดของ `Glucose` ในรายงานสรุปข้อมูล)
- ใช้หลายเงื่อนไขร่วมกัน เช่น `(Glucose >= 160) or (Glucose >= 120 and BMI >= 30)` หรือเพิ่มตัวแปร `Age`
- เลือกเกณฑ์ตัดสินจากฮิสโทแกรมของชุดข้อมูลฝึก หรือจาก `train.groupby("Outcome").median(numeric_only=True)` ห้ามใช้ข้อมูลผู้ป่วยชุดทดสอบ
- เขียน loop เพื่อทดลองเกณฑ์ตัดสินหลายค่า เรียก `evaluate` สำหรับแต่ละค่า แล้วเลือกค่าที่ดีที่สุด *จากชุดข้อมูลฝึก*

```python
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทาย — ไม่บังคับ)
#   พื้นที่สำหรับผู้เรียนปรับปรุงกฎคัดกรองของตนเอง ชื่อ my_screen
#   - ค่าเริ่มต้นเหมือนกฎพื้นฐาน (Glucose >= 140) ผู้เรียนสามารถแก้ไขเกณฑ์ตัดสิน
#     หรือเพิ่มเงื่อนไขอื่น เช่น BMI, Age ภายในฟังก์ชัน
#   - จากนั้นวัดผลบนชุดข้อมูลฝึก (เปรียบเทียบกับ baseline) และสร้างไฟล์/ส่งผลไปยังตารางอันดับ
#   สิ่งที่ผู้เรียนจะเห็น: ค่าความไว ความจำเพาะ balanced accuracy ของกฎใหม่ และผลการส่ง
#   ข้อควรระวัง: เลือกเกณฑ์จากชุดข้อมูลฝึกเท่านั้น ห้ามใช้ข้อมูลผู้ป่วยชุดทดสอบ
# ============================================================================

# 🏆 กฎที่ผู้เรียนปรับปรุง (ไม่บังคับ) - แก้ไขเงื่อนไข แล้วรันเซลล์นี้เพื่อประเมินผลและส่งผล
# สร้างฟังก์ชันกฎ my_screen รับข้อมูลผู้ป่วย 1 ราย (row) แล้วคืน 1 (เป็นโรค) หรือ 0 (ไม่เป็นโรค)
def my_screen(row):
    # ค่าเริ่มต้นเหมือนกฎพื้นฐาน ให้ปรับเกณฑ์ตัดสินหรือเพิ่มเงื่อนไข (BMI, Age, ...)
    # if คือการตรวจเงื่อนไข: หากค่า Glucose ของผู้ป่วยรายนี้ >= 140 จะทำคำสั่งที่ย่อหน้าอยู่ด้านล่าง
    #   (สามารถรวมหลายเงื่อนไขด้วย and = "และ", or = "หรือ")
    if row["Glucose"] >= 140:
        # คืนค่า 1 = ทำนายว่าเป็นโรคเบาหวาน (return จะจบการทำงานของฟังก์ชันทันที)
        return 1
    # หากเงื่อนไขด้านบนไม่เป็นจริง โปรแกรมจะมาถึงบรรทัดนี้ และคืนค่า 0 = ทำนายว่าไม่เป็นโรคเบาหวาน
    return 0

# วัดผลกฎ my_screen บนชุดข้อมูลฝึก (เปรียบเทียบกับผลของ baseline ในเซลล์ก่อนหน้า)
evaluate(my_screen)                                        # คะแนนจากชุดข้อมูลฝึก - เปรียบเทียบกับ baseline
# ใช้กฎ my_screen กับผู้ป่วยชุดทดสอบ บันทึกไฟล์ lab01_submission.csv (ทับไฟล์เดิม) และส่งผล (หากกรอก NAME/JOIN_CODE)
make_submission(my_screen)
```

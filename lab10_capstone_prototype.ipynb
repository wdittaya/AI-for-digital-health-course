```python
# @title ⚙️ การตั้งค่า — โปรดรันเซลล์นี้ก่อน (NAME และ JOIN_CODE ใช้เฉพาะโจทย์ท้าทายบนตารางอันดับ ซึ่งไม่บังคับ)
!pip -q install gradio_client huggingface_hub

HF_OWNER  = "wdittaya"     # บัญชี Hugging Face ที่ใช้โฮสต์ตารางอันดับของรายวิชา (ผู้สอนเป็นผู้กำหนด)
LAB_ID    = "lab10"
NAME      = ""            # ไม่บังคับ: ชื่อที่จะแสดงบนตารางอันดับแบบสาธารณะ (สามารถใช้นามแฝงได้)
JOIN_CODE = ""            # ไม่บังคับ: รหัสเข้าร่วมซึ่งผู้สอนประกาศในการเรียนการสอนผ่าน Zoom

SPACE_ID  = f"{HF_OWNER}/aidh-leaderboard"
DATA_REPO = f"{HF_OWNER}/aidh-lab-data"

from huggingface_hub import hf_hub_download

def lab_data(path):
    """ดาวน์โหลดไฟล์ข้อมูลทดสอบสาธารณะของรายวิชา 1 ไฟล์ (เช่น 'pima/test_ids.csv') และคืนค่าตำแหน่งไฟล์ในเครื่อง"""
    return hf_hub_download(DATA_REPO, path, repo_type="dataset")

def submit_to_leaderboard(path, lab=LAB_ID):
    """ส่งไฟล์ผลการทำนายไปยังตารางอันดับของรายวิชา และแสดงคะแนนพร้อมอันดับ"""
    from gradio_client import Client, handle_file
    assert NAME and JOIN_CODE, "โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่าก่อน (จำเป็นเฉพาะการส่งผลไปยังตารางอันดับ)"
    client = Client(SPACE_ID, verbose=False)
    print(client.predict(lab=lab, name=NAME, code=JOIN_CODE, file=handle_file(path), api_name="/submit"))

print(f"Leaderboard: https://huggingface.co/spaces/{SPACE_ID}  |  lab={LAB_ID}  name={NAME or '(ยังไม่ได้ระบุ — ตารางอันดับไม่บังคับ)'}")
```

## ขั้นที่ 1 — กำหนดกรอบปัญหา (กรอกแบบฟอร์ม)

ก่อนเริ่มพัฒนา ให้อธิบายปัญหาด้วยภาษาที่ชัดเจน กรอบปัญหาที่ดีควรมีขนาดเล็กพอที่จะพัฒนาเสร็จภายในคาบเรียน และระบุความเสี่ยงอย่างตรงไปตรงมา

**ตัวอย่าง** (โจทย์การกลับมารักษาซ้ำ):

| หัวข้อ | ตัวอย่าง |
|---|---|
| ปัญหา (Problem) | ผู้ป่วยเบาหวานที่รับไว้ในโรงพยาบาลจำนวนมากกลับมารักษาซ้ำภายใน 30 วัน จึงต้องการระบุผู้ป่วยกลุ่มเสี่ยงสูงก่อนจำหน่าย |
| ผู้ใช้ (Users) | พยาบาลวางแผนจำหน่ายและผู้จัดการรายกรณี (case manager) ในขั้นตอนการวางแผนจำหน่าย |
| ประเภทงาน AI (AI task type) | การจำแนกประเภท (classification): กลับมารักษาซ้ำภายใน 30 วันหรือไม่ โดยแสดงผลเป็นคะแนนความเสี่ยง |
| ข้อมูลเข้า → ผลลัพธ์ (Input → Output) | ข้อมูลการเข้ารับบริการ (อายุ จำนวนวันนอน ประวัติการมารับบริการ ยาที่ได้รับ ฯลฯ) → ความเสี่ยง 0–100 % |
| ชุดข้อมูล (Dataset) | [Diabetes 130-US Hospitals](https://archive.ics.uci.edu/dataset/296/) (สาธารณะ ถอดข้อมูลระบุตัวตนแล้ว ปี 1999–2008) |
| ตัวชี้วัดความสำเร็จ (Success metric) | AUROC สูงกว่า 0.5 (การสุ่ม) อย่างชัดเจน โมเดลที่ตีพิมพ์มีค่าประมาณ 0.65–0.70 |
| ความเสี่ยงหลัก (Key risk) | ข้อมูลเก่าจากสหรัฐอเมริกา (การเปลี่ยนแปลงของการกระจายข้อมูลเมื่อนำมาใช้ในโรงพยาบาลไทย) และอาจทำงานได้ไม่ดีในกลุ่มผู้ป่วยที่มีข้อมูลการติดตามน้อย |

*ศัพท์:* **AUROC** (area under the ROC curve) คือค่าที่วัดความสามารถของโมเดลในการจัดอันดับผู้ป่วยที่เกิดเหตุการณ์ให้สูงกว่าผู้ที่ไม่เกิดเหตุการณ์ โดย 0.5 เท่ากับการสุ่ม และ 1.0 คือสมบูรณ์ **การเปลี่ยนแปลงของการกระจายข้อมูล (distribution shift)** คือสภาวะที่ข้อมูลในสถานที่ใช้งานจริงแตกต่างจากข้อมูลที่ใช้ฝึกโมเดล

```python
# @title ▶️ รันต้นแบบ tabular: โหลด → ทำความสะอาด → ฝึกโมเดล 2 แบบ → ประเมินผล (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
if TRACK != "tabular":
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
else:
    import io, zipfile, requests
    import numpy as np, pandas as pd, matplotlib.pyplot as plt
    from sklearn.model_selection import train_test_split, GroupShuffleSplit
    from sklearn.pipeline import make_pipeline
    from sklearn.impute import SimpleImputer
    from sklearn.preprocessing import StandardScaler
    from sklearn.linear_model import LogisticRegression
    from sklearn.ensemble import HistGradientBoostingClassifier
    from sklearn.metrics import roc_auc_score, roc_curve

    # 1) โหลดข้อมูล และแปลงผลลัพธ์เป็น 0 (ไม่เกิด) / 1 (เกิด)
    group_col = None                      # คอลัมน์ที่ระบุตัวผู้ป่วย กรณีผู้ป่วย 1 รายมีหลายแถว
    if TABULAR_DATASET == "pima":
        df = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv")
        zero_as_missing = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
        df[zero_as_missing] = df[zero_as_missing].replace(0, np.nan)   # BMI เท่ากับ 0 คือค่าที่หายไป (ปฏิบัติการที่ 1)
        target = "Outcome"
    elif TABULAR_DATASET == "heart":
        cols = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg", "thalach",
                "exang", "oldpeak", "slope", "ca", "thal", "target"]
        df = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/"
                         "processed.cleveland.data", names=cols, na_values="?")
        df["target"] = (df["target"] > 0).astype(int)                  # มีโรคหัวใจ เทียบกับ ไม่มีโรค
        target = "target"
    elif TABULAR_DATASET == "readmission":
        URL = "https://archive.ics.uci.edu/static/public/296/diabetes+130-us+hospitals+for+years+1999-2008.zip"
        z = zipfile.ZipFile(io.BytesIO(requests.get(URL, timeout=120).content))
        df = pd.read_csv(z.open(next(n for n in z.namelist() if n.endswith("diabetic_data.csv"))),
                         na_values="?", low_memory=False)
        df["readmitted_30d"] = (df.pop("readmitted") == "<30").astype(int)
        try:   # ตัดผู้ป่วยในชุดทดสอบที่ซ่อนไว้ของตารางอันดับออก เพื่อให้การประเมินตรงไปตรงมา
            hidden = pd.read_csv(lab_data("readmission/test_ids.csv"))["id"]
            df = df[~df["encounter_id"].isin(hidden)]
        except Exception as e:
            print("หมายเหตุ: ไม่สามารถดาวน์โหลดรหัสชุดทดสอบของตารางอันดับได้ จึงใช้ข้อมูลทั้งหมด", type(e).__name__)
        df = df.drop(columns=["encounter_id"])
        target, group_col = "readmitted_30d", "patient_nbr"
    elif TABULAR_DATASET == "my_csv":
        df = pd.read_csv(MY_CSV_URL)
        df = df[df[MY_TARGET].notna()]
        col = df[MY_TARGET]
        if pd.api.types.is_numeric_dtype(col) and isinstance(MY_POSITIVE_VALUE, (int, float)):
            df[MY_TARGET] = (col == MY_POSITIVE_VALUE).astype(int)
        else:
            df[MY_TARGET] = (col.astype(str).str.strip().str.lower() == str(MY_POSITIVE_VALUE).strip().lower()).astype(int)
        target = MY_TARGET
    else:
        raise ValueError(f"ไม่รู้จัก TABULAR_DATASET {TABULAR_DATASET!r} โปรดใช้ 'pima', 'heart', 'readmission' หรือ 'my_csv'")
    df = df.reset_index(drop=True)
    y = df[target].astype(int)
    print(f"ข้อมูล {len(df):,} แถว {df.shape[1] - 1} คอลัมน์; ผลลัพธ์ '{target}' เป็น 'ใช่' {y.mean():.1%} ของแถวทั้งหมด")
    assert 0 < y.mean() < 1, "คอลัมน์ผลลัพธ์มีค่าเพียงค่าเดียว โปรดตรวจสอบ MY_TARGET / MY_POSITIVE_VALUE"

    # 2) ทำความสะอาด: คอลัมน์ตัวเลขคงเดิม คอลัมน์ข้อความแปลงเป็นคอลัมน์ 0/1 (15 ค่าที่พบบ่อยที่สุด ที่เหลือเป็น 'other')
    X_raw = df.drop(columns=[target] + ([group_col] if group_col else []))
    num = X_raw.select_dtypes("number")
    cat = X_raw.select_dtypes(exclude="number").fillna("missing").astype(str)
    for c in cat.columns:
        top = cat[c].value_counts().index[:15]
        cat[c] = cat[c].where(cat[c].isin(top), "other")
    X = pd.concat([num, pd.get_dummies(cat, dtype=int)], axis=1) if cat.shape[1] else num.copy()

    # 3) แบ่งข้อมูล: กันผู้ป่วย 25% ไว้ไม่ใช้ฝึก และใช้เฉพาะการประเมินผลขั้นสุดท้าย
    if group_col:   # ผู้ป่วยมีหลายแถว -> แบ่งตามผู้ป่วย เพื่อไม่ให้ผู้ป่วยรายเดียวกันอยู่ทั้งสองส่วน
        tr, te = next(GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42).split(X, y, groups=df[group_col]))
    else:
        tr, te = train_test_split(np.arange(len(X)), test_size=0.25, stratify=y, random_state=42)
    X_train, X_test, y_train, y_test = X.iloc[tr], X.iloc[te], y.iloc[tr], y.iloc[te]
    raw_test = df.iloc[te]
    threshold = y_train.mean()   # ระบุผู้ป่วยเป็นกลุ่มเสี่ยงเมื่อความเสี่ยงที่ทำนายสูงกว่าความเสี่ยงเฉลี่ย

    # 4) ฝึกโมเดล 2 แบบ และ 5) ประเมินผลกับผู้ป่วยที่กันไว้
    models = {
        "Logistic regression": make_pipeline(SimpleImputer(strategy="median"), StandardScaler(),
                                             LogisticRegression(max_iter=2000)),
        "Gradient boosting": HistGradientBoostingClassifier(random_state=42),
    }
    rows, test_probs = [], {}
    for name, m in models.items():
        m.fit(X_train, y_train)
        p = m.predict_proba(X_test)[:, 1]
        test_probs[name] = p
        flag = p >= threshold
        rows.append({"model": name,
                     "AUROC": roc_auc_score(y_test, p),
                     "sensitivity": (flag & (y_test == 1)).sum() / (y_test == 1).sum(),
                     "specificity": (~flag & (y_test == 0)).sum() / (y_test == 0).sum()})
    rows.append({"model": "Baseline: coin flip", "AUROC": 0.5, "sensitivity": 0.5, "specificity": 0.5})
    results = pd.DataFrame(rows).set_index("model").round(3)
    best_name = results.drop(index="Baseline: coin flip")["AUROC"].idxmax()
    best_model = models[best_name]
    print(f"\nชุดทดสอบ: {len(y_test):,} แถวที่โมเดลไม่เคยเห็น เกณฑ์ตัดสิน (threshold) = ความเสี่ยงที่ทำนาย {threshold:.0%}\n")
    print(results.to_string())
    print(f"\nโมเดลที่ดีที่สุด: {best_name} (AUROC {results.loc[best_name, 'AUROC']:.3f})")

    plt.figure(figsize=(5, 4))
    for name, p in test_probs.items():
        fpr, tpr, _ = roc_curve(y_test, p)
        plt.plot(fpr, tpr, label=f"{name} (AUROC {roc_auc_score(y_test, p):.2f})")
    plt.plot([0, 1], [0, 1], "k--", label="coin flip (0.50)")
    plt.xlabel("1 - specificity (false alarms)"); plt.ylabel("sensitivity (cases found)")
    plt.title(f"ROC curve - {TABULAR_DATASET}"); plt.legend(fontsize=8); plt.show()

    def demo(patient_number):
        """แสดงข้อมูลผู้ป่วย 1 รายที่โมเดลไม่เคยเห็น พร้อมความเสี่ยงที่โมเดลทำนาย"""
        i = int(patient_number)
        risk = best_model.predict_proba(X_test.iloc[[i]])[0, 1]
        print(raw_test.iloc[i].drop(labels=[target]).head(12).to_string())
        return (f"\nผู้ป่วยหมายเลข {i}: ความเสี่ยงที่ทำนาย {risk:.0%} -> "
                f"{'ระบุเป็นกลุ่มเสี่ยง ควรติดตาม' if risk >= threshold else 'ไม่ระบุเป็นกลุ่มเสี่ยง'} (เกณฑ์ตัดสิน {threshold:.0%}) "
                f"ผลลัพธ์จริง: {'เกิดเหตุการณ์' if y_test.iloc[i] == 1 else 'ไม่เกิดเหตุการณ์'}")
    print(f"\nฟังก์ชัน demo() พร้อมใช้งาน: ทดลอง demo(0) ถึง demo({len(y_test) - 1})")
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
if is_blank(globals().get("SYSTEM_PROMPT")):
    SYSTEM_PROMPT = ("You are a clinical guideline assistant for primary-care nurses. Answer in 1-3 sentences "
                     "using ONLY the numbered context passages, and cite them like [1]. If the answer is not "
                     "in the context, reply exactly: 'Not in the guidelines.'")
if is_blank(globals().get("KNOWLEDGE")):
    KNOWLEDGE = "course_kb"
if "MY_PASSAGES" not in globals():
    MY_PASSAGES = []
if is_blank(globals().get("TEST_CASES")):
    TEST_CASES = [
        ("What is the blood pressure target for most adults with diabetes?", "130/80"),
        ("What HbA1c target is appropriate for a frail older adult with diabetes?", "8"),
        ("How is mild hypoglycaemia treated in a conscious patient?", "15"),
        ("How often should a person with diabetes have a foot examination?", "annual"),
        ("What is the first-line treatment for scabies?", "Not in the guidelines"),   # อยู่นอกขอบเขตฐานความรู้
    ]
print("แหล่งความรู้:", KNOWLEDGE, "| จำนวนกรณีทดสอบ:", len(TEST_CASES))
```

```python
# @title ▶️ รันต้นแบบ LLM: โหลดโมเดล → (สร้างดัชนีแหล่งความรู้) → รันกรณีทดสอบ (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
if TRACK != "llm":
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
else:
    import subprocess, sys, json
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "sentence-transformers",
                    "transformers>=4.44", "accelerate"], check=True)
    import numpy as np, torch
    from transformers import pipeline

    # 1) โหลด LLM แบบเปิด (ตัวเดียวกับปฏิบัติการที่ 8-9)
    gpu = torch.cuda.is_available()
    llm_name = "Qwen/Qwen2.5-1.5B-Instruct" if gpu else "Qwen/Qwen2.5-0.5B-Instruct"
    if not gpu:
        print("⚠️ ไม่พบ GPU จึงใช้โมเดลขนาดเล็กกว่า", llm_name, "(ควรเปลี่ยนเป็น runtime แบบ T4 GPU เพื่อคุณภาพคำตอบที่ดีขึ้น)")
    llm = pipeline("text-generation", model=llm_name, device_map="auto",
                   torch_dtype=torch.float16 if gpu else torch.float32)

    def chat(system, user, max_new_tokens=250):
        messages = [{"role": "system", "content": system}, {"role": "user", "content": user}]
        return llm(messages, max_new_tokens=max_new_tokens, do_sample=False)[0]["generated_text"][-1]["content"]

    # 2) แหล่งความรู้: แปลงทุกข้อความเป็นเวกเตอร์ (embedding) เพื่อค้นหาข้อความที่ใกล้เคียงกับคำถามที่สุด (ปฏิบัติการที่ 9)
    if KNOWLEDGE == "course_kb":
        try:
            passages = [d["text"] for d in json.load(open(lab_data("rag/kb.json")))]
        except Exception as e:   # เข้าถึงข้อมูลของรายวิชาไม่ได้ -> ใช้ฐานความรู้ขนาดเล็กจากปฏิบัติการที่ 9
            print("หมายเหตุ: ไม่สามารถเข้าถึงฐานความรู้ของรายวิชาได้ จึงใช้ฐานความรู้ 6 ข้อความจากปฏิบัติการที่ 9", type(e).__name__)
            passages = [
                "Type 2 diabetes is diagnosed when HbA1c is 6.5% or higher, or fasting glucose >= 126 mg/dL.",
                "First-line pharmacotherapy for type 2 diabetes is metformin unless contraindicated.",
                "Blood pressure target for most adults with diabetes is below 130/80 mmHg.",
                "Statins are recommended for diabetic patients aged 40-75 to reduce cardiovascular risk.",
                "Metformin is contraindicated when eGFR is below 30 mL/min/1.73m2.",
                "Annual retinal screening is recommended for patients with type 2 diabetes.",
            ]
    elif KNOWLEDGE == "my_passages":
        passages = list(MY_PASSAGES)
        assert passages, "MY_PASSAGES ยังว่างอยู่ โปรดวางข้อความ หรือกำหนด KNOWLEDGE = 'course_kb'"
    else:
        passages = []
    if passages:
        from sentence_transformers import SentenceTransformer
        embedder = SentenceTransformer("all-MiniLM-L6-v2")
        passage_vecs = embedder.encode(passages, normalize_embeddings=True)
        print(f"สร้างดัชนีข้อความแล้ว {len(passages)} ข้อความ")

    def retrieve(question, k=3):
        sims = passage_vecs @ embedder.encode([question], normalize_embeddings=True)[0]
        return [passages[i] for i in np.argsort(-sims)[:k]]

    def demo(question):
        """ตอบคำถาม 1 ข้อ (RAG) หรือประมวลผลข้อความที่ป้อน 1 ข้อความ (ใช้คำสั่งอย่างเดียว)"""
        if passages:
            context = "\n".join(f"[{i + 1}] {p}" for i, p in enumerate(retrieve(question)))
            answer = chat(SYSTEM_PROMPT, f"Context:\n{context}\n\nQuestion: {question}")
            return f"{answer}\n\n--- retrieved passages ---\n{context}"
        return chat(SYSTEM_PROMPT, question)

    # 3) ประเมินผล: คำตอบแต่ละข้อมีคำที่คาดหวังหรือไม่
    hits = 0
    for question, expected in TEST_CASES:
        answer = demo(question).split("\n\n--- retrieved")[0]
        ok = expected.lower() in answer.lower()
        hits += ok
        print(f"{'✅' if ok else '❌'} {question}\n   คำที่คาดหวัง '{expected}' | คำตอบ: {answer[:200]}\n")
    print(f"คะแนน: ผ่าน {hits}/{len(TEST_CASES)} กรณีทดสอบ ({hits / len(TEST_CASES):.0%})")
```

## ขั้นที่ 3C — แนวทาง image: โมเดลจำแนกภาพทางการแพทย์ (ปฏิบัติการที่ 4) · แนะนำ GPU

โค้ดเริ่มต้นจะดาวน์โหลดชุดข้อมูล [MedMNIST](https://medmnist.com/) ขนาดเล็ก ปรับโมเดล **ResNet-18** ที่ฝึกไว้ล่วงหน้าด้วย ImageNet (การเรียนรู้แบบถ่ายโอน (transfer learning) เช่นเดียวกับปฏิบัติการที่ 4) และฝึกโมเดลกับภาพเป็นจำนวนรอบที่กำหนด (**epoch**) จากนั้นประเมินผลกับภาพชุดทดสอบ

> ทำงานเร็วขึ้นเมื่อเลือก `Runtime → Change runtime type → T4 GPU` หากใช้ CPU ให้กำหนด `EPOCHS` ไว้ที่ 2–3 (ใช้เวลาไม่กี่นาที)

```python
# ⭐ Exercise 2C — แบบฝึกหัดที่ 2C: กำหนดชุดภาพให้ต้นแบบ image (เฉพาะแนวทาง image)
# เลือกชุดข้อมูล MedMNIST 1 ชุด:
#   "pneumoniamnist" -> ภาพเอกซเรย์ทรวงอก: ปกติ เทียบกับ ปอดอักเสบ (ปฏิบัติการที่ 4)   ประมาณ 5,900 ภาพ
#   "dermamnist"     -> รอยโรคผิวหนัง 7 ประเภท                                   ประมาณ 10,000 ภาพ
#   "bloodmnist"     -> ภาพเซลล์เม็ดเลือดจากกล้องจุลทรรศน์ 8 ชนิด                  ประมาณ 17,000 ภาพ (ช้าบน CPU ควรใช้ GPU)
IMAGE_DATASET = ...    # TODO เช่น "pneumoniamnist"
EPOCHS = 3             # จำนวนรอบการฝึกกับภาพทั้งหมด: มากขึ้นมักแม่นยำขึ้นแต่ช้าลง (1-10)

print("ชุดภาพ:", IMAGE_DATASET, "| epochs:", EPOCHS)
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
if is_blank(globals().get("IMAGE_DATASET")):
    IMAGE_DATASET = "pneumoniamnist"
if is_blank(globals().get("EPOCHS")):
    EPOCHS = 3
print("ชุดภาพ:", IMAGE_DATASET, "| epochs:", EPOCHS)
```

```python
# @title ▶️ รันต้นแบบ image: ดาวน์โหลด → ปรับแต่ง ResNet-18 → ประเมินผล (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
if TRACK != "image":
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
else:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "medmnist"], check=True)
    import numpy as np, matplotlib.pyplot as plt, torch, torch.nn as nn
    import medmnist
    from medmnist import INFO
    from torchvision import transforms, models
    from torch.utils.data import DataLoader
    from sklearn.metrics import roc_auc_score, confusion_matrix, ConfusionMatrixDisplay

    # 1) ข้อมูล
    info = INFO[IMAGE_DATASET]
    assert "multi-label" not in info["task"], "โปรดเลือกชุดข้อมูลที่แต่ละภาพมีประเภทเดียว (single-label)"
    class_names = [info["label"][str(i)] for i in range(len(info["label"]))]
    n_ch = info["n_channels"]
    device = "cuda" if torch.cuda.is_available() else "cpu"
    tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize([.5] * n_ch, [.5] * n_ch)])
    DataClass = getattr(medmnist, info["python_class"])
    train_ds = DataClass(split="train", transform=tf, download=True, size=28)
    test_ds = DataClass(split="test", transform=tf, download=True, size=28)
    print(f"{IMAGE_DATASET}: ภาพฝึก {len(train_ds)} ภาพ / ภาพทดสอบ {len(test_ds)} ภาพ ประเภท: {class_names} อุปกรณ์: {device}")
    train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
    test_loader = DataLoader(test_ds, batch_size=256)

    # 2) โมเดล: ResNet-18 ที่ฝึกไว้ล่วงหน้า ปรับให้ตรงกับจำนวนช่องสีและจำนวนประเภท (ปฏิบัติการที่ 4 แบบฝึกหัดที่ 1)
    net = models.resnet18(weights="IMAGENET1K_V1")
    if n_ch == 1:
        net.conv1 = nn.Conv2d(1, 64, kernel_size=7, stride=2, padding=3, bias=False)
    net.fc = nn.Linear(net.fc.in_features, len(class_names))
    net = net.to(device)
    opt = torch.optim.Adam(net.parameters(), lr=1e-4)   # อัตราการเรียนรู้ต่ำ: ปรับแต่งโมเดลอย่างค่อยเป็นค่อยไป
    crit = nn.CrossEntropyLoss()

    # 3) ฝึกโมเดล
    for epoch in range(int(EPOCHS)):
        net.train(); correct = total = 0
        for x, yb in train_loader:
            x, yb = x.to(device), yb.squeeze(1).long().to(device)
            out = net(x); loss = crit(out, yb)
            opt.zero_grad(); loss.backward(); opt.step()
            correct += (out.argmax(1) == yb).sum().item(); total += len(yb)
        print(f"epoch {epoch + 1}/{EPOCHS}: ความแม่นยำกับข้อมูลฝึก (train accuracy) {correct / total:.3f}")

    # 4) ประเมินผลกับภาพทดสอบที่โมเดลไม่เคยเห็น
    net.eval(); probs, y_true = [], []
    with torch.no_grad():
        for x, yb in test_loader:
            probs.append(torch.softmax(net(x.to(device)), 1).cpu().numpy()); y_true += yb.squeeze(1).tolist()
    probs, y_true = np.concatenate(probs), np.array(y_true)
    y_pred = probs.argmax(1)
    auroc = roc_auc_score(y_true, probs[:, 1]) if len(class_names) == 2 else roc_auc_score(y_true, probs, multi_class="ovr")
    majority = np.bincount(y_true).max() / len(y_true)
    print(f"\nความแม่นยำกับชุดทดสอบ (test accuracy) {np.mean(y_pred == y_true):.3f} "
          f"(ค่าอ้างอิง 'ทายประเภทที่พบบ่อยที่สุดเสมอ': {majority:.3f}) | AUROC {auroc:.3f}")
    cm = confusion_matrix(y_true, y_pred, labels=range(len(class_names)))
    for i, name in enumerate(class_names):
        print(f"  จากภาพ '{name}' ทั้งหมด โมเดลจำแนกถูก {cm[i, i] / max(cm[i].sum(), 1):.0%}")
    ConfusionMatrixDisplay(cm, display_labels=[n[:12] for n in class_names]).plot(xticks_rotation=45)
    plt.title("Confusion matrix (rows = truth)"); plt.show()

    def demo(image_number):
        """แสดงภาพทดสอบ 1 ภาพที่โมเดลไม่เคยเห็น พร้อมผลการทำนาย"""
        i = int(image_number)
        img = test_ds.imgs[i]
        plt.figure(figsize=(2.5, 2.5)); plt.imshow(img, cmap="gray" if n_ch == 1 else None); plt.axis("off"); plt.show()
        p = probs[i]
        return (f"ภาพหมายเลข {i}: โมเดลทำนายว่า '{class_names[p.argmax()]}' (ความมั่นใจ {p.max():.0%}) "
                f"ประเภทจริง: '{class_names[y_true[i]]}'")
    print(f"\nฟังก์ชัน demo() พร้อมใช้งาน: ทดลอง demo(0) ถึง demo({len(test_ds) - 1})")
```

💡 **คำอธิบายผลลัพธ์** โมเดลได้เรียนรู้ลักษณะทั่วไปของภาพ (ขอบ พื้นผิว) จากภาพทั่วไปจำนวนหลายล้านภาพไว้แล้ว การฝึกในขั้นนี้เป็นเพียงการปรับโมเดลให้เข้ากับภาพทางการแพทย์ ซึ่งเรียกว่า **การเรียนรู้แบบถ่ายโอน (transfer learning)** ค่า **ความแม่นยำ (accuracy)** เพียงค่าเดียวอาจทำให้เข้าใจผิดเมื่อมีประเภทใดประเภทหนึ่งเป็นส่วนใหญ่ จึงต้องเปรียบเทียบกับค่าอ้างอิง "ทายประเภทที่พบบ่อยที่สุดเสมอ" และพิจารณาอัตราการจำแนกถูกรายประเภทร่วมกับ **เมทริกซ์ความสับสน (confusion matrix)** ซึ่งแสดงว่าประเภทใดถูกจำแนกสลับกัน

1. *(ไม่บังคับ)* รันเซลล์ช่วยด้านล่างเพื่อดาวน์โหลดไฟล์ zip ของภาพเอกซเรย์ทรวงอกตัวอย่าง (PneumoniaMNIST) ซึ่งแบ่งเป็นโฟลเดอร์ `train/` และ `test/` หรือใช้ภาพสาธารณะอื่น (เช่น ภาพรอยโรคผิวหนังจาก [ISIC](https://gallery.isic-archive.com/)) หรือเสียงไอ / เสียงหายใจโดยใช้ **Audio project**

```python
# @title 📦 เซลล์ช่วยสำหรับแนวทาง nocode: ดาวน์โหลดภาพเอกซเรย์ตัวอย่างสำหรับ Teachable Machine (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
if TRACK != "nocode":
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
else:
    import subprocess, sys, os, shutil
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "medmnist"], check=True)
    from medmnist import PneumoniaMNIST
    names = {0: "normal", 1: "pneumonia"}
    for split, per_class in [("train", 40), ("test", 15)]:
        ds = PneumoniaMNIST(split=split, download=True, size=64)   # ภาพเอกซเรย์ทรวงอกขนาด 64x64 พิกเซล
        count = {0: 0, 1: 0}
        for img, label in ds:
            lab = int(label[0])
            if count[lab] < per_class:
                folder = f"nocode_images/{split}/{names[lab]}"
                os.makedirs(folder, exist_ok=True)
                img.save(f"{folder}/{split}_{names[lab]}_{count[lab]:02d}.png")
                count[lab] += 1
            if min(count.values()) >= per_class:
                break
    shutil.make_archive("nocode_images", "zip", "nocode_images")
    print("สร้างไฟล์ nocode_images.zip แล้ว: train/ = 40 + 40 ภาพ, test/ = 15 + 15 ภาพ")
    if "google.colab" in sys.modules:
        from google.colab import files
        files.download("nocode_images.zip")   # แตกไฟล์ในเครื่องคอมพิวเตอร์ แล้วอัปโหลดโฟลเดอร์ไปยัง Teachable Machine
```

โจทย์คือการทำนายว่าผู้ป่วยเบาหวานที่รับไว้ในโรงพยาบาลจะ **กลับมารักษาซ้ำภายใน 30 วัน** หรือไม่ (Diabetes 130-US hospitals, 101,766 ครั้งของการนอนโรงพยาบาล, ปี 1999–2008) ชุดทดสอบที่กันไว้แบ่ง **ตามผู้ป่วย** (20 % ของผู้ป่วย = 20,352 ครั้ง) ผู้ป่วยรายเดียวกันจึงไม่ปรากฏทั้งในข้อมูลฝึกและข้อมูลทดสอบ ซึ่งเป็นการป้องกันการรั่วไหลของข้อมูล (data leakage) ที่พบบ่อยในงานวิจัยด้วยข้อมูล EHR และทำให้ผลลัพธ์สูงเกินจริง **ตัวชี้วัด: ROC AUC** (โมเดลที่ตีพิมพ์มีค่าประมาณ 0.65–0.70 และกลุ่มผลบวกมีเพียงประมาณ 11 % ค่า accuracy จึงไม่มีความหมาย) ไฟล์ที่ส่ง: `lab10_submission.csv` มีคอลัมน์ `id,prob` (`id` = `encounter_id` จาก `readmission/test_ids.csv`, `prob` = ความน่าจะเป็นที่จะกลับมารักษาซ้ำภายใน 30 วัน)

โมเดลอ้างอิง (baseline) ด้านล่างทำงานได้ทันที โดยสร้างไฟล์ส่งที่ถูกต้อง และจะส่งผลไปยังตารางอันดับเมื่อผู้เรียนระบุ `NAME` และ `JOIN_CODE` ในเซลล์การตั้งค่าแล้ว (จากนั้นรันเซลล์การตั้งค่าอีกครั้ง)

> **หลักความซื่อสัตย์ทางวิชาการ (Honour code)** ป้ายกำกับ (label) ที่ซ่อนไว้มาจากชุดข้อมูลสาธารณะ การค้นหาป้ายกำกับเหล่านั้นจึงทำได้ง่ายแต่ไม่มีประโยชน์ใด ๆ ตารางอันดับมีไว้เพื่อการเรียนรู้ และโน้ตบุ๊กของผู้เรียนจะได้รับการประเมินด้วย ห้ามแก้ไขผลการทำนายด้วยมือ

```python
# @title 🏆 โมเดลอ้างอิง: จำนวนการใช้บริการ + gradient boosting → lab10_submission.csv
import io, zipfile, requests, pandas as pd
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import GroupKFold, cross_val_score

URL = "https://archive.ics.uci.edu/static/public/296/diabetes+130-us+hospitals+for+years+1999-2008.zip"
z = zipfile.ZipFile(io.BytesIO(requests.get(URL, timeout=120).content))
rdf = pd.read_csv(z.open(next(n for n in z.namelist() if n.endswith("diabetic_data.csv"))), low_memory=False)
rdf["y"] = (rdf["readmitted"] == "<30").astype(int)

test_ids = pd.read_csv(lab_data("readmission/test_ids.csv"))["id"]
is_test = rdf["encounter_id"].isin(test_ids)
lb_train = rdf[~is_test]
lb_test = rdf[is_test].drop(columns=["readmitted", "y"])        # ป้ายกำกับถูกซ่อน ห้ามนำมาใช้
print(lb_train.shape, lb_test.shape, "สัดส่วนผลบวกในข้อมูลฝึก:", round(lb_train["y"].mean(), 3))

features = ["time_in_hospital", "num_lab_procedures", "num_procedures", "num_medications",
            "number_outpatient", "number_emergency", "number_inpatient", "number_diagnoses"]
clf = HistGradientBoostingClassifier(random_state=42)
cv = GroupKFold(n_splits=5)                                      # แบ่งตามผู้ป่วย เช่นเดียวกับชุดทดสอบของตารางอันดับ
auc = cross_val_score(clf, lb_train[features], lb_train["y"], cv=cv, groups=lb_train["patient_nbr"], scoring="roc_auc")
print("grouped CV AUC ของโมเดลอ้างอิง:", auc.mean().round(3))

clf.fit(lb_train[features], lb_train["y"])
prob = clf.predict_proba(lb_test[features])[:, 1]
pd.DataFrame({"id": lb_test["encounter_id"].values, "prob": prob}).to_csv("lab10_submission.csv", index=False)
print("บันทึกไฟล์ lab10_submission.csv แล้ว จำนวน", len(prob), "แถว")

if NAME and JOIN_CODE:
    submit_to_leaderboard("lab10_submission.csv")
else:
    print("ยังไม่ได้ส่งผล: โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่า รันเซลล์การตั้งค่าอีกครั้ง แล้วรันเซลล์นี้ใหม่")
```

**แนวทางปรับปรุงให้ดีกว่าโมเดลอ้างอิง**

- เพิ่มตัวแปรเชิงกลุ่ม (categorical features) เช่น ช่วงอายุ `age`, `admission_type_id`, `discharge_disposition_id` (เช่น จำหน่ายไปยังสถานดูแล เทียบกับ กลับบ้าน), `admission_source_id`, `A1Cresult`, `insulin`, `change`, `diabetesMed`
- จัดกลุ่มรหัส ICD-9 ใน `diag_1`/`diag_2`/`diag_3` ตามหมวด (ระบบไหลเวียนโลหิต 390–459, ระบบหายใจ 460–519, เบาหวาน 250.xx …)
- ประวัติการใช้บริการ: จำนวนครั้งรวม = outpatient + emergency + inpatient และจำนวนครั้งที่ผู้ป่วยรายเดียวกันเคยเข้ารับบริการก่อนหน้า
- ปรับไฮเปอร์พารามิเตอร์ (hyperparameter) ของโมเดล (`learning_rate`, `max_leaf_nodes`, `l2_regularization`) โดยใช้ `GroupKFold` แบ่งตาม `patient_nbr` ทุกครั้ง
- ผู้ป่วยที่เสียชีวิตหรือจำหน่ายไปยังสถานดูแลระยะสุดท้าย (hospice) (`discharge_disposition_id` 11, 13, 14, 19–21) ไม่สามารถกลับมารักษาซ้ำได้ ซึ่งเป็นลักษณะเฉพาะที่ทราบกันดีของชุดข้อมูลนี้และควรนำมาพิจารณา

```python
# 🏆 ไม่บังคับ — โมเดลที่ผู้เรียนปรับปรุง (รันได้ทันที; เพิ่มคอลัมน์ใน EXTRA_CATEGORICAL เพื่อเพิ่มคะแนน)
EXTRA_CATEGORICAL = []   # เช่น ["age", "discharge_disposition_id", "admission_source_id", "A1Cresult", "insulin", "change", "diabetesMed"]
SUBMIT_IMPROVED = False  # กำหนดเป็น True เพื่อเขียนทับ lab10_submission.csv ด้วยโมเดลนี้และส่งผล

def make_X(frame, ref=None):
    """ตัวแปรตัวเลขของโมเดลอ้างอิง + one-hot ของตัวแปรเชิงกลุ่มที่เพิ่ม (30 ค่าที่พบบ่อยที่สุด ที่เหลือเป็น 'other')"""
    parts = [frame[features]]
    for c in EXTRA_CATEGORICAL:
        top = (ref if ref is not None else frame)[c].astype(str).value_counts().index[:30]
        s = frame[c].astype(str).where(frame[c].astype(str).isin(top), "other")
        parts.append(pd.get_dummies(s, prefix=c, dtype=int))
    return pd.concat(parts, axis=1)

X_lb = make_X(lb_train)
X_lb_test = make_X(lb_test, ref=lb_train).reindex(columns=X_lb.columns, fill_value=0)
clf2 = HistGradientBoostingClassifier(random_state=42)
auc2 = cross_val_score(clf2, X_lb, lb_train["y"], cv=GroupKFold(n_splits=5), groups=lb_train["patient_nbr"], scoring="roc_auc")
print(f"grouped CV AUC ของโมเดลที่ปรับปรุง: {auc2.mean():.3f}  (โมเดลอ้างอิง {auc.mean():.3f}) จำนวนตัวแปร {X_lb.shape[1]}")

if SUBMIT_IMPROVED:
    clf2.fit(X_lb, lb_train["y"])
    pd.DataFrame({"id": lb_test["encounter_id"].values,
                  "prob": clf2.predict_proba(X_lb_test)[:, 1]}).to_csv("lab10_submission.csv", index=False)
    if NAME and JOIN_CODE:
        submit_to_leaderboard("lab10_submission.csv")
    else:
        print("บันทึกไฟล์ lab10_submission.csv แล้ว โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่าเพื่อส่งผล")
```

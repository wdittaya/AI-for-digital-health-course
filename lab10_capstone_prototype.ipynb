## วัตถุประสงค์การเรียนรู้

- กำหนดขอบเขตปัญหาด้านสุขภาพดิจิทัลจริงให้เป็นต้นแบบปัญญาประดิษฐ์ (AI) ที่สร้างได้
- ประกอบขั้นตอนการทำงาน (pipeline) โดยนำเทคนิคจากปฏิบัติการที่ 1–9 มาใช้ซ้ำ
- ประเมินผลอย่างตรงไปตรงมาและระบุข้อจำกัด
- เตรียมการสาธิต (demo) และการนำเสนอสดผ่าน Zoom
- *(ไม่บังคับ)* ร่วมโจทย์มาตรฐานกลาง **การกลับมารักษาซ้ำภายใน 30 วัน (30-day readmission)** บนตารางอันดับ (leaderboard) ด้วยโมเดลที่ไม่มีการรั่วไหลของข้อมูล (data leakage)

> **แนวทางการทำปฏิบัติการ** (งานเดี่ยว)
> - ✅ **ส่วนหลัก — สำหรับผู้เรียนทุกคน ไม่จำเป็นต้องมีพื้นฐานการเขียนโปรแกรม** ให้รันเซลล์ตามลำดับจากบนลงล่าง (กด Shift+Enter หรือคลิกปุ่ม ▶) แบบฝึกหัดแต่ละข้อมีเซลล์ ✅ เฉลย อยู่ถัดไป ให้ทดลองทำด้วยตนเองก่อน หากไม่สามารถดำเนินการต่อได้ ให้รันเซลล์เฉลยแล้วทำขั้นตอนถัดไป จากนั้นตอบคำถาม ✍️ ในเซลล์ข้อความ ซึ่งถือเป็นชิ้นงานที่ต้องส่ง แบบฝึกหัดในปฏิบัติการนี้ส่วนใหญ่เป็น *การเลือก* ได้แก่ การเลือกแนวทาง (track) การกำหนดข้อมูลหรือคำสั่ง (prompt) ให้ต้นแบบ และการเลือกตัวอย่างสำหรับการสาธิต เซลล์เฉลยจะเติมเฉพาะค่าที่ผู้เรียนเว้นว่างไว้ และจะไม่เขียนทับค่าที่ผู้เรียนเลือกไว้แล้ว
> - 🏆 **โจทย์ท้าทายบนตารางอันดับ — ไม่บังคับ สำหรับผู้เรียนที่เขียนโปรแกรมได้** อยู่ท้ายโน้ตบุ๊ก (notebook) และไม่นับเป็นส่วนหนึ่งของชิ้นงานที่ต้องส่ง
> - หากพบปัญหา สามารถสอบถามผู้สอนผ่านช่องแชตของ Zoom

> **สภาพแวดล้อมการรัน (Runtime):** ทุกขั้นตอนรันได้บน [Google Colab](https://colab.research.google.com) รุ่นฟรี แนวทาง *tabular* และ *nocode* ใช้ CPU ตามค่าเริ่มต้น แนวทาง *llm* ต้องใช้ GPU โดยเลือก `Runtime → Change runtime type → T4 GPU` **ก่อน** เริ่มทำ (การเปลี่ยน runtime จะเริ่มระบบใหม่ จึงต้องรันโน้ตบุ๊กตั้งแต่ต้นอีกครั้ง) แนวทาง *image* ทำงานเร็วขึ้นเมื่อใช้ GPU แต่สามารถรันบน CPU ได้

```python
# @title ⚙️ การตั้งค่า — โปรดรันเซลล์นี้ก่อน (NAME และ JOIN_CODE ใช้เฉพาะโจทย์ท้าทายบนตารางอันดับ ซึ่งไม่บังคับ)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้เป็นการ "เตรียมความพร้อม" ก่อนเริ่มปฏิบัติการ โดยทำ 3 อย่าง ได้แก่
#   1) ติดตั้งเครื่องมือ (ไลบรารี) เพิ่มเติมที่จำเป็นต้องใช้ใน Google Colab
#   2) กำหนดค่าพื้นฐาน เช่น รหัสปฏิบัติการ ชื่อที่จะแสดง และรหัสเข้าร่วมตารางอันดับ
#   3) สร้าง "ฟังก์ชัน" (ชุดคำสั่งสำเร็จรูปที่เรียกใช้ซ้ำได้) 2 ตัว คือ
#        - lab_data(...)              : ดาวน์โหลดไฟล์ข้อมูลของรายวิชา
#        - submit_to_leaderboard(...) : ส่งผลการทำนายไปยังตารางอันดับ (ไม่บังคับ)
#   ผู้เรียนเพียงกดรันเซลล์นี้ 1 ครั้ง ไม่จำเป็นต้องแก้ไขโค้ด
#   (ยกเว้นต้องการเข้าร่วมตารางอันดับ ให้กรอก NAME และ JOIN_CODE ด้านล่าง)
# ============================================================================

# ติดตั้งไลบรารีเพิ่มเติมลงในเครื่อง Colab
#   - เครื่องหมาย ! ด้านหน้า หมายถึงสั่งงานระบบปฏิบัติการ (ไม่ใช่คำสั่งภาษา Python)
#   - pip install : คำสั่งติดตั้งไลบรารีของ Python
#   - -q (quiet)  : ให้แสดงข้อความระหว่างติดตั้งให้น้อยที่สุด
#   - gradio_client   : ไลบรารีสำหรับติดต่อกับเว็บแอปตารางอันดับ (Hugging Face Space)
#   - huggingface_hub : ไลบรารีสำหรับดาวน์โหลดไฟล์จากเว็บไซต์ Hugging Face
!pip -q install gradio_client huggingface_hub

# สร้างตัวแปร (ช่องเก็บค่า) HF_OWNER แล้วใส่ข้อความ "wdittaya" ไว้
#   เครื่องหมาย = หมายถึง "นำค่าทางขวา ไปเก็บไว้ในชื่อทางซ้าย"
#   ข้อความในภาษา Python ต้องอยู่ในเครื่องหมายคำพูด " "
HF_OWNER  = "wdittaya"     # บัญชี Hugging Face ที่ใช้โฮสต์ตารางอันดับของรายวิชา (ผู้สอนเป็นผู้กำหนด)
# รหัสของปฏิบัติการนี้ ใช้ระบุว่าผลที่ส่งเป็นของปฏิบัติการใด (ไม่ต้องแก้ไข)
LAB_ID    = "lab10"
# ชื่อที่จะแสดงบนตารางอันดับ: พิมพ์ชื่อไว้ระหว่างเครื่องหมาย " " เช่น NAME = "หมอเอ"
NAME      = ""            # ไม่บังคับ: ชื่อที่จะแสดงบนตารางอันดับแบบสาธารณะ (สามารถใช้นามแฝงได้)
# รหัสเข้าร่วมตารางอันดับ: พิมพ์รหัสที่ผู้สอนแจ้งไว้ระหว่างเครื่องหมาย " "
JOIN_CODE = ""            # ไม่บังคับ: รหัสเข้าร่วมซึ่งผู้สอนประกาศในการเรียนการสอนผ่าน Zoom

# สร้างชื่อเว็บแอปตารางอันดับ โดยนำ HF_OWNER มาต่อกับข้อความ
#   - f"..." (f-string) คือข้อความที่แทรกค่าตัวแปรได้ โดยเขียนชื่อตัวแปรไว้ใน { }
#   - ผลลัพธ์คือ "wdittaya/aidh-leaderboard"
SPACE_ID  = f"{HF_OWNER}/aidh-leaderboard"
# สร้างชื่อคลังข้อมูล (dataset) ของรายวิชาในลักษณะเดียวกัน ผลลัพธ์คือ "wdittaya/aidh-lab-data"
DATA_REPO = f"{HF_OWNER}/aidh-lab-data"

# นำเข้า (import) ฟังก์ชัน hf_hub_download จากไลบรารี huggingface_hub
#   ฟังก์ชันนี้ใช้ดาวน์โหลดไฟล์จากเว็บไซต์ Hugging Face มาเก็บไว้ในเครื่อง
from huggingface_hub import hf_hub_download

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน lab_data สำหรับดาวน์โหลดไฟล์ข้อมูลของรายวิชา
#   - def คือคำสั่งสร้างฟังก์ชันใหม่ ตามด้วยชื่อฟังก์ชันและพารามิเตอร์ในวงเล็บ
#   - path : พารามิเตอร์ (ค่าที่ผู้ใช้ส่งเข้ามา) คือชื่อไฟล์ที่ต้องการ เช่น "pima/test_ids.csv"
#   - บรรทัดที่ย่อหน้าเข้าไปด้านล่าง คือคำสั่งที่อยู่ภายในฟังก์ชัน
# ----------------------------------------------------------------------------
def lab_data(path):
    """ดาวน์โหลดไฟล์ข้อมูลทดสอบสาธารณะของรายวิชา 1 ไฟล์ (เช่น 'pima/test_ids.csv') และคืนค่าตำแหน่งไฟล์ในเครื่อง"""
    # เรียก hf_hub_download(...) เพื่อดาวน์โหลดไฟล์ แล้วส่งตำแหน่งไฟล์ในเครื่องกลับไป (return)
    #   - DATA_REPO           : ชื่อคลังข้อมูลบน Hugging Face ที่เก็บไฟล์
    #   - path                : ชื่อไฟล์ที่ต้องการดาวน์โหลดภายในคลังข้อมูลนั้น
    #   - repo_type="dataset" : ระบุว่าคลังนี้เป็นประเภท "ชุดข้อมูล" (ไม่ใช่โมเดล)
    return hf_hub_download(DATA_REPO, path, repo_type="dataset")

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน submit_to_leaderboard สำหรับส่งไฟล์ผลการทำนายไปยังตารางอันดับ
#   - path   : ชื่อไฟล์ผลการทำนายที่ต้องการส่ง
#   - lab=LAB_ID : รหัสปฏิบัติการ หากไม่ระบุ จะใช้ค่า LAB_ID ที่กำหนดไว้ด้านบนโดยอัตโนมัติ
# ----------------------------------------------------------------------------
def submit_to_leaderboard(path, lab=LAB_ID):
    """ส่งไฟล์ผลการทำนายไปยังตารางอันดับของรายวิชา และแสดงคะแนนพร้อมอันดับ"""
    # นำเข้าเครื่องมือ 2 ตัวจากไลบรารี gradio_client
    #   - Client      : ใช้เชื่อมต่อกับเว็บแอปตารางอันดับ
    #   - handle_file : ใช้เตรียมไฟล์ในเครื่องให้อยู่ในรูปแบบที่ส่งขึ้นเว็บแอปได้
    from gradio_client import Client, handle_file
    # assert คือการตรวจสอบเงื่อนไข หากเงื่อนไขเป็นเท็จ โปรแกรมจะหยุดและแสดงข้อความเตือน
    #   - NAME and JOIN_CODE : เงื่อนไขว่าต้องกรอกทั้ง NAME และ JOIN_CODE แล้ว (ไม่เป็นข้อความว่าง)
    #   - ข้อความหลังเครื่องหมายจุลภาค คือข้อความเตือนที่จะแสดงเมื่อยังไม่ได้กรอก
    assert NAME and JOIN_CODE, "โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่าก่อน (จำเป็นเฉพาะการส่งผลไปยังตารางอันดับ)"
    # เชื่อมต่อกับเว็บแอปตารางอันดับ แล้วเก็บการเชื่อมต่อไว้ในตัวแปร client
    #   - SPACE_ID      : ชื่อเว็บแอปตารางอันดับที่กำหนดไว้ด้านบน
    #   - verbose=False : ไม่ต้องแสดงข้อความรายละเอียดระหว่างเชื่อมต่อ
    client = Client(SPACE_ID, verbose=False)
    # ส่งข้อมูลไปยังเว็บแอปด้วย client.predict(...) แล้วแสดงผลลัพธ์ (คะแนนและอันดับ) ด้วย print(...)
    #   - lab=lab                 : รหัสปฏิบัติการ
    #   - name=NAME               : ชื่อที่จะแสดงบนตารางอันดับ
    #   - code=JOIN_CODE          : รหัสเข้าร่วม
    #   - file=handle_file(path)  : ไฟล์ผลการทำนายที่เตรียมพร้อมสำหรับส่ง
    #   - api_name="/submit"      : ชื่อช่องทางรับข้อมูลของเว็บแอป (ส่วน "ส่งผล")
    print(client.predict(lab=lab, name=NAME, code=JOIN_CODE, file=handle_file(path), api_name="/submit"))

# แสดงข้อความยืนยันว่าการตั้งค่าเสร็จแล้ว ด้วยฟังก์ชัน print(...) (พิมพ์ข้อความออกทางหน้าจอ)
#   - ข้อความเป็น f-string จึงแทรกค่า SPACE_ID, LAB_ID และ NAME ลงในข้อความได้
#   - NAME or '(ยังไม่ได้ระบุ ...)' : หาก NAME ว่าง จะแสดงข้อความในวงเล็บแทน
print(f"Leaderboard: https://huggingface.co/spaces/{SPACE_ID}  |  lab={LAB_ID}  name={NAME or '(ยังไม่ได้ระบุ — ตารางอันดับไม่บังคับ)'}")
```

```python
# @title 🧰 ฟังก์ชันช่วยสำหรับเซลล์เฉลย — ให้รันเซลล์นี้
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้สร้าง "ฟังก์ชันช่วย" ชื่อ is_blank(...) ไว้ 1 ตัว เพื่อให้เซลล์ ✅ เฉลย
#   ในโน้ตบุ๊กนี้ใช้ตรวจสอบว่า ผู้เรียนกรอกค่าในแบบฝึกหัดแล้วหรือยัง
#     - หากผู้เรียน "ยังไม่ได้กรอก" (ยังเป็น ... หรือว่างเปล่า) เซลล์เฉลยจะเติมค่าตัวอย่างให้
#     - หากผู้เรียน "กรอกแล้ว" เซลล์เฉลยจะคงค่าของผู้เรียนไว้ ไม่เขียนทับ
#   ข้อมูลเข้า : ไม่มี (เพียงสร้างฟังก์ชันไว้ใช้ภายหลัง)
#   ผลลัพธ์   : ข้อความ "ฟังก์ชันช่วยพร้อมใช้งาน" แสดงใต้เซลล์
#   ผู้เรียนเพียงกดรันเซลล์นี้ 1 ครั้ง ไม่จำเป็นต้องแก้ไขโค้ด
# ============================================================================

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน is_blank สำหรับตรวจว่า "ค่านี้ยังว่างอยู่หรือไม่"
#   - def คือคำสั่งสร้างฟังก์ชันใหม่ (ชุดคำสั่งที่ตั้งชื่อไว้และเรียกใช้ซ้ำได้)
#   - value : พารามิเตอร์ (ค่าที่ส่งเข้ามาตรวจ) เช่น ค่าของตัวแปร TRACK
#   - ข้อความในเครื่องหมาย """ """ ใต้บรรทัด def เรียกว่า docstring
#     เป็นคำอธิบายการทำงานของฟังก์ชัน (ไม่มีผลต่อการทำงาน)
# ----------------------------------------------------------------------------
def is_blank(value):
    """คืนค่า True หากค่าตั้งยังว่างอยู่ (`...`, None หรือ "") เซลล์เฉลยจะเติมเฉพาะค่าที่ยังว่างเท่านั้น"""
    # return คือการส่งผลลัพธ์กลับไปยังผู้เรียกใช้ฟังก์ชัน ผลลัพธ์ที่นี่เป็น True (จริง) หรือ False (เท็จ)
    #   เงื่อนไขย่อย 3 ข้อเชื่อมด้วย or (หรือ) — หากข้อใดข้อหนึ่งเป็นจริง ผลลัพธ์คือ True
    #   - value is ...      : ค่าเป็นเครื่องหมาย ... (Ellipsis) ซึ่งใช้เป็น "ช่องว่างให้เติม" ในแบบฝึกหัด
    #                          (is คือการตรวจว่าเป็น "สิ่งเดียวกัน" ทุกประการ)
    #   - value is None     : ค่าเป็น None (ค่าว่างพิเศษของ Python หมายถึง "ไม่มีค่า")
    #   - (isinstance(value, str) and value.strip() in ("", "...")) :
    #       * isinstance(value, str) : ฟังก์ชันในตัวของ Python ตรวจว่า value เป็นข้อความ (str) หรือไม่
    #       * and                    : ต้องจริงทั้งสองด้าน
    #       * value.strip()          : ตัดช่องว่างหน้า-หลังข้อความออก
    #       * in ("", "...")         : ตรวจว่าข้อความที่ตัดแล้วเป็นข้อความว่าง "" หรือข้อความ "..." หรือไม่
    return value is ... or value is None or (isinstance(value, str) and value.strip() in ("", "..."))

# แสดงข้อความยืนยันว่าสร้างฟังก์ชันเสร็จแล้ว ด้วยฟังก์ชัน print(...) (พิมพ์ข้อความออกทางหน้าจอ)
print("ฟังก์ชันช่วยพร้อมใช้งาน")
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (เฉลยแบบฝึกหัดที่ 1)
#   ตรวจสอบค่า TRACK ที่ผู้เรียนกรอกไว้ในแบบฝึกหัดที่ 1
#     - หากเป็นหนึ่งในแนวทางที่ถูกต้อง ("tabular", "llm", "image", "nocode")
#       จะคงค่าที่ผู้เรียนเลือกไว้ (แปลงเป็นตัวพิมพ์เล็กและตัดช่องว่างให้)
#     - หากยังไม่ได้กรอก หรือกรอกผิด จะกำหนดเป็น "tabular" ซึ่งเร็วที่สุด
#       และรันบน CPU ของ Colab รุ่นฟรีได้
#   ผลลัพธ์ : ข้อความ "แนวทางที่เลือก: ..." แสดงใต้เซลล์
# ============================================================================

# สร้างตัวแปร VALID_TRACKS เก็บรายชื่อแนวทางที่ถูกต้องทั้ง 4 แนวทาง
#   - ( ... , ... ) ในวงเล็บธรรมดาเรียกว่า tuple คือรายการหลายค่าที่แก้ไขไม่ได้
VALID_TRACKS = ("tabular", "llm", "image", "nocode")
# อ่านค่า TRACK ที่ผู้เรียนกรอก แล้วปรับให้อยู่ในรูปมาตรฐาน เก็บไว้ในตัวแปร _choice
#   (ชื่อตัวแปรขึ้นต้นด้วย _ เป็นธรรมเนียมว่าเป็นตัวแปรชั่วคราวภายในเซลล์)
#   บรรทัดนี้เขียนแบบ "A if เงื่อนไข else B" คือ ถ้าเงื่อนไขจริงได้ A มิฉะนั้นได้ B
#   - globals()                : ฟังก์ชันในตัวของ Python คืนค่าตัวแปรทั้งหมดที่สร้างไว้ในโน้ตบุ๊ก (เป็น dict)
#   - .get("TRACK")            : ดึงค่าของตัวแปรชื่อ "TRACK" ถ้ายังไม่เคยสร้างตัวแปรนี้จะได้ None (ไม่เกิดข้อผิดพลาด)
#   - isinstance(..., str)     : ตรวจว่าค่าที่ได้เป็นข้อความ (str) หรือไม่
#   - TRACK.strip()            : ตัดช่องว่างหน้า-หลังออก
#   - .lower()                 : แปลงเป็นตัวพิมพ์เล็กทั้งหมด เช่น "Tabular" -> "tabular"
#   - else None                : หาก TRACK ไม่ใช่ข้อความ (เช่น ยังเป็น ...) ให้ _choice เป็น None
_choice = TRACK.strip().lower() if isinstance(globals().get("TRACK"), str) else None
# if คือคำสั่งตรวจเงื่อนไข: ถ้า _choice อยู่ใน (in) รายชื่อแนวทางที่ถูกต้อง ให้ทำคำสั่งที่ย่อหน้าด้านล่าง
if _choice in VALID_TRACKS:
    # คงค่าที่ผู้เรียนเลือกไว้ (ในรูปแบบที่ปรับให้เป็นมาตรฐานแล้ว)
    TRACK = _choice                       # คงค่าที่ผู้เรียนเลือกไว้
# else คือกรณี "มิฉะนั้น" (เงื่อนไขด้านบนเป็นเท็จ — ยังไม่ได้เลือก หรือพิมพ์ผิด)
else:
    # กำหนดแนวทางเป็น "tabular" แทน
    TRACK = "tabular"                     # แนวทางที่เร็วที่สุด รันบน CPU รุ่นฟรีได้
    # แจ้งผู้เรียนว่าเซลล์เฉลยได้เลือกแนวทางให้แล้ว
    print("ยังไม่ได้เลือกแนวทางที่ถูกต้อง เซลล์เฉลยจึงกำหนดเป็น 'tabular'")
# แสดงแนวทางที่ใช้จริงในโน้ตบุ๊กนี้
print("แนวทางที่เลือก:", TRACK)
```

```python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (เฉลยแบบฝึกหัดที่ 2A)
#   - หากผู้เรียนยังไม่ได้เลือกชุดข้อมูล จะกำหนด TABULAR_DATASET = "pima"
#     (ชุดข้อมูลเบาหวาน Pima ขนาดเล็ก ฝึกโมเดลได้เร็ว)
#   - หากตัวแปร MY_CSV_URL, MY_TARGET, MY_POSITIVE_VALUE ยังไม่เคยถูกสร้าง
#     (เช่น ผู้เรียนข้ามเซลล์แบบฝึกหัด) จะสร้างให้ด้วยค่าเริ่มต้น
#     เพื่อไม่ให้เซลล์ถัดไปเกิดข้อผิดพลาด
#   ผลลัพธ์ : ข้อความ "ชุดข้อมูล tabular: ..." แสดงใต้เซลล์
# ============================================================================

# ตรวจว่า TABULAR_DATASET ยังว่างอยู่หรือไม่ ด้วยฟังก์ชัน is_blank(...) ที่สร้างไว้ในเซลล์ฟังก์ชันช่วย
#   - globals().get("TABULAR_DATASET") : ดึงค่าตัวแปรชื่อนี้ หากยังไม่เคยสร้างจะได้ None (ซึ่งนับว่า "ว่าง")
if is_blank(globals().get("TABULAR_DATASET")):
    # หากว่าง: กำหนดให้ใช้ชุดข้อมูล "pima"
    TABULAR_DATASET = "pima"              # ขนาดเล็กและเร็ว ผู้เรียนสามารถทดลอง "heart" หรือ "readmission" ต่อได้
# for คือคำสั่งวนซ้ำ (loop): ทำคำสั่งที่ย่อหน้าด้านล่างซ้ำกับสมาชิกทีละตัวในรายการ
#   - [ ... ] : list (รายการ) ที่มีสมาชิก 3 ตัว แต่ละตัวเป็นคู่ (ชื่อตัวแปร, ค่าเริ่มต้น)
#   - _name, _default : ในแต่ละรอบ แยกคู่ค่าออกเป็น 2 ตัวแปร คือชื่อตัวแปร และค่าเริ่มต้น
#     เช่น รอบแรก _name = "MY_CSV_URL", _default = ""
for _name, _default in [("MY_CSV_URL", ""), ("MY_TARGET", ""), ("MY_POSITIVE_VALUE", 1)]:
    # ตรวจว่า "ยังไม่มี" (not in) ตัวแปรชื่อนี้ในโน้ตบุ๊กหรือไม่
    if _name not in globals():
        # สร้างตัวแปรชื่อ _name ให้มีค่า _default
        #   - globals()[_name] = ... : การกำหนดค่าผ่าน dict ของตัวแปรทั้งหมด ซึ่งให้ผลเหมือนเขียน MY_CSV_URL = "" โดยตรง
        globals()[_name] = _default
# แสดงชื่อชุดข้อมูลที่จะใช้
print("ชุดข้อมูล tabular:", TABULAR_DATASET)
```

```python
# @title ▶️ รันต้นแบบ tabular: โหลด → ทำความสะอาด → ฝึกโมเดล 2 แบบ → ประเมินผล (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (แนวทาง tabular)
#   เซลล์นี้คือ "ต้นแบบโมเดลทำนายความเสี่ยงจากตารางข้อมูลผู้ป่วย" ทั้งขั้นตอน (pipeline)
#   โดยนำเทคนิคจากปฏิบัติการที่ 1–3 มาใช้ซ้ำ แบ่งเป็น 5 ขั้น ได้แก่
#   1) โหลดข้อมูล ตามชุดข้อมูลที่เลือกใน TABULAR_DATASET และแปลงผลลัพธ์เป็น 0 / 1
#   2) ทำความสะอาดข้อมูล: แปลงคอลัมน์ข้อความให้เป็นตัวเลข 0/1 ที่โมเดลเข้าใจได้
#   3) แบ่งข้อมูล: กันผู้ป่วย 25% ไว้เป็น "ชุดทดสอบ (test set)" ที่โมเดลไม่เคยเห็น
#   4) ฝึกโมเดล 2 แบบ คือ การถดถอยโลจิสติก (logistic regression) และ gradient boosting
#   5) ประเมินผลด้วย AUROC ความไว (sensitivity) และความจำเพาะ (specificity)
#      เทียบกับค่าอ้างอิงอย่างง่าย (baseline) "การโยนเหรียญ" แล้ววาดเส้นโค้ง ROC
#   สุดท้ายสร้างฟังก์ชัน demo(...) ไว้สำหรับแสดงผู้ป่วย 1 รายในการสาธิตสด
#   ข้อมูลเข้า : TRACK, TABULAR_DATASET (และ MY_... หากใช้ไฟล์ของตนเอง)
#   ผลลัพธ์   : ตารางผลการประเมิน กราฟ ROC และฟังก์ชัน demo() พร้อมใช้งาน
#   หากผู้เรียนเลือกแนวทางอื่น เซลล์นี้จะแสดงข้อความแล้วข้ามการทำงานทั้งหมด
# ============================================================================

# ตรวจว่าแนวทางที่เลือก "ไม่ใช่" (!= หมายถึง ไม่เท่ากับ) "tabular" หรือไม่
if TRACK != "tabular":
    # หากไม่ใช่ ให้แสดงข้อความแจ้งว่าข้ามเซลล์นี้ (f-string แทรกค่า TRACK ลงในข้อความผ่าน { })
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
# else: กรณีเลือกแนวทาง tabular — คำสั่งทั้งหมดที่ย่อหน้าด้านล่างจะทำงาน
else:
    # ------------------------------------------------------------------------
    # นำเข้า (import) ไลบรารีที่จำเป็น — ไลบรารีคือชุดเครื่องมือสำเร็จรูปที่ผู้อื่นเขียนไว้
    # ------------------------------------------------------------------------
    # นำเข้าไลบรารีมาตรฐานของ Python 3 ตัว (คั่นด้วยจุลภาค)
    #   - io       : จัดการข้อมูลในหน่วยความจำให้ทำงานเหมือนไฟล์
    #   - zipfile  : เปิดอ่านไฟล์บีบอัด .zip
    #   - requests : ดาวน์โหลดข้อมูลจากอินเทอร์เน็ต
    import io, zipfile, requests
    # นำเข้าไลบรารีวิเคราะห์ข้อมูล พร้อมตั้งชื่อย่อด้วย as (ชื่อย่อเป็นธรรมเนียมที่ใช้กันทั่วไป)
    #   - numpy as np               : คำนวณตัวเลขและอาร์เรย์ (array)
    #   - pandas as pd              : จัดการข้อมูลแบบตาราง (DataFrame) คล้าย Excel
    #   - matplotlib.pyplot as plt  : วาดกราฟ
    import numpy as np, pandas as pd, matplotlib.pyplot as plt
    # นำเข้าเครื่องมือแบ่งข้อมูลจากไลบรารี scikit-learn (sklearn) ส่วน model_selection
    #   - train_test_split  : แบ่งข้อมูลเป็นชุดฝึกและชุดทดสอบแบบสุ่ม
    #   - GroupShuffleSplit : แบ่งข้อมูลแบบสุ่ม "ตามกลุ่ม" (เช่น ตามผู้ป่วย) ให้ทั้งกลุ่มอยู่ส่วนเดียวกัน
    from sklearn.model_selection import train_test_split, GroupShuffleSplit
    # make_pipeline : นำขั้นตอนหลายขั้น (เช่น เติมค่าที่หายไป → ปรับสเกล → โมเดล) มาต่อกันเป็นชุดเดียว
    from sklearn.pipeline import make_pipeline
    # SimpleImputer : เติมค่าที่หายไป (missing value) ด้วยค่าแทน เช่น ค่ามัธยฐาน
    from sklearn.impute import SimpleImputer
    # StandardScaler : ปรับสเกลตัวเลขแต่ละคอลัมน์ให้มีค่าเฉลี่ย 0 และส่วนเบี่ยงเบนมาตรฐาน 1
    from sklearn.preprocessing import StandardScaler
    # LogisticRegression : โมเดลการถดถอยโลจิสติก (logistic regression) ให้ผลเป็นความน่าจะเป็น ตีความได้ง่าย
    from sklearn.linear_model import LogisticRegression
    # HistGradientBoostingClassifier : โมเดล gradient boosting (ต้นไม้ตัดสินใจขนาดเล็กจำนวนมากทำงานร่วมกัน)
    #   รองรับค่าที่หายไป (NaN) ได้เองโดยไม่ต้องเติมค่า
    from sklearn.ensemble import HistGradientBoostingClassifier
    # เครื่องมือวัดผลจาก sklearn.metrics
    #   - roc_auc_score : คำนวณค่า AUROC (พื้นที่ใต้เส้นโค้ง ROC)
    #   - roc_curve     : คำนวณจุดต่าง ๆ บนเส้นโค้ง ROC เพื่อนำไปวาดกราฟ
    from sklearn.metrics import roc_auc_score, roc_curve

    # ------------------------------------------------------------------------
    # ขั้นที่ 1) โหลดข้อมูลตามชุดข้อมูลที่เลือก
    # ------------------------------------------------------------------------
    # 1) โหลดข้อมูล และแปลงผลลัพธ์เป็น 0 (ไม่เกิด) / 1 (เกิด)
    # สร้างตัวแปร group_col เก็บชื่อคอลัมน์รหัสผู้ป่วย เริ่มต้นเป็น None (ไม่มี)
    #   จะกำหนดค่าเฉพาะชุดข้อมูลที่ผู้ป่วย 1 รายมีหลายแถว (readmission)
    group_col = None                      # คอลัมน์ที่ระบุตัวผู้ป่วย กรณีผู้ป่วย 1 รายมีหลายแถว
    # กรณีที่ 1: ชุดข้อมูลเบาหวาน Pima (== หมายถึง "เท่ากับหรือไม่" ใช้เปรียบเทียบ ต่างจาก = ที่ใช้เก็บค่า)
    if TABULAR_DATASET == "pima":
        # อ่านไฟล์ CSV จากอินเทอร์เน็ตมาเป็นตาราง (DataFrame) แล้วเก็บไว้ในตัวแปร df
        #   - pd.read_csv(...) : ฟังก์ชันของ pandas สำหรับอ่านไฟล์ CSV (ระบุเป็นตำแหน่งไฟล์หรือ URL ก็ได้)
        #   - อาร์กิวเมนต์เดียวคือ URL ของไฟล์ข้อมูล Pima Diabetes
        df = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv")
        # สร้าง list รายชื่อคอลัมน์ที่ค่า 0 เป็นไปไม่ได้ทางสรีรวิทยา (เช่น ความดันโลหิต 0)
        #   ในชุดข้อมูลนี้ค่า 0 จึงหมายถึง "ไม่ได้บันทึก" หรือค่าที่หายไปซึ่งแฝงอยู่
        zero_as_missing = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
        # แทนค่า 0 ในคอลัมน์เหล่านั้นด้วย NaN (ค่าที่หายไป)
        #   - df[zero_as_missing]       : เลือกเฉพาะคอลัมน์ตามรายชื่อใน list
        #   - .replace(0, np.nan)       : แทนที่ค่า 0 ด้วย np.nan (Not a Number = ค่าว่างของ numpy)
        #   - นำผลลัพธ์ไปเขียนทับคอลัมน์เดิมใน df
        df[zero_as_missing] = df[zero_as_missing].replace(0, np.nan)   # BMI เท่ากับ 0 คือค่าที่หายไป (ปฏิบัติการที่ 1)
        # กำหนดชื่อคอลัมน์ผลลัพธ์ (target) ที่ต้องการทำนาย: Outcome (1 = เป็นเบาหวาน, 0 = ไม่เป็น)
        target = "Outcome"
    # elif ("else if") : กรณีที่ 2 ตรวจเมื่อกรณีก่อนหน้าไม่เป็นจริง — ชุดข้อมูลโรคหัวใจ UCI Cleveland
    elif TABULAR_DATASET == "heart":
        # สร้าง list ชื่อคอลัมน์ 14 คอลัมน์ (ไฟล์ต้นฉบับไม่มีแถวหัวตาราง จึงต้องตั้งชื่อเอง)
        #   เช่น age อายุ, sex เพศ, cp ชนิดการเจ็บหน้าอก, trestbps ความดันขณะพัก, chol คอเลสเตอรอล,
        #   thalach อัตราการเต้นหัวใจสูงสุด, target การวินิจฉัยโรคหัวใจ
        #   (list ยาวหลายบรรทัดได้ เพราะอยู่ในวงเล็บ [ ])
        cols = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg", "thalach",
                "exang", "oldpeak", "slope", "ca", "thal", "target"]
        # อ่านไฟล์ข้อมูลโรคหัวใจจากคลังข้อมูล UCI
        #   - อาร์กิวเมนต์แรก : URL (ข้อความ 2 ส่วนที่วางติดกันจะถูกต่อเป็นข้อความเดียวโดยอัตโนมัติ)
        #   - names=cols      : กำหนดชื่อคอลัมน์ตาม list cols
        #   - na_values="?"   : ให้ถือว่าเครื่องหมาย ? ในไฟล์คือค่าที่หายไป (NaN)
        df = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/"
                         "processed.cleveland.data", names=cols, na_values="?")
        # แปลงคอลัมน์ target (เดิมมีค่า 0–4 ตามระดับความรุนแรง) ให้เป็น 0/1
        #   - df["target"] > 0 : ได้ True หากมีโรค (ค่ามากกว่า 0) และ False หากไม่มีโรค
        #   - .astype(int)     : แปลง True/False เป็นตัวเลข 1/0
        df["target"] = (df["target"] > 0).astype(int)                  # มีโรคหัวใจ เทียบกับ ไม่มีโรค
        # กำหนดชื่อคอลัมน์ผลลัพธ์
        target = "target"
    # กรณีที่ 3: ชุดข้อมูลการกลับมารักษาซ้ำ (readmission) Diabetes 130-US Hospitals
    elif TABULAR_DATASET == "readmission":
        # เก็บ URL ของไฟล์ zip ชุดข้อมูลไว้ในตัวแปร URL
        URL = "https://archive.ics.uci.edu/static/public/296/diabetes+130-us+hospitals+for+years+1999-2008.zip"
        # ดาวน์โหลดไฟล์ zip แล้วเปิดอ่านในหน่วยความจำ (ไม่ต้องบันทึกลงดิสก์) ทำงานจากในสุดออกมา:
        #   - requests.get(URL, timeout=120) : ดาวน์โหลดจาก URL โดยรอได้นานสุด 120 วินาที
        #   - .content                       : เนื้อหาไฟล์ที่ดาวน์โหลดได้ (เป็นข้อมูลไบต์)
        #   - io.BytesIO(...)                : ห่อข้อมูลไบต์ให้ใช้งานได้เหมือนไฟล์
        #   - zipfile.ZipFile(...)           : เปิดเป็นไฟล์ zip เก็บไว้ในตัวแปร z
        z = zipfile.ZipFile(io.BytesIO(requests.get(URL, timeout=120).content))
        # อ่านไฟล์ diabetic_data.csv ที่อยู่ภายใน zip มาเป็นตาราง df
        #   - z.namelist()                               : รายชื่อไฟล์ทั้งหมดใน zip
        #   - (n for n in ... if n.endswith("diabetic_data.csv")) : วนหาชื่อไฟล์ที่ลงท้ายด้วย diabetic_data.csv
        #   - next(...)                                  : หยิบชื่อไฟล์แรกที่พบ
        #   - z.open(...)                                : เปิดไฟล์นั้นจากใน zip
        #   - na_values="?"                              : ถือว่า ? คือค่าที่หายไป
        #   - low_memory=False                           : ให้ pandas อ่านทั้งไฟล์ก่อนเดาชนิดข้อมูลแต่ละคอลัมน์
        #                                                  (ลดคำเตือนเรื่องชนิดข้อมูลปนกัน)
        df = pd.read_csv(z.open(next(n for n in z.namelist() if n.endswith("diabetic_data.csv"))),
                         na_values="?", low_memory=False)
        # สร้างคอลัมน์ผลลัพธ์ใหม่ readmitted_30d (1 = กลับมารักษาซ้ำภายใน 30 วัน, 0 = ไม่ใช่)
        #   - df.pop("readmitted") : ดึงคอลัมน์ readmitted ออกจากตาราง (และลบคอลัมน์เดิมทิ้ง)
        #                            ค่าเดิมมี 3 แบบ คือ "<30", ">30", "NO"
        #   - == "<30"             : ได้ True เฉพาะแถวที่กลับมาภายใน 30 วัน
        #   - .astype(int)         : แปลง True/False เป็น 1/0
        df["readmitted_30d"] = (df.pop("readmitted") == "<30").astype(int)
        # try / except : "ลองทำ" คำสั่งใน try หากเกิดข้อผิดพลาด จะไปทำคำสั่งใน except แทน (โปรแกรมไม่หยุด)
        try:   # ตัดผู้ป่วยในชุดทดสอบที่ซ่อนไว้ของตารางอันดับออก เพื่อให้การประเมินตรงไปตรงมา
            # ดาวน์โหลดไฟล์รหัสการนอนโรงพยาบาลของชุดทดสอบตารางอันดับ ด้วย lab_data(...) จากเซลล์การตั้งค่า
            #   - pd.read_csv(...) อ่านเป็นตาราง แล้ว ["id"] เลือกเฉพาะคอลัมน์ id
            hidden = pd.read_csv(lab_data("readmission/test_ids.csv"))["id"]
            # เก็บไว้เฉพาะแถวที่ "ไม่อยู่" ในชุดทดสอบที่ซ่อนไว้
            #   - df["encounter_id"].isin(hidden) : True หากรหัสการนอนโรงพยาบาลอยู่ในรายการ hidden
            #   - ~                               : กลับค่า True <-> False (คือ "ไม่อยู่ใน")
            #   - df[ ... ]                       : เลือกเฉพาะแถวที่เงื่อนไขเป็น True
            df = df[~df["encounter_id"].isin(hidden)]
        # หากดาวน์โหลดไม่สำเร็จ: เก็บข้อผิดพลาดไว้ในตัวแปร e (Exception = ข้อผิดพลาดทุกประเภท)
        except Exception as e:
            # แจ้งเตือนและแสดงชื่อประเภทข้อผิดพลาด (type(e).__name__) แล้วใช้ข้อมูลทั้งหมดต่อไป
            print("หมายเหตุ: ไม่สามารถดาวน์โหลดรหัสชุดทดสอบของตารางอันดับได้ จึงใช้ข้อมูลทั้งหมด", type(e).__name__)
        # ลบคอลัมน์ encounter_id (รหัสการนอนโรงพยาบาล) ออก เพราะเป็นเพียงรหัส ไม่ใช่ข้อมูลทางคลินิก
        #   - df.drop(columns=[...]) : ลบคอลัมน์ตามรายชื่อที่ระบุ
        df = df.drop(columns=["encounter_id"])
        # กำหนดค่า 2 ตัวแปรพร้อมกันในบรรทัดเดียว (ค่าทางขวาจับคู่กับชื่อทางซ้ายตามลำดับ)
        #   - target    = "readmitted_30d" : คอลัมน์ผลลัพธ์
        #   - group_col = "patient_nbr"    : คอลัมน์รหัสผู้ป่วย ใช้แบ่งข้อมูลตามผู้ป่วยในขั้นที่ 3
        target, group_col = "readmitted_30d", "patient_nbr"
    # กรณีที่ 4: ไฟล์ CSV สาธารณะของผู้เรียนเอง
    elif TABULAR_DATASET == "my_csv":
        # อ่านไฟล์ CSV จากลิงก์ที่ผู้เรียนระบุใน MY_CSV_URL
        df = pd.read_csv(MY_CSV_URL)
        # ตัดแถวที่คอลัมน์ผลลัพธ์ว่างออก
        #   - df[MY_TARGET].notna() : True หากค่าในคอลัมน์ผลลัพธ์ "ไม่ว่าง"
        df = df[df[MY_TARGET].notna()]
        # เก็บคอลัมน์ผลลัพธ์ไว้ในตัวแปร col เพื่อเรียกใช้สั้นลง
        col = df[MY_TARGET]
        # ตรวจว่าคอลัมน์ผลลัพธ์เป็นตัวเลข และ MY_POSITIVE_VALUE ก็เป็นตัวเลขด้วยหรือไม่
        #   - pd.api.types.is_numeric_dtype(col)          : ฟังก์ชันของ pandas ตรวจว่าคอลัมน์เป็นชนิดตัวเลข
        #   - isinstance(MY_POSITIVE_VALUE, (int, float)) : ตรวจว่าเป็นจำนวนเต็ม (int) หรือทศนิยม (float)
        if pd.api.types.is_numeric_dtype(col) and isinstance(MY_POSITIVE_VALUE, (int, float)):
            # กรณีตัวเลข: แถวที่ค่าเท่ากับ MY_POSITIVE_VALUE ได้ 1 ที่เหลือได้ 0
            df[MY_TARGET] = (col == MY_POSITIVE_VALUE).astype(int)
        # กรณีอื่น (เช่น คอลัมน์เป็นข้อความ "Yes"/"No"): เปรียบเทียบแบบข้อความ
        else:
            # แปลงทั้งสองฝั่งให้เป็นข้อความ ตัดช่องว่าง และเป็นตัวพิมพ์เล็กก่อนเปรียบเทียบ
            #   - col.astype(str)        : แปลงทุกค่าในคอลัมน์เป็นข้อความ
            #   - .str.strip()           : ตัดช่องว่างหน้า-หลังของทุกค่า (.str ใช้เรียกคำสั่งข้อความกับทั้งคอลัมน์)
            #   - .str.lower()           : แปลงเป็นตัวพิมพ์เล็ก
            #   - str(MY_POSITIVE_VALUE).strip().lower() : ทำแบบเดียวกันกับค่าที่ถือว่าเป็นผลบวก
            #   - == ... แล้ว .astype(int) : ได้ 1 หากตรงกัน 0 หากไม่ตรง
            df[MY_TARGET] = (col.astype(str).str.strip().str.lower() == str(MY_POSITIVE_VALUE).strip().lower()).astype(int)
        # กำหนดชื่อคอลัมน์ผลลัพธ์
        target = MY_TARGET
    # กรณีชื่อชุดข้อมูลไม่ตรงกับตัวเลือกใดเลย
    else:
        # raise คือการสั่งให้เกิดข้อผิดพลาดและหยุดเซลล์ พร้อมข้อความอธิบาย
        #   - ValueError        : ประเภทข้อผิดพลาด "ค่าไม่ถูกต้อง"
        #   - {TABULAR_DATASET!r} : แทรกค่าใน f-string โดย !r แสดงในรูปที่มีเครื่องหมายคำพูด เช่น 'piam'
        raise ValueError(f"ไม่รู้จัก TABULAR_DATASET {TABULAR_DATASET!r} โปรดใช้ 'pima', 'heart', 'readmission' หรือ 'my_csv'")
    # จัดเลขแถว (index) ใหม่ให้เรียง 0, 1, 2, ... หลังจากตัดแถวออกไป
    #   - drop=True : ไม่ต้องเก็บเลขแถวเดิมไว้เป็นคอลัมน์ใหม่
    df = df.reset_index(drop=True)
    # ดึงคอลัมน์ผลลัพธ์ออกมาเป็นตัวแปร y และแปลงเป็นจำนวนเต็ม (0/1)
    y = df[target].astype(int)
    # แสดงขนาดข้อมูลและสัดส่วนผลบวก ด้วย f-string และรหัสจัดรูปแบบ (format code) หลังเครื่องหมาย :
    #   - len(df)        : จำนวนแถว; :, คือใส่จุลภาคคั่นหลักพัน เช่น 81,412
    #   - df.shape[1]    : จำนวนคอลัมน์ (shape = (แถว, คอลัมน์)) ลบ 1 เพราะไม่นับคอลัมน์ผลลัพธ์
    #   - y.mean()       : ค่าเฉลี่ยของ 0/1 คือสัดส่วนผลบวก; :.1% คือแสดงเป็นร้อยละทศนิยม 1 ตำแหน่ง
    print(f"ข้อมูล {len(df):,} แถว {df.shape[1] - 1} คอลัมน์; ผลลัพธ์ '{target}' เป็น 'ใช่' {y.mean():.1%} ของแถวทั้งหมด")
    # assert : ตรวจเงื่อนไข หากเป็นเท็จจะหยุดพร้อมข้อความเตือน
    #   - 0 < y.mean() < 1 : ต้องมีทั้งผลบวกและผลลบ (ถ้ามีค่าเดียว โมเดลเรียนรู้ไม่ได้)
    assert 0 < y.mean() < 1, "คอลัมน์ผลลัพธ์มีค่าเพียงค่าเดียว โปรดตรวจสอบ MY_TARGET / MY_POSITIVE_VALUE"

    # ------------------------------------------------------------------------
    # ขั้นที่ 2) ทำความสะอาดข้อมูลและเตรียมตัวแปรทำนาย (features)
    # ------------------------------------------------------------------------
    # 2) ทำความสะอาด: คอลัมน์ตัวเลขคงเดิม คอลัมน์ข้อความแปลงเป็นคอลัมน์ 0/1 (15 ค่าที่พบบ่อยที่สุด ที่เหลือเป็น 'other')
    # สร้างตาราง X_raw = ข้อมูลทุกคอลัมน์ ยกเว้นคอลัมน์ผลลัพธ์และคอลัมน์รหัสผู้ป่วย (ห้ามให้โมเดลเห็นคำตอบ)
    #   - [target] + ([group_col] if group_col else []) : list ชื่อคอลัมน์ที่จะลบ
    #     (+ ต่อ list สองชุดเข้าด้วยกัน; ถ้าไม่มี group_col จะต่อด้วย list ว่าง [])
    X_raw = df.drop(columns=[target] + ([group_col] if group_col else []))
    # เลือกเฉพาะคอลัมน์ที่เป็นตัวเลข เก็บใน num
    #   - .select_dtypes("number") : เลือกคอลัมน์ตามชนิดข้อมูล โดย "number" คือชนิดตัวเลขทุกแบบ
    num = X_raw.select_dtypes("number")
    # เลือกคอลัมน์ที่ "ไม่ใช่" ตัวเลข (ข้อความ/หมวดหมู่) แล้วเตรียมให้พร้อมใช้ เก็บใน cat
    #   - .select_dtypes(exclude="number") : เลือกคอลัมน์ที่ไม่ใช่ตัวเลข
    #   - .fillna("missing")               : เติมช่องว่างด้วยข้อความ "missing" (ให้ "ค่าหายไป" เป็นหมวดหนึ่ง)
    #   - .astype(str)                     : แปลงทุกค่าเป็นข้อความ
    cat = X_raw.select_dtypes(exclude="number").fillna("missing").astype(str)
    # วนซ้ำทีละคอลัมน์ข้อความ (c คือชื่อคอลัมน์ในแต่ละรอบ)
    for c in cat.columns:
        # หา 15 ค่าที่พบบ่อยที่สุดในคอลัมน์นี้
        #   - cat[c].value_counts() : นับจำนวนครั้งของแต่ละค่า เรียงจากมากไปน้อย
        #   - .index                : รายชื่อค่า (ไม่เอาจำนวนนับ)
        #   - [:15]                 : เลือก 15 ตัวแรก (การตัดช่วง slicing ตั้งแต่ต้นถึงก่อนตำแหน่ง 15)
        top = cat[c].value_counts().index[:15]
        # ค่าที่ไม่อยู่ใน 15 อันดับแรกให้รวมเป็น "other" เพื่อไม่ให้มีคอลัมน์มากเกินไป
        #   - cat[c].isin(top)        : True หากค่านั้นอยู่ใน top
        #   - .where(เงื่อนไข, "other") : คงค่าเดิมเมื่อเงื่อนไขจริง มิฉะนั้นแทนด้วย "other"
        cat[c] = cat[c].where(cat[c].isin(top), "other")
    # รวมเป็นตารางตัวแปรทำนาย X ที่เป็นตัวเลขทั้งหมด
    #   - pd.get_dummies(cat, dtype=int) : การเข้ารหัสแบบ one-hot คือแปลงแต่ละค่าในคอลัมน์ข้อความเป็นคอลัมน์ใหม่ 0/1
    #                                      เช่น sex -> sex_Male, sex_Female; dtype=int ให้เป็นตัวเลข 0/1
    #   - pd.concat([num, ...], axis=1)  : นำตารางมาต่อกัน "ด้านข้าง" (axis=1 = ต่อคอลัมน์)
    #   - ... if cat.shape[1] else num.copy() : ถ้ามีคอลัมน์ข้อความอย่างน้อย 1 คอลัมน์ให้ต่อกัน
    #                                           มิฉะนั้นใช้สำเนา (copy) ของคอลัมน์ตัวเลขอย่างเดียว
    X = pd.concat([num, pd.get_dummies(cat, dtype=int)], axis=1) if cat.shape[1] else num.copy()

    # ------------------------------------------------------------------------
    # ขั้นที่ 3) แบ่งข้อมูลเป็นชุดฝึก (train) 75% และชุดทดสอบ (test) 25%
    # ------------------------------------------------------------------------
    # 3) แบ่งข้อมูล: กันผู้ป่วย 25% ไว้ไม่ใช้ฝึก และใช้เฉพาะการประเมินผลขั้นสุดท้าย
    # หากมีคอลัมน์รหัสผู้ป่วย (ชุดข้อมูล readmission) ให้แบ่งตามผู้ป่วย
    if group_col:   # ผู้ป่วยมีหลายแถว -> แบ่งตามผู้ป่วย เพื่อไม่ให้ผู้ป่วยรายเดียวกันอยู่ทั้งสองส่วน
        # แบ่งข้อมูลตามกลุ่มผู้ป่วยด้วย GroupShuffleSplit ได้ตำแหน่งแถวของชุดฝึก (tr) และชุดทดสอบ (te)
        #   - n_splits=1             : แบ่งเพียง 1 ครั้ง
        #   - test_size=0.25         : ให้ชุดทดสอบเป็น 25% ของกลุ่ม (ผู้ป่วย)
        #   - random_state=42        : กำหนดค่าเริ่มต้นการสุ่ม ให้ได้ผลเหมือนเดิมทุกครั้งที่รัน
        #   - .split(X, y, groups=df[group_col]) : สั่งแบ่ง โดยใช้รหัสผู้ป่วยเป็นกลุ่ม
        #   - next(...)              : หยิบผลการแบ่งครั้งแรก (ซึ่งมีครั้งเดียว) ออกมาเป็นคู่ (tr, te)
        tr, te = next(GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42).split(X, y, groups=df[group_col]))
    # กรณีผู้ป่วย 1 รายมี 1 แถว: แบ่งแบบสุ่มปกติ
    else:
        # แบ่งตำแหน่งแถวด้วย train_test_split
        #   - np.arange(len(X))  : ลำดับเลข 0, 1, 2, ... ถึงจำนวนแถว - 1 (ตำแหน่งของทุกแถว)
        #   - test_size=0.25     : ชุดทดสอบ 25%
        #   - stratify=y         : ให้สัดส่วนผลบวก/ผลลบในชุดฝึกและชุดทดสอบใกล้เคียงกับข้อมูลทั้งหมด
        #   - random_state=42    : ให้ผลการสุ่มเหมือนเดิมทุกครั้ง
        tr, te = train_test_split(np.arange(len(X)), test_size=0.25, stratify=y, random_state=42)
    # ใช้ตำแหน่งแถว tr / te เลือกข้อมูลออกมาเป็น 4 ส่วน (.iloc[...] = เลือกแถวตามตำแหน่ง)
    #   - X_train, y_train : ตัวแปรทำนายและผลลัพธ์ของชุดฝึก
    #   - X_test,  y_test  : ตัวแปรทำนายและผลลัพธ์ของชุดทดสอบ
    X_train, X_test, y_train, y_test = X.iloc[tr], X.iloc[te], y.iloc[tr], y.iloc[te]
    # เก็บข้อมูลดิบ (ก่อนแปลงเป็นตัวเลข) ของชุดทดสอบไว้ใช้แสดงผลที่อ่านง่ายในฟังก์ชัน demo
    raw_test = df.iloc[te]
    # กำหนดเกณฑ์ตัดสิน (threshold) = สัดส่วนผลบวกในชุดฝึก (ความเสี่ยงเฉลี่ย)
    #   ผู้ป่วยที่ความเสี่ยงที่ทำนายได้ ≥ ค่านี้ จะถูกระบุเป็นกลุ่มเสี่ยง
    threshold = y_train.mean()   # ระบุผู้ป่วยเป็นกลุ่มเสี่ยงเมื่อความเสี่ยงที่ทำนายสูงกว่าความเสี่ยงเฉลี่ย

    # ------------------------------------------------------------------------
    # ขั้นที่ 4–5) ฝึกโมเดล 2 แบบ และประเมินผลกับชุดทดสอบ
    # ------------------------------------------------------------------------
    # 4) ฝึกโมเดล 2 แบบ และ 5) ประเมินผลกับผู้ป่วยที่กันไว้
    # สร้าง dict (พจนานุกรม { ชื่อ: ค่า }) ชื่อ models เก็บโมเดล 2 แบบ โดยใช้ชื่อโมเดลเป็นกุญแจ (key)
    models = {
        # โมเดลที่ 1: ต่อขั้นตอน 3 ขั้นเป็นชุดเดียวด้วย make_pipeline(...)
        #   - SimpleImputer(strategy="median") : เติมค่าที่หายไปด้วยค่ามัธยฐาน (median) ของแต่ละคอลัมน์
        #   - StandardScaler()                 : ปรับสเกลตัวเลขให้อยู่ในระดับเดียวกัน
        #   - LogisticRegression(max_iter=2000): การถดถอยโลจิสติก; max_iter คือจำนวนรอบคำนวณสูงสุด
        #                                        (ตั้งไว้สูงเพื่อให้คำนวณจนลู่เข้า ไม่เกิดคำเตือน)
        "Logistic regression": make_pipeline(SimpleImputer(strategy="median"), StandardScaler(),
                                             LogisticRegression(max_iter=2000)),
        # โมเดลที่ 2: gradient boosting; random_state=42 ให้ผลเหมือนเดิมทุกครั้งที่รัน
        "Gradient boosting": HistGradientBoostingClassifier(random_state=42),
    }
    # สร้างตัวแปร 2 ตัวพร้อมกัน
    #   - rows       = [] : list ว่าง ไว้เก็บผลการประเมินของแต่ละโมเดล (1 แถวต่อ 1 โมเดล)
    #   - test_probs = {} : dict ว่าง ไว้เก็บความน่าจะเป็นที่แต่ละโมเดลทำนายกับชุดทดสอบ
    rows, test_probs = [], {}
    # วนซ้ำทีละโมเดล
    #   - models.items() : ได้คู่ (ชื่อโมเดล, ตัวโมเดล) ทีละคู่ เก็บใน name และ m
    for name, m in models.items():
        # ฝึกโมเดล (fit) ด้วยข้อมูลชุดฝึก: X_train = ตัวแปรทำนาย, y_train = คำตอบที่ถูกต้อง
        m.fit(X_train, y_train)
        # ให้โมเดลทำนายความน่าจะเป็นกับชุดทดสอบ
        #   - m.predict_proba(X_test) : ได้ตารางความน่าจะเป็น 2 คอลัมน์ [P(ไม่เกิด), P(เกิด)]
        #   - [:, 1]                  : เลือกทุกแถว (:) ของคอลัมน์ที่ 1 คือความน่าจะเป็นที่ "เกิดเหตุการณ์"
        p = m.predict_proba(X_test)[:, 1]
        # เก็บความน่าจะเป็นไว้ใน dict โดยใช้ชื่อโมเดลเป็นกุญแจ (ไว้วาดกราฟ ROC ภายหลัง)
        test_probs[name] = p
        # ระบุผู้ป่วยกลุ่มเสี่ยง: True เมื่อความน่าจะเป็น >= เกณฑ์ตัดสิน
        flag = p >= threshold
        # คำนวณตัวชี้วัด แล้วเพิ่มเป็น 1 แถว (dict) ต่อท้าย list rows ด้วย .append(...)
        #   - "model"       : ชื่อโมเดล
        #   - "AUROC"       : roc_auc_score(คำตอบจริง, ความน่าจะเป็น) — 0.5 = สุ่ม, 1.0 = สมบูรณ์
        #   - "sensitivity" : ความไว = จำนวนที่ระบุถูกว่าเกิด (flag & y_test == 1) ÷ จำนวนที่เกิดจริงทั้งหมด
        #                     (& = "และ" ทีละแถว; .sum() นับจำนวน True)
        #   - "specificity" : ความจำเพาะ = จำนวนที่ไม่ถูกระบุและไม่เกิดจริง (~flag & y_test == 0) ÷ จำนวนที่ไม่เกิดทั้งหมด
        rows.append({"model": name,
                     "AUROC": roc_auc_score(y_test, p),
                     "sensitivity": (flag & (y_test == 1)).sum() / (y_test == 1).sum(),
                     "specificity": (~flag & (y_test == 0)).sum() / (y_test == 0).sum()})
    # เพิ่มแถวค่าอ้างอิงอย่างง่าย (baseline) "การโยนเหรียญ" ซึ่งทุกค่าเท่ากับ 0.5 ไว้เปรียบเทียบ
    rows.append({"model": "Baseline: coin flip", "AUROC": 0.5, "sensitivity": 0.5, "specificity": 0.5})
    # แปลง list ผลลัพธ์เป็นตาราง
    #   - pd.DataFrame(rows)   : สร้างตารางจาก list ของ dict (1 dict = 1 แถว)
    #   - .set_index("model")  : ใช้คอลัมน์ model เป็นชื่อแถว
    #   - .round(3)            : ปัดเศษทศนิยม 3 ตำแหน่ง
    results = pd.DataFrame(rows).set_index("model").round(3)
    # หาชื่อโมเดลที่ AUROC สูงสุด (ไม่นับแถว baseline)
    #   - .drop(index="Baseline: coin flip") : ตัดแถว baseline ออก
    #   - ["AUROC"]                          : เลือกคอลัมน์ AUROC
    #   - .idxmax()                          : คืนชื่อแถวที่มีค่ามากที่สุด
    best_name = results.drop(index="Baseline: coin flip")["AUROC"].idxmax()
    # ดึงตัวโมเดลที่ดีที่สุดจาก dict models ด้วยชื่อ (การเข้าถึง dict ด้วย [กุญแจ])
    best_model = models[best_name]
    # แสดงขนาดชุดทดสอบและเกณฑ์ตัดสิน
    #   - \n          : ขึ้นบรรทัดใหม่
    #   - :,          : คั่นหลักพัน
    #   - :.0%        : แสดงเป็นร้อยละไม่มีทศนิยม เช่น 0.35 -> 35%
    print(f"\nชุดทดสอบ: {len(y_test):,} แถวที่โมเดลไม่เคยเห็น เกณฑ์ตัดสิน (threshold) = ความเสี่ยงที่ทำนาย {threshold:.0%}\n")
    # แสดงตารางผลการประเมินทั้งหมด (.to_string() แปลงตารางเป็นข้อความเพื่อแสดงครบทุกแถว)
    print(results.to_string())
    # แสดงชื่อโมเดลที่ดีที่สุดและค่า AUROC
    #   - results.loc[best_name, 'AUROC'] : เลือกค่าด้วยชื่อแถวและชื่อคอลัมน์ (.loc)
    #   - :.3f                            : แสดงทศนิยม 3 ตำแหน่ง
    print(f"\nโมเดลที่ดีที่สุด: {best_name} (AUROC {results.loc[best_name, 'AUROC']:.3f})")

    # ------------------------------------------------------------------------
    # วาดกราฟเส้นโค้ง ROC (ROC curve) ของทั้ง 2 โมเดล เทียบกับเส้นการโยนเหรียญ
    # ------------------------------------------------------------------------
    # สร้างพื้นที่วาดกราฟใหม่; figsize=(5, 4) คือกว้าง 5 นิ้ว สูง 4 นิ้ว
    plt.figure(figsize=(5, 4))
    # วนซ้ำทีละโมเดล: name = ชื่อโมเดล, p = ความน่าจะเป็นที่ทำนายไว้
    for name, p in test_probs.items():
        # คำนวณจุดบนเส้นโค้ง ROC ด้วย roc_curve(คำตอบจริง, ความน่าจะเป็น) ได้ 3 ค่า
        #   - fpr : อัตราผลบวกลวง (false positive rate) = 1 - ความจำเพาะ
        #   - tpr : อัตราผลบวกจริง (true positive rate) = ความไว
        #   - _   : เกณฑ์ตัดสินแต่ละจุด (ไม่ใช้ จึงตั้งชื่อเป็น _ ตามธรรมเนียม)
        fpr, tpr, _ = roc_curve(y_test, p)
        # วาดเส้น ROC: แกน x = fpr, แกน y = tpr
        #   - label=... : ข้อความในคำอธิบายกราฟ (legend) แสดงชื่อโมเดลและ AUROC ทศนิยม 2 ตำแหน่ง (:.2f)
        plt.plot(fpr, tpr, label=f"{name} (AUROC {roc_auc_score(y_test, p):.2f})")
    # วาดเส้นทแยงมุมจาก (0,0) ถึง (1,1) แทนการโยนเหรียญ
    #   - [0, 1], [0, 1] : พิกัด x และ y ของจุดปลายทั้งสอง
    #   - "k--"          : สีดำ (k) เส้นประ (--)
    #   - label=...      : ข้อความในคำอธิบายกราฟ
    plt.plot([0, 1], [0, 1], "k--", label="coin flip (0.50)")
    # ตั้งชื่อแกน x และแกน y (เครื่องหมาย ; ใช้เขียนหลายคำสั่งในบรรทัดเดียว)
    #   - plt.xlabel(...) : ชื่อแกน x = 1 - ความจำเพาะ (การเตือนผิด)
    #   - plt.ylabel(...) : ชื่อแกน y = ความไว (กรณีที่ตรวจพบ)
    plt.xlabel("1 - specificity (false alarms)"); plt.ylabel("sensitivity (cases found)")
    # ตั้งชื่อกราฟ แสดงคำอธิบายกราฟ (legend) ขนาดตัวอักษร 8 และแสดงกราฟออกทางหน้าจอ
    #   - plt.title(...)          : ชื่อกราฟ พร้อมชื่อชุดข้อมูล
    #   - plt.legend(fontsize=8)  : แสดงกล่องคำอธิบายเส้นแต่ละเส้น
    #   - plt.show()              : แสดงกราฟ
    plt.title(f"ROC curve - {TABULAR_DATASET}"); plt.legend(fontsize=8); plt.show()

    # ------------------------------------------------------------------------
    # สร้างฟังก์ชัน demo สำหรับการสาธิตสด: แสดงผู้ป่วย 1 รายจากชุดทดสอบ พร้อมความเสี่ยงที่ทำนาย
    #   - patient_number : หมายเลขผู้ป่วยในชุดทดสอบ (0 ถึง จำนวนผู้ป่วยทดสอบ - 1)
    # ------------------------------------------------------------------------
    def demo(patient_number):
        """แสดงข้อมูลผู้ป่วย 1 รายที่โมเดลไม่เคยเห็น พร้อมความเสี่ยงที่โมเดลทำนาย"""
        # แปลงหมายเลขที่รับเข้ามาเป็นจำนวนเต็มด้วย int(...) (เช่น ข้อความ "5" -> 5) เก็บใน i
        i = int(patient_number)
        # ให้โมเดลที่ดีที่สุดทำนายความเสี่ยงของผู้ป่วยรายที่ i
        #   - X_test.iloc[[i]] : เลือกแถวที่ i โดยใช้วงเล็บ 2 ชั้น เพื่อให้ยังเป็นตาราง (1 แถว)
        #   - .predict_proba(...)[0, 1] : แถวแรก (0) คอลัมน์ "เกิดเหตุการณ์" (1)
        risk = best_model.predict_proba(X_test.iloc[[i]])[0, 1]
        # แสดงข้อมูลดิบของผู้ป่วยรายนี้
        #   - raw_test.iloc[i]          : แถวที่ i ของข้อมูลดิบชุดทดสอบ
        #   - .drop(labels=[target])    : ซ่อนคอลัมน์ผลลัพธ์ (ไม่เฉลยคำตอบก่อน)
        #   - .head(12)                 : แสดงเพียง 12 รายการแรก
        #   - .to_string()              : แปลงเป็นข้อความสำหรับพิมพ์
        print(raw_test.iloc[i].drop(labels=[target]).head(12).to_string())
        # ส่งข้อความสรุปกลับ (return) — ข้อความยาวในวงเล็บ ( ) ต่อกันจาก f-string 3 ส่วน
        #   - {risk:.0%}  : ความเสี่ยงที่ทำนาย เป็นร้อยละ
        #   - {'A' if เงื่อนไข else 'B'} : เลือกข้อความตามเงื่อนไข
        #       * ถ้า risk >= threshold -> "ระบุเป็นกลุ่มเสี่ยง ควรติดตาม"
        #       * ถ้า y_test.iloc[i] == 1 (ผลลัพธ์จริงของผู้ป่วยรายนี้) -> "เกิดเหตุการณ์"
        return (f"\nผู้ป่วยหมายเลข {i}: ความเสี่ยงที่ทำนาย {risk:.0%} -> "
                f"{'ระบุเป็นกลุ่มเสี่ยง ควรติดตาม' if risk >= threshold else 'ไม่ระบุเป็นกลุ่มเสี่ยง'} (เกณฑ์ตัดสิน {threshold:.0%}) "
                f"ผลลัพธ์จริง: {'เกิดเหตุการณ์' if y_test.iloc[i] == 1 else 'ไม่เกิดเหตุการณ์'}")
    # แจ้งว่าฟังก์ชัน demo พร้อมใช้ และบอกช่วงหมายเลขผู้ป่วยที่ใช้ได้
    print(f"\nฟังก์ชัน demo() พร้อมใช้งาน: ทดลอง demo(0) ถึง demo({len(y_test) - 1})")
```

💡 **คำอธิบายผลลัพธ์** โค้ดเริ่มต้นกันผู้ป่วยไว้ 25 % ฝึกโมเดลทั้งสองแบบด้วยข้อมูลส่วนที่เหลือ แล้วประเมินผลกับผู้ป่วยที่กันไว้ ซึ่งเทียบได้กับการตรวจสอบความถูกต้องของคะแนนทางคลินิกในกลุ่มผู้ป่วยใหม่ ค่า **AUROC** ใช้เปรียบเทียบกับการสุ่ม (0.50) **ความไว (sensitivity)** คือสัดส่วนของผู้ป่วยที่เกิดเหตุการณ์จริงซึ่งโมเดลระบุได้ และ **ความจำเพาะ (specificity)** คือสัดส่วนของผู้ที่ไม่เกิดเหตุการณ์ซึ่งโมเดลไม่ได้ระบุ การปรับ **เกณฑ์ตัดสิน (threshold)** ขึ้นหรือลงจะทำให้ค่าทั้งสองเปลี่ยนแปลงในทิศทางตรงข้ามกัน สำหรับชุดข้อมูล *readmission* การแบ่งข้อมูลทำตาม **ผู้ป่วย** เพื่อไม่ให้ผู้ป่วยรายเดียวกันอยู่ทั้งในข้อมูลฝึกและข้อมูลทดสอบ มิฉะนั้นผลลัพธ์จะดูดีเกินความเป็นจริง

## ขั้นที่ 3B — แนวทาง llm: ระบบคำสั่งหรือระบบตอบคำถามจากแนวปฏิบัติ (ปฏิบัติการที่ 8–9) · T4 GPU

โค้ดเริ่มต้นจะโหลดโมเดลภาษาขนาดใหญ่ (large language model; LLM) แบบเปิดตัวเดียวกับปฏิบัติการที่ 8–9 (`Qwen2.5-1.5B-Instruct` ไม่มีค่าใช้จ่ายและไม่ต้องใช้ API key) หากกำหนด **แหล่งความรู้** (ข้อความสั้นหลายข้อความ) ระบบจะทำงานแบบ **RAG** (Retrieval-Augmented Generation, ปฏิบัติการที่ 9) กล่าวคือ สำหรับแต่ละคำถาม ระบบจะค้นหาข้อความที่เกี่ยวข้องที่สุด 3 ข้อความก่อน แล้วสั่งให้ LLM ตอบโดยอ้างอิง *เฉพาะ* ข้อความเหล่านั้น หากกำหนด `KNOWLEDGE = "none"` ระบบจะเป็นต้นแบบที่ใช้ **คำสั่ง (prompt)** เพียงอย่างเดียว (ปฏิบัติการที่ 8) เช่น "เรียบเรียงคำแนะนำก่อนจำหน่ายนี้ให้เป็นภาษาที่เข้าใจง่าย"

ผู้เรียนต้องเขียน **กรณีทดสอบ (test cases)** 3–5 กรณี แต่ละกรณีประกอบด้วยข้อมูลเข้าและคำที่คำตอบที่ถูกต้องต้องมี ซึ่งใช้เป็นการประเมินผล

```python
# @title ▶️ รันต้นแบบ LLM: โหลดโมเดล → (สร้างดัชนีแหล่งความรู้) → รันกรณีทดสอบ (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (แนวทาง llm)
#   เซลล์นี้คือ "ต้นแบบระบบตอบคำถามจากแนวปฏิบัติ (RAG) หรือระบบที่ใช้คำสั่ง (prompt)"
#   โดยนำเทคนิคจากปฏิบัติการที่ 8–9 มาใช้ซ้ำ แบ่งเป็น 3 ขั้น ได้แก่
#   1) โหลดโมเดลภาษาขนาดใหญ่แบบเปิด (open LLM) Qwen2.5 (ใช้ GPU ถ้ามี)
#      และสร้างฟังก์ชัน chat(...) สำหรับส่งคำสั่งและรับคำตอบจากโมเดล
#   2) เตรียมแหล่งความรู้: แปลงทุกข้อความเป็นเวกเตอร์ (embedding — ชุดตัวเลขที่แทนความหมาย)
#      แล้วสร้างฟังก์ชัน retrieve(...) ค้นหาข้อความที่ใกล้เคียงคำถามที่สุด
#      และฟังก์ชัน demo(...) ที่รวมการค้นคืน (retrieval) + การสร้างคำตอบ (generation) = RAG
#   3) ประเมินผล: รันกรณีทดสอบทุกข้อ ตรวจว่าคำตอบมีคำที่คาดหวังหรือไม่ แล้วสรุปคะแนน
#   ข้อมูลเข้า : TRACK, SYSTEM_PROMPT, KNOWLEDGE, MY_PASSAGES, TEST_CASES
#   ผลลัพธ์   : ผลแต่ละกรณีทดสอบ (✅/❌) คะแนนรวม และฟังก์ชัน demo() พร้อมใช้งาน
#   ควรเลือก runtime แบบ T4 GPU; เซลล์นี้ใช้เวลาประมาณ 5 นาที
# ============================================================================

# ตรวจว่าแนวทางที่เลือกไม่ใช่ "llm" หรือไม่ (!= หมายถึง ไม่เท่ากับ)
if TRACK != "llm":
    # หากไม่ใช่ ให้แสดงข้อความแล้วข้ามเซลล์นี้
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
# กรณีเลือกแนวทาง llm — คำสั่งที่ย่อหน้าด้านล่างทั้งหมดจะทำงาน
else:
    # ------------------------------------------------------------------------
    # ติดตั้งและนำเข้าไลบรารี
    # ------------------------------------------------------------------------
    # นำเข้าไลบรารีมาตรฐานของ Python
    #   - subprocess : สั่งรันโปรแกรมอื่นจากใน Python (ใช้สั่ง pip ติดตั้งไลบรารี)
    #   - sys        : ข้อมูลเกี่ยวกับตัว Python ที่กำลังทำงาน
    #   - json       : อ่าน/เขียนไฟล์รูปแบบ JSON
    import subprocess, sys, json
    # ติดตั้งไลบรารีเพิ่มเติมด้วย pip (ทำงานเหมือน !pip install แต่สั่งจากใน Python)
    #   - [ ... ]           : list ของคำสั่งและตัวเลือก ต่อกันเป็นคำสั่งเดียว
    #       * sys.executable : ตำแหน่งโปรแกรม Python ที่กำลังใช้
    #       * "-m", "pip", "install" : เรียกโมดูล pip เพื่อติดตั้ง
    #       * "-q"           : แสดงข้อความน้อยที่สุด
    #       * "sentence-transformers" : ไลบรารีแปลงข้อความเป็นเวกเตอร์ (embedding)
    #       * "transformers>=4.44"    : ไลบรารีโมเดลภาษาของ Hugging Face รุ่น 4.44 ขึ้นไป
    #       * "accelerate"            : ไลบรารีช่วยจัดสรรโมเดลลง GPU/CPU อัตโนมัติ
    #   - check=True        : หากติดตั้งไม่สำเร็จ ให้แจ้งข้อผิดพลาดและหยุด
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "sentence-transformers",
                    "transformers>=4.44", "accelerate"], check=True)
    # นำเข้า numpy (คำนวณตัวเลข ชื่อย่อ np) และ torch (PyTorch ไลบรารีการเรียนรู้เชิงลึก)
    import numpy as np, torch
    # นำเข้า pipeline จาก transformers : ฟังก์ชันสร้าง "ชุดเครื่องมือสำเร็จรูป" สำหรับงานแต่ละประเภท เช่น สร้างข้อความ
    from transformers import pipeline

    # ------------------------------------------------------------------------
    # ขั้นที่ 1) โหลดโมเดลภาษา และสร้างฟังก์ชัน chat
    # ------------------------------------------------------------------------
    # 1) โหลด LLM แบบเปิด (ตัวเดียวกับปฏิบัติการที่ 8-9)
    # ตรวจว่าเครื่องมี GPU (การ์ดประมวลผลกราฟิก ซึ่งคำนวณโมเดลได้เร็วมาก) หรือไม่ ได้ True/False เก็บใน gpu
    #   - torch.cuda.is_available() : ฟังก์ชันของ PyTorch ตรวจว่ามี GPU แบบ CUDA ใช้งานได้
    gpu = torch.cuda.is_available()
    # เลือกชื่อโมเดล: ถ้ามี GPU ใช้รุ่น 1.5B (1.5 พันล้านพารามิเตอร์) มิฉะนั้นใช้รุ่นเล็ก 0.5B
    #   (รูปแบบ A if เงื่อนไข else B)
    llm_name = "Qwen/Qwen2.5-1.5B-Instruct" if gpu else "Qwen/Qwen2.5-0.5B-Instruct"
    # ถ้า "ไม่มี" GPU (not = กลับค่าความจริง)
    if not gpu:
        # แจ้งเตือนผู้เรียนว่ากำลังใช้โมเดลขนาดเล็ก และแนะนำให้เปลี่ยน runtime
        print("⚠️ ไม่พบ GPU จึงใช้โมเดลขนาดเล็กกว่า", llm_name, "(ควรเปลี่ยนเป็น runtime แบบ T4 GPU เพื่อคุณภาพคำตอบที่ดีขึ้น)")
    # โหลดโมเดลภาษาเป็นเครื่องมือสร้างข้อความ เก็บไว้ในตัวแปร llm
    #   - "text-generation" : ประเภทงาน คือ การสร้างข้อความต่อจากข้อความที่ให้
    #   - model=llm_name    : ชื่อโมเดลบน Hugging Face ที่จะดาวน์โหลด
    #   - device_map="auto" : ให้ระบบวางโมเดลบน GPU หรือ CPU อัตโนมัติ
    #   - torch_dtype=...   : ชนิดตัวเลขที่ใช้เก็บโมเดล
    #       * torch.float16 (ถ้ามี GPU) : ความละเอียดครึ่งหนึ่ง ใช้หน่วยความจำน้อยและเร็ว
    #       * torch.float32 (ถ้าไม่มี)  : ความละเอียดมาตรฐาน เหมาะกับ CPU
    llm = pipeline("text-generation", model=llm_name, device_map="auto",
                   torch_dtype=torch.float16 if gpu else torch.float32)

    # สร้างฟังก์ชัน chat สำหรับ "ส่งคำสั่งให้โมเดล แล้วรับคำตอบกลับ"
    #   - system              : คำสั่งระบบ (บทบาทและกติกา)
    #   - user                : ข้อความ/คำถามจากผู้ใช้
    #   - max_new_tokens=250  : จำนวนโทเคน (token — หน่วยย่อยของข้อความ ประมาณคำหรือส่วนของคำ)
    #                           สูงสุดที่โมเดลสร้างได้; หากไม่ระบุใช้ค่าเริ่มต้น 250
    def chat(system, user, max_new_tokens=250):
        # สร้างบทสนทนาเป็น list ของ dict 2 ข้อความ ตามรูปแบบแชตของโมเดล
        #   - {"role": "system", "content": system} : ข้อความบทบาท "ระบบ"
        #   - {"role": "user",   "content": user}   : ข้อความบทบาท "ผู้ใช้"
        messages = [{"role": "system", "content": system}, {"role": "user", "content": user}]
        # ส่งบทสนทนาให้โมเดล แล้วดึงเฉพาะข้อความคำตอบกลับไป (return)
        #   - llm(messages, ...)          : ให้โมเดลสร้างคำตอบต่อจากบทสนทนา
        #       * max_new_tokens=...      : จำกัดความยาวคำตอบ
        #       * do_sample=False         : ไม่สุ่ม — เลือกคำที่น่าจะเป็นที่สุดเสมอ จึงได้คำตอบเหมือนเดิมทุกครั้ง
        #   - [0]                         : ผลลัพธ์ชุดแรก (มีชุดเดียว)
        #   - ["generated_text"]          : บทสนทนาทั้งหมดรวมคำตอบใหม่ที่ต่อท้าย
        #   - [-1]                        : ข้อความสุดท้าย (ตำแหน่ง -1 = นับจากท้าย) คือคำตอบของโมเดล
        #   - ["content"]                 : ตัวเนื้อหาข้อความคำตอบ
        return llm(messages, max_new_tokens=max_new_tokens, do_sample=False)[0]["generated_text"][-1]["content"]

    # ------------------------------------------------------------------------
    # ขั้นที่ 2) เตรียมแหล่งความรู้ และสร้างดัชนีค้นหา (index)
    # ------------------------------------------------------------------------
    # 2) แหล่งความรู้: แปลงทุกข้อความเป็นเวกเตอร์ (embedding) เพื่อค้นหาข้อความที่ใกล้เคียงกับคำถามที่สุด (ปฏิบัติการที่ 9)
    # กรณีใช้ฐานความรู้ของรายวิชา
    if KNOWLEDGE == "course_kb":
        # ลองดาวน์โหลดฐานความรู้ (หากไม่สำเร็จจะไปทำใน except)
        try:
            # โหลดฐานความรู้และดึงเฉพาะข้อความ เก็บเป็น list ใน passages
            #   - lab_data("rag/kb.json") : ดาวน์โหลดไฟล์ฐานความรู้ของรายวิชา (ฟังก์ชันจากเซลล์การตั้งค่า)
            #   - open(...)               : เปิดไฟล์
            #   - json.load(...)          : อ่านไฟล์ JSON ได้เป็น list ของ dict (1 dict ต่อ 1 ข้อความ)
            #   - [d["text"] for d in ...] : list comprehension — วนทีละ dict (d) แล้วเก็บเฉพาะค่า "text"
            passages = [d["text"] for d in json.load(open(lab_data("rag/kb.json")))]
        # หากเกิดข้อผิดพลาดใด ๆ (เก็บไว้ในตัวแปร e)
        except Exception as e:   # เข้าถึงข้อมูลของรายวิชาไม่ได้ -> ใช้ฐานความรู้ขนาดเล็กจากปฏิบัติการที่ 9
            # แจ้งผู้เรียนพร้อมชื่อประเภทข้อผิดพลาด
            print("หมายเหตุ: ไม่สามารถเข้าถึงฐานความรู้ของรายวิชาได้ จึงใช้ฐานความรู้ 6 ข้อความจากปฏิบัติการที่ 9", type(e).__name__)
            # ใช้ฐานความรู้สำรอง 6 ข้อความ (แนวปฏิบัติโรคเบาหวานชนิดที่ 2 แบบสังเคราะห์) เป็น list ของข้อความ
            passages = [
                # เกณฑ์วินิจฉัยเบาหวานชนิดที่ 2: HbA1c ≥ 6.5% หรือน้ำตาลขณะอดอาหาร ≥ 126 mg/dL
                "Type 2 diabetes is diagnosed when HbA1c is 6.5% or higher, or fasting glucose >= 126 mg/dL.",
                # ยาลำดับแรกคือ metformin เว้นแต่มีข้อห้ามใช้
                "First-line pharmacotherapy for type 2 diabetes is metformin unless contraindicated.",
                # เป้าหมายความดันโลหิตในผู้ใหญ่ที่เป็นเบาหวานส่วนใหญ่ ต่ำกว่า 130/80 mmHg
                "Blood pressure target for most adults with diabetes is below 130/80 mmHg.",
                # แนะนำยากลุ่ม statin ในผู้ป่วยเบาหวานอายุ 40–75 ปี เพื่อลดความเสี่ยงโรคหัวใจและหลอดเลือด
                "Statins are recommended for diabetic patients aged 40-75 to reduce cardiovascular risk.",
                # ห้ามใช้ metformin เมื่อ eGFR ต่ำกว่า 30
                "Metformin is contraindicated when eGFR is below 30 mL/min/1.73m2.",
                # แนะนำตรวจคัดกรองจอประสาทตาปีละครั้ง
                "Annual retinal screening is recommended for patients with type 2 diabetes.",
            ]
    # กรณีใช้ข้อความของผู้เรียนเอง
    elif KNOWLEDGE == "my_passages":
        # คัดลอกข้อความจาก MY_PASSAGES มาเป็น list ใหม่ด้วย list(...)
        passages = list(MY_PASSAGES)
        # ตรวจว่ามีข้อความอย่างน้อย 1 ข้อความ (list ว่างถือเป็นเท็จ) หากว่างจะหยุดพร้อมข้อความเตือน
        assert passages, "MY_PASSAGES ยังว่างอยู่ โปรดวางข้อความ หรือกำหนด KNOWLEDGE = 'course_kb'"
    # กรณีอื่น (เช่น "none"): ไม่ใช้แหล่งความรู้ — เป็นต้นแบบที่ใช้คำสั่งอย่างเดียว
    else:
        # กำหนดเป็น list ว่าง
        passages = []
    # ถ้ามีข้อความในแหล่งความรู้ (list ไม่ว่าง) ให้สร้างดัชนีเวกเตอร์
    if passages:
        # นำเข้า SentenceTransformer : โมเดลแปลงประโยคเป็นเวกเตอร์ จากไลบรารี sentence_transformers
        from sentence_transformers import SentenceTransformer
        # โหลดโมเดล embedding ขนาดเล็กชื่อ "all-MiniLM-L6-v2" (เร็ว ใช้กันแพร่หลาย) เก็บใน embedder
        embedder = SentenceTransformer("all-MiniLM-L6-v2")
        # แปลงทุกข้อความเป็นเวกเตอร์ เก็บใน passage_vecs (ตาราง: 1 แถวต่อ 1 ข้อความ)
        #   - passages                 : list ของข้อความ
        #   - normalize_embeddings=True: ปรับความยาวเวกเตอร์ให้เท่ากับ 1 เพื่อให้ผลคูณจุด (dot product)
        #                                เท่ากับความคล้ายแบบโคไซน์ (cosine similarity)
        passage_vecs = embedder.encode(passages, normalize_embeddings=True)
        # แจ้งจำนวนข้อความที่สร้างดัชนีแล้ว
        print(f"สร้างดัชนีข้อความแล้ว {len(passages)} ข้อความ")

    # สร้างฟังก์ชัน retrieve สำหรับ "ค้นคืน" ข้อความที่เกี่ยวข้องกับคำถามมากที่สุด
    #   - question : คำถาม
    #   - k=3      : จำนวนข้อความที่ต้องการ (ค่าเริ่มต้น 3 ข้อความ)
    def retrieve(question, k=3):
        # คำนวณความคล้ายระหว่างคำถามกับทุกข้อความ เก็บใน sims (1 ค่าต่อ 1 ข้อความ ยิ่งมากยิ่งคล้าย)
        #   - embedder.encode([question], normalize_embeddings=True) : แปลงคำถามเป็นเวกเตอร์
        #   - [0]                                                     : เวกเตอร์ของคำถามแรก (มีข้อเดียว)
        #   - passage_vecs @ ...                                      : @ คือการคูณเมทริกซ์ (ผลคูณจุดกับทุกข้อความ)
        sims = passage_vecs @ embedder.encode([question], normalize_embeddings=True)[0]
        # ส่งกลับข้อความ k ข้อความที่คล้ายที่สุด
        #   - np.argsort(-sims) : เรียงตำแหน่งตามค่า -sims จากน้อยไปมาก = ตามความคล้ายจากมากไปน้อย
        #   - [:k]              : เลือก k ตำแหน่งแรก
        #   - [passages[i] for i in ...] : ดึงข้อความตามตำแหน่งเหล่านั้นมาเป็น list
        return [passages[i] for i in np.argsort(-sims)[:k]]

    # สร้างฟังก์ชัน demo สำหรับการสาธิตและการประเมินผล
    #   - question : คำถาม (กรณี RAG) หรือข้อความที่ป้อน (กรณีใช้คำสั่งอย่างเดียว)
    def demo(question):
        """ตอบคำถาม 1 ข้อ (RAG) หรือประมวลผลข้อความที่ป้อน 1 ข้อความ (ใช้คำสั่งอย่างเดียว)"""
        # ถ้ามีแหล่งความรู้ ให้ทำงานแบบ RAG
        if passages:
            # สร้างข้อความอ้างอิง (context) จากข้อความที่ค้นคืนได้ ใส่หมายเลข [1], [2], [3]
            #   - retrieve(question)        : ได้ list ข้อความที่เกี่ยวข้อง 3 ข้อความ
            #   - enumerate(...)            : ให้ทั้งลำดับ i (เริ่ม 0) และข้อความ p ในแต่ละรอบ
            #   - f"[{i + 1}] {p}"          : จัดรูปแบบเป็น "[1] ข้อความ" (บวก 1 ให้เริ่มนับที่ 1)
            #   - "\n".join(...)            : นำทุกบรรทัดมาต่อกันโดยคั่นด้วยการขึ้นบรรทัดใหม่
            context = "\n".join(f"[{i + 1}] {p}" for i, p in enumerate(retrieve(question)))
            # ส่งคำสั่งระบบ + ข้อความอ้างอิง + คำถาม ให้โมเดลตอบ ผ่านฟังก์ชัน chat
            #   - SYSTEM_PROMPT : คำสั่งระบบจากแบบฝึกหัดที่ 2B
            #   - f"Context:\n{context}\n\nQuestion: {question}" : ข้อความผู้ใช้ที่รวมบริบทและคำถาม
            answer = chat(SYSTEM_PROMPT, f"Context:\n{context}\n\nQuestion: {question}")
            # ส่งกลับคำตอบ ตามด้วยเส้นคั่น และข้อความที่ค้นคืนได้ (เพื่อให้ตรวจสอบแหล่งอ้างอิงได้)
            return f"{answer}\n\n--- retrieved passages ---\n{context}"
        # ถ้าไม่มีแหล่งความรู้: ส่งคำสั่งระบบและข้อความที่ป้อนให้โมเดลโดยตรง แล้วส่งคำตอบกลับ
        return chat(SYSTEM_PROMPT, question)

    # ------------------------------------------------------------------------
    # ขั้นที่ 3) ประเมินผลด้วยกรณีทดสอบ
    # ------------------------------------------------------------------------
    # 3) ประเมินผล: คำตอบแต่ละข้อมีคำที่คาดหวังหรือไม่
    # ตัวนับจำนวนกรณีที่ผ่าน เริ่มต้นที่ 0
    hits = 0
    # วนซ้ำทีละกรณีทดสอบ: แยกแต่ละคู่เป็น question (คำถาม) และ expected (คำที่คาดหวัง)
    for question, expected in TEST_CASES:
        # เรียก demo เพื่อให้ได้คำตอบ แล้วตัดส่วน "ข้อความที่ค้นคืน" ทิ้ง เหลือเฉพาะคำตอบของโมเดล
        #   - .split("\n\n--- retrieved") : แยกข้อความ ณ ตำแหน่งเส้นคั่น ได้เป็น list
        #   - [0]                         : เลือกส่วนแรก (ก่อนเส้นคั่น) คือคำตอบ
        answer = demo(question).split("\n\n--- retrieved")[0]
        # ตรวจว่าคำที่คาดหวังปรากฏในคำตอบหรือไม่ (แปลงเป็นตัวพิมพ์เล็กทั้งคู่ด้วย .lower() เพื่อไม่สนตัวพิมพ์)
        #   ได้ True (ผ่าน) หรือ False (ไม่ผ่าน) เก็บใน ok
        ok = expected.lower() in answer.lower()
        # เพิ่มตัวนับ: += คือบวกเพิ่มเข้าไป; True มีค่าเท่ากับ 1 และ False เท่ากับ 0
        hits += ok
        # แสดงผลกรณีนี้
        #   - {'✅' if ok else '❌'} : เครื่องหมายผ่าน/ไม่ผ่าน
        #   - \n และช่องว่างนำหน้า   : ขึ้นบรรทัดใหม่และย่อหน้า
        #   - {answer[:200]}         : แสดงคำตอบเพียง 200 ตัวอักษรแรก
        print(f"{'✅' if ok else '❌'} {question}\n   คำที่คาดหวัง '{expected}' | คำตอบ: {answer[:200]}\n")
    # สรุปคะแนนรวม: จำนวนที่ผ่าน / จำนวนทั้งหมด และร้อยละ (:.0% = ร้อยละไม่มีทศนิยม)
    print(f"คะแนน: ผ่าน {hits}/{len(TEST_CASES)} กรณีทดสอบ ({hits / len(TEST_CASES):.0%})")
```

```python
# @title ▶️ รันต้นแบบ image: ดาวน์โหลด → ปรับแต่ง ResNet-18 → ประเมินผล (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (แนวทาง image)
#   เซลล์นี้คือ "ต้นแบบโมเดลจำแนกภาพทางการแพทย์" โดยนำเทคนิคจากปฏิบัติการที่ 4 มาใช้ซ้ำ
#   แบ่งเป็น 4 ขั้น ได้แก่
#   1) ดาวน์โหลดชุดภาพ MedMNIST ที่เลือก (ภาพขนาด 28x28 พิกเซล) แยกเป็นชุดฝึกและชุดทดสอบ
#   2) เตรียมโมเดล ResNet-18 ที่ฝึกไว้ล่วงหน้าด้วยภาพทั่วไป (ImageNet) แล้วปรับให้เข้ากับ
#      จำนวนช่องสีและจำนวนประเภทของภาพทางการแพทย์ — เรียกว่าการเรียนรู้แบบถ่ายโอน (transfer learning)
#   3) ฝึกโมเดลตามจำนวนรอบ EPOCHS
#   4) ประเมินผลกับภาพทดสอบที่โมเดลไม่เคยเห็น: ความแม่นยำ (accuracy) เทียบกับค่าอ้างอิง
#      "ทายประเภทที่พบบ่อยที่สุดเสมอ", AUROC, อัตราจำแนกถูกรายประเภท และเมทริกซ์ความสับสน (confusion matrix)
#   สุดท้ายสร้างฟังก์ชัน demo(...) สำหรับแสดงภาพ 1 ภาพพร้อมผลการทำนายในการสาธิตสด
#   ข้อมูลเข้า : TRACK, IMAGE_DATASET, EPOCHS
#   ผลลัพธ์   : ความแม่นยำแต่ละ epoch ผลการประเมิน ภาพเมทริกซ์ความสับสน และฟังก์ชัน demo()
#   ใช้เวลาประมาณ 3–8 นาที (เร็วขึ้นเมื่อใช้ GPU)
# ============================================================================

# ตรวจว่าแนวทางที่เลือกไม่ใช่ "image" หรือไม่
if TRACK != "image":
    # หากไม่ใช่ ให้แสดงข้อความแล้วข้ามเซลล์นี้
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
# กรณีเลือกแนวทาง image — คำสั่งที่ย่อหน้าด้านล่างทั้งหมดจะทำงาน
else:
    # ------------------------------------------------------------------------
    # ติดตั้งและนำเข้าไลบรารี
    # ------------------------------------------------------------------------
    # นำเข้า subprocess (สั่งรันโปรแกรมอื่น) และ sys (ข้อมูลของ Python ที่กำลังทำงาน)
    import subprocess, sys
    # ติดตั้งไลบรารี medmnist (ชุดภาพทางการแพทย์ขนาดเล็ก) ด้วย pip
    #   - sys.executable, "-m", "pip", "install" : ใช้ pip ของ Python ตัวที่กำลังทำงาน
    #   - "-q"        : แสดงข้อความน้อยที่สุด
    #   - check=True  : หากติดตั้งไม่สำเร็จให้หยุดพร้อมแจ้งข้อผิดพลาด
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "medmnist"], check=True)
    # นำเข้าไลบรารีหลัก
    #   - numpy as np               : คำนวณตัวเลข/อาร์เรย์
    #   - matplotlib.pyplot as plt  : วาดกราฟและแสดงภาพ
    #   - torch                     : PyTorch ไลบรารีการเรียนรู้เชิงลึก (deep learning)
    #   - torch.nn as nn            : ส่วนประกอบของโครงข่ายประสาทเทียม (neural network) เช่น ชั้นต่าง ๆ และฟังก์ชันความสูญเสีย
    import numpy as np, matplotlib.pyplot as plt, torch, torch.nn as nn
    # นำเข้าไลบรารี medmnist ทั้งไลบรารี
    import medmnist
    # นำเข้า INFO : dict ที่เก็บข้อมูลอธิบายของแต่ละชุดภาพ (ประเภทงาน ชื่อประเภท จำนวนช่องสี ฯลฯ)
    from medmnist import INFO
    # นำเข้าจาก torchvision (ไลบรารีด้านภาพของ PyTorch)
    #   - transforms : เครื่องมือแปลง/ปรับภาพก่อนป้อนโมเดล
    #   - models     : โมเดลภาพสำเร็จรูป เช่น ResNet-18
    from torchvision import transforms, models
    # นำเข้า DataLoader : ตัวช่วยแบ่งภาพเป็นกลุ่มย่อย (batch) และป้อนให้โมเดลทีละกลุ่ม
    from torch.utils.data import DataLoader
    # นำเข้าเครื่องมือวัดผลจาก scikit-learn
    #   - roc_auc_score          : คำนวณ AUROC
    #   - confusion_matrix       : สร้างเมทริกซ์ความสับสน (ตารางนับ ประเภทจริง × ประเภทที่ทำนาย)
    #   - ConfusionMatrixDisplay : วาดเมทริกซ์ความสับสนเป็นภาพ
    from sklearn.metrics import roc_auc_score, confusion_matrix, ConfusionMatrixDisplay

    # ------------------------------------------------------------------------
    # ขั้นที่ 1) เตรียมข้อมูลภาพ
    # ------------------------------------------------------------------------
    # 1) ข้อมูล
    # ดึงข้อมูลอธิบายของชุดภาพที่เลือก (เช่น INFO["pneumoniamnist"]) เก็บใน info
    info = INFO[IMAGE_DATASET]
    # ตรวจว่าชุดภาพนี้ "ไม่ใช่" งานหลายป้ายกำกับ (multi-label = 1 ภาพมีได้หลายประเภทพร้อมกัน)
    #   เพราะโค้ดนี้รองรับเฉพาะงานที่ 1 ภาพมี 1 ประเภท หากไม่ผ่านจะหยุดพร้อมข้อความเตือน
    assert "multi-label" not in info["task"], "โปรดเลือกชุดข้อมูลที่แต่ละภาพมีประเภทเดียว (single-label)"
    # สร้าง list ชื่อประเภททั้งหมดเรียงตามหมายเลข 0, 1, 2, ...
    #   - info["label"]              : dict ที่ใช้หมายเลข (เป็นข้อความ "0", "1", ...) เป็นกุญแจ และชื่อประเภทเป็นค่า
    #   - range(len(info["label"]))  : ลำดับเลข 0 ถึง จำนวนประเภท - 1
    #   - str(i)                     : แปลงเลขเป็นข้อความ เพื่อใช้เป็นกุญแจของ dict
    class_names = [info["label"][str(i)] for i in range(len(info["label"]))]
    # จำนวนช่องสีของภาพ: 1 = ภาพขาวดำ (เช่น เอกซเรย์), 3 = ภาพสี RGB (เช่น ภาพผิวหนัง)
    n_ch = info["n_channels"]
    # เลือกอุปกรณ์คำนวณ: "cuda" (GPU) ถ้ามี มิฉะนั้น "cpu"
    device = "cuda" if torch.cuda.is_available() else "cpu"
    # สร้างชุดการแปลงภาพ tf ที่ทำต่อกันตามลำดับด้วย transforms.Compose([...])
    #   - transforms.ToTensor()       : แปลงภาพเป็น tensor (อาร์เรย์ตัวเลขของ PyTorch) ค่าพิกเซลอยู่ในช่วง 0–1
    #   - transforms.Normalize(mean, std) : ปรับค่าเป็น (ค่า - mean) / std
    #       * [.5] * n_ch : list ค่า 0.5 จำนวนเท่ากับช่องสี เช่น [0.5] หรือ [0.5, 0.5, 0.5]
    #       * ใช้ 0.5 ทั้ง mean และ std ทำให้ค่าพิกเซลอยู่ในช่วง -1 ถึง 1
    tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize([.5] * n_ch, [.5] * n_ch)])
    # ดึง "คลาส" (แม่แบบ) ของชุดภาพตามชื่อที่เก็บใน info["python_class"] เช่น PneumoniaMNIST
    #   - getattr(medmnist, "ชื่อ") : ฟังก์ชันในตัวของ Python ดึงสิ่งที่ชื่อนั้นออกจากไลบรารี
    DataClass = getattr(medmnist, info["python_class"])
    # โหลดชุดภาพสำหรับฝึก
    #   - split="train" : ส่วนข้อมูลฝึก
    #   - transform=tf  : แปลงภาพแต่ละภาพด้วย tf ทุกครั้งที่ดึงภาพ
    #   - download=True : ดาวน์โหลดไฟล์หากยังไม่มีในเครื่อง
    #   - size=28       : ใช้ภาพขนาด 28x28 พิกเซล
    train_ds = DataClass(split="train", transform=tf, download=True, size=28)
    # โหลดชุดภาพสำหรับทดสอบ (split="test") ด้วยการตั้งค่าเดียวกัน
    test_ds = DataClass(split="test", transform=tf, download=True, size=28)
    # แสดงจำนวนภาพฝึก/ทดสอบ ชื่อประเภท และอุปกรณ์ที่ใช้ (len นับจำนวนภาพ)
    print(f"{IMAGE_DATASET}: ภาพฝึก {len(train_ds)} ภาพ / ภาพทดสอบ {len(test_ds)} ภาพ ประเภท: {class_names} อุปกรณ์: {device}")
    # สร้างตัวป้อนภาพฝึก
    #   - batch_size=64 : ป้อนครั้งละ 64 ภาพ
    #   - shuffle=True  : สลับลำดับภาพทุก epoch เพื่อให้โมเดลไม่จำลำดับ
    train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
    # สร้างตัวป้อนภาพทดสอบ ครั้งละ 256 ภาพ (ไม่สลับลำดับ เพื่อให้ตรงกับหมายเลขภาพ)
    test_loader = DataLoader(test_ds, batch_size=256)

    # ------------------------------------------------------------------------
    # ขั้นที่ 2) เตรียมโมเดล ResNet-18 (การเรียนรู้แบบถ่ายโอน)
    # ------------------------------------------------------------------------
    # 2) โมเดล: ResNet-18 ที่ฝึกไว้ล่วงหน้า ปรับให้ตรงกับจำนวนช่องสีและจำนวนประเภท (ปฏิบัติการที่ 4 แบบฝึกหัดที่ 1)
    # โหลดโมเดล ResNet-18 (โครงข่ายประสาทเทียมแบบคอนโวลูชัน 18 ชั้น)
    #   - weights="IMAGENET1K_V1" : ใช้ค่าน้ำหนักที่ฝึกไว้แล้วกับภาพทั่วไป ImageNet (ประมาณ 1.2 ล้านภาพ 1,000 ประเภท)
    net = models.resnet18(weights="IMAGENET1K_V1")
    # ถ้าภาพเป็นขาวดำ (1 ช่องสี) ต้องเปลี่ยนชั้นแรกซึ่งเดิมรับภาพสี 3 ช่อง
    if n_ch == 1:
        # แทนที่ชั้นคอนโวลูชันแรก (conv1) ด้วยชั้นใหม่ที่รับภาพ 1 ช่อง (ชั้นใหม่นี้เริ่มจากค่าสุ่ม ต้องเรียนรู้ใหม่)
        #   - 1             : จำนวนช่องสีของภาพเข้า
        #   - 64            : จำนวนตัวกรอง (filter) ที่ส่งออก (เท่ากับชั้นเดิม เพื่อให้ต่อกับชั้นถัดไปได้)
        #   - kernel_size=7 : ขนาดตัวกรอง 7x7 พิกเซล
        #   - stride=2      : เลื่อนตัวกรองทีละ 2 พิกเซล (ย่อขนาดภาพลงครึ่งหนึ่ง)
        #   - padding=3     : เติมขอบภาพ 3 พิกเซล
        #   - bias=False    : ไม่ใช้ค่าคงที่บวกเพิ่ม (เหมือนชั้นเดิมของ ResNet)
        net.conv1 = nn.Conv2d(1, 64, kernel_size=7, stride=2, padding=3, bias=False)
    # แทนที่ชั้นสุดท้าย (fc — fully connected) ให้ส่งออกจำนวนค่าเท่ากับจำนวนประเภทของเรา (แทน 1,000 ประเภทเดิม)
    #   - net.fc.in_features : จำนวนค่าที่ป้อนเข้าชั้นนี้ (ของ ResNet-18 คือ 512)
    #   - len(class_names)   : จำนวนประเภท เช่น 2 สำหรับ ปกติ/ปอดอักเสบ
    net.fc = nn.Linear(net.fc.in_features, len(class_names))
    # ย้ายโมเดลไปไว้บนอุปกรณ์ที่เลือก (GPU หรือ CPU)
    net = net.to(device)
    # สร้างตัวปรับค่าน้ำหนัก (optimizer) แบบ Adam
    #   - net.parameters() : ค่าน้ำหนักทั้งหมดของโมเดลที่จะปรับ
    #   - lr=1e-4          : อัตราการเรียนรู้ (learning rate) = 0.0001 (1e-4 คือ 1 × 10^-4)
    opt = torch.optim.Adam(net.parameters(), lr=1e-4)   # อัตราการเรียนรู้ต่ำ: ปรับแต่งโมเดลอย่างค่อยเป็นค่อยไป
    # ฟังก์ชันความสูญเสีย (loss function) แบบ cross-entropy วัดว่าการทำนายผิดจากคำตอบมากเพียงใด (ใช้กับงานจำแนกประเภท)
    crit = nn.CrossEntropyLoss()

    # ------------------------------------------------------------------------
    # ขั้นที่ 3) ฝึกโมเดล
    # ------------------------------------------------------------------------
    # 3) ฝึกโมเดล
    # วนซ้ำตามจำนวน epoch: range(n) ให้ลำดับเลข 0 ถึง n-1; int(...) แปลง EPOCHS เป็นจำนวนเต็ม
    for epoch in range(int(EPOCHS)):
        # ตั้งโมเดลเป็นโหมดฝึก (net.train()) และตั้งตัวนับ correct (ทายถูก) และ total (ทั้งหมด) เป็น 0
        #   - correct = total = 0 : กำหนดค่า 0 ให้ทั้งสองตัวแปรพร้อมกัน
        net.train(); correct = total = 0
        # วนซ้ำทีละกลุ่มภาพ (batch): x = ภาพ, yb = ป้ายกำกับ (label) ประเภทจริง
        for x, yb in train_loader:
            # ย้ายภาพและป้ายกำกับไปยังอุปกรณ์คำนวณ
            #   - x.to(device)       : ย้ายภาพ
            #   - yb.squeeze(1)      : ลดมิติป้ายกำกับจากรูป (จำนวนภาพ, 1) เป็น (จำนวนภาพ,)
            #   - .long()            : แปลงเป็นจำนวนเต็มแบบ long ที่ CrossEntropyLoss ต้องการ
            #   - .to(device)        : ย้ายไปยังอุปกรณ์
            x, yb = x.to(device), yb.squeeze(1).long().to(device)
            # ป้อนภาพเข้าโมเดลได้ผลทำนาย out (คะแนนของแต่ละประเภท) แล้วคำนวณความสูญเสีย loss เทียบกับคำตอบ yb
            out = net(x); loss = crit(out, yb)
            # 3 คำสั่งของการเรียนรู้ 1 ก้าว
            #   - opt.zero_grad() : ล้างค่าความชัน (gradient) ที่ค้างจากรอบก่อน
            #   - loss.backward() : คำนวณค่าความชันย้อนกลับ (backpropagation) ว่าควรปรับน้ำหนักแต่ละตัวทิศทางใด
            #   - opt.step()      : ปรับค่าน้ำหนักของโมเดลตามค่าความชัน
            opt.zero_grad(); loss.backward(); opt.step()
            # นับจำนวนภาพที่ทายถูก และจำนวนภาพทั้งหมด
            #   - out.argmax(1)    : ประเภทที่ได้คะแนนสูงสุดของแต่ละภาพ (คำทำนาย)
            #   - == yb            : เทียบกับคำตอบจริง ได้ True/False
            #   - .sum().item()    : นับจำนวน True แล้วแปลงเป็นตัวเลขธรรมดาของ Python
            #   - total += len(yb) : บวกจำนวนภาพใน batch นี้
            correct += (out.argmax(1) == yb).sum().item(); total += len(yb)
        # แสดงความแม่นยำกับข้อมูลฝึกเมื่อจบแต่ละ epoch (epoch + 1 เพื่อนับเริ่มที่ 1; :.3f = ทศนิยม 3 ตำแหน่ง)
        print(f"epoch {epoch + 1}/{EPOCHS}: ความแม่นยำกับข้อมูลฝึก (train accuracy) {correct / total:.3f}")

    # ------------------------------------------------------------------------
    # ขั้นที่ 4) ประเมินผลกับภาพทดสอบ
    # ------------------------------------------------------------------------
    # 4) ประเมินผลกับภาพทดสอบที่โมเดลไม่เคยเห็น
    # ตั้งโมเดลเป็นโหมดประเมินผล (net.eval()) และสร้าง list ว่าง 2 ตัว
    #   - probs  : เก็บความน่าจะเป็นที่ทำนาย
    #   - y_true : เก็บประเภทจริง
    net.eval(); probs, y_true = [], []
    # with torch.no_grad(): ภายในบล็อกนี้ไม่ต้องคำนวณค่าความชัน (ไม่ได้ฝึก) จึงเร็วขึ้นและประหยัดหน่วยความจำ
    #   (with คือคำสั่งเปิดบริบทชั่วคราว เมื่อจบบล็อกที่ย่อหน้าจะกลับสู่ปกติ)
    with torch.no_grad():
        # วนซ้ำทีละกลุ่มภาพทดสอบ
        for x, yb in test_loader:
            # 2 คำสั่งในบรรทัดเดียว:
            #   (1) ทำนายแล้วเก็บความน่าจะเป็น
            #       - net(x.to(device))       : ป้อนภาพเข้าโมเดลได้คะแนนแต่ละประเภท
            #       - torch.softmax(..., 1)   : แปลงคะแนนเป็นความน่าจะเป็น (รวมกันได้ 1) ตามมิติที่ 1 (ประเภท)
            #       - .cpu().numpy()          : ย้ายกลับมา CPU และแปลงเป็นอาร์เรย์ numpy
            #       - probs.append(...)       : เพิ่มต่อท้าย list
            #   (2) เพิ่มประเภทจริงต่อท้าย y_true
            #       - yb.squeeze(1).tolist()  : ลดมิติแล้วแปลงเป็น list ธรรมดา
            probs.append(torch.softmax(net(x.to(device)), 1).cpu().numpy()); y_true += yb.squeeze(1).tolist()
    # รวมผลทุก batch เป็นอาร์เรย์เดียว
    #   - np.concatenate(probs) : ต่ออาร์เรย์ความน่าจะเป็นทุก batch เป็นตารางเดียว (1 แถวต่อ 1 ภาพ)
    #   - np.array(y_true)      : แปลง list ประเภทจริงเป็นอาร์เรย์
    probs, y_true = np.concatenate(probs), np.array(y_true)
    # ประเภทที่ทำนาย = ประเภทที่มีความน่าจะเป็นสูงสุดของแต่ละภาพ (argmax ตามมิติที่ 1)
    y_pred = probs.argmax(1)
    # คำนวณ AUROC
    #   - ถ้ามี 2 ประเภท : roc_auc_score(ประเภทจริง, ความน่าจะเป็นของประเภทที่ 1)
    #   - ถ้ามากกว่า 2   : ใช้ multi_class="ovr" (one-vs-rest) คือคิด AUROC ของแต่ละประเภทเทียบกับที่เหลือ แล้วเฉลี่ย
    auroc = roc_auc_score(y_true, probs[:, 1]) if len(class_names) == 2 else roc_auc_score(y_true, probs, multi_class="ovr")
    # ค่าอ้างอิง "ทายประเภทที่พบบ่อยที่สุดเสมอ" = จำนวนภาพของประเภทที่พบมากที่สุด ÷ จำนวนภาพทั้งหมด
    #   - np.bincount(y_true) : นับจำนวนภาพของแต่ละประเภท
    #   - .max()              : จำนวนที่มากที่สุด
    majority = np.bincount(y_true).max() / len(y_true)
    # แสดงความแม่นยำกับชุดทดสอบ เทียบค่าอ้างอิง และ AUROC (ข้อความ f-string 2 ส่วนต่อกันในวงเล็บ)
    #   - np.mean(y_pred == y_true) : สัดส่วนภาพที่ทายถูก (ค่าเฉลี่ยของ True=1/False=0)
    print(f"\nความแม่นยำกับชุดทดสอบ (test accuracy) {np.mean(y_pred == y_true):.3f} "
          f"(ค่าอ้างอิง 'ทายประเภทที่พบบ่อยที่สุดเสมอ': {majority:.3f}) | AUROC {auroc:.3f}")
    # สร้างเมทริกซ์ความสับสน: แถว = ประเภทจริง, คอลัมน์ = ประเภทที่ทำนาย
    #   - labels=range(len(class_names)) : ระบุให้มีครบทุกประเภท แม้บางประเภทจะไม่ถูกทำนายเลย
    cm = confusion_matrix(y_true, y_pred, labels=range(len(class_names)))
    # วนซ้ำทีละประเภท: i = หมายเลขประเภท, name = ชื่อประเภท (enumerate ให้ทั้งลำดับและค่า)
    for i, name in enumerate(class_names):
        # แสดงอัตราการจำแนกถูกของประเภทนี้ (recall รายประเภท)
        #   - cm[i, i]              : จำนวนภาพประเภท i ที่ทายถูก (ช่องแนวทแยง)
        #   - cm[i].sum()           : จำนวนภาพประเภท i ทั้งหมด (ผลรวมทั้งแถว)
        #   - max(..., 1)           : ป้องกันการหารด้วยศูนย์ หากไม่มีภาพประเภทนี้
        print(f"  จากภาพ '{name}' ทั้งหมด โมเดลจำแนกถูก {cm[i, i] / max(cm[i].sum(), 1):.0%}")
    # วาดเมทริกซ์ความสับสนเป็นภาพ
    #   - cm                                     : ตารางตัวเลขที่จะวาด
    #   - display_labels=[n[:12] for n in ...]   : ชื่อประเภทบนแกน ตัดเหลือ 12 ตัวอักษรแรกเพื่อไม่ให้ยาวเกิน
    #   - .plot(xticks_rotation=45)              : วาดภาพ โดยหมุนชื่อบนแกน x 45 องศา
    ConfusionMatrixDisplay(cm, display_labels=[n[:12] for n in class_names]).plot(xticks_rotation=45)
    # ตั้งชื่อภาพ (แถว = ประเภทจริง) แล้วแสดงภาพ
    plt.title("Confusion matrix (rows = truth)"); plt.show()

    # ------------------------------------------------------------------------
    # สร้างฟังก์ชัน demo สำหรับการสาธิตสด: แสดงภาพทดสอบ 1 ภาพพร้อมผลการทำนาย
    #   - image_number : หมายเลขภาพในชุดทดสอบ
    # ------------------------------------------------------------------------
    def demo(image_number):
        """แสดงภาพทดสอบ 1 ภาพที่โมเดลไม่เคยเห็น พร้อมผลการทำนาย"""
        # แปลงหมายเลขภาพเป็นจำนวนเต็ม
        i = int(image_number)
        # ดึงภาพต้นฉบับ (ก่อนแปลงด้วย tf) ลำดับที่ i จากชุดทดสอบ (.imgs เก็บภาพทั้งหมดเป็นอาร์เรย์)
        img = test_ds.imgs[i]
        # แสดงภาพ (4 คำสั่งคั่นด้วย ;)
        #   - plt.figure(figsize=(2.5, 2.5))  : สร้างพื้นที่ภาพขนาด 2.5 x 2.5 นิ้ว
        #   - plt.imshow(img, cmap=...)       : วาดภาพ; ภาพขาวดำ (n_ch == 1) ใช้โทนสีเทา "gray"
        #                                        ภาพสีใช้ None (สีตามจริง)
        #   - plt.axis("off")                 : ซ่อนแกนและตัวเลขรอบภาพ
        #   - plt.show()                      : แสดงภาพ
        plt.figure(figsize=(2.5, 2.5)); plt.imshow(img, cmap="gray" if n_ch == 1 else None); plt.axis("off"); plt.show()
        # ความน่าจะเป็นของทุกประเภทสำหรับภาพนี้ (คำนวณไว้แล้วในขั้นที่ 4)
        p = probs[i]
        # ส่งกลับข้อความสรุปผล
        #   - class_names[p.argmax()] : ชื่อประเภทที่ทำนาย (ความน่าจะเป็นสูงสุด)
        #   - p.max():.0%             : ความมั่นใจ = ความน่าจะเป็นสูงสุด แสดงเป็นร้อยละ
        #   - class_names[y_true[i]]  : ชื่อประเภทจริงของภาพ
        return (f"ภาพหมายเลข {i}: โมเดลทำนายว่า '{class_names[p.argmax()]}' (ความมั่นใจ {p.max():.0%}) "
                f"ประเภทจริง: '{class_names[y_true[i]]}'")
    # แจ้งว่าฟังก์ชัน demo พร้อมใช้ และช่วงหมายเลขภาพที่ใช้ได้
    print(f"\nฟังก์ชัน demo() พร้อมใช้งาน: ทดลอง demo(0) ถึง demo({len(test_ds) - 1})")
```

💡 **คำอธิบายผลลัพธ์** โมเดลได้เรียนรู้ลักษณะทั่วไปของภาพ (ขอบ พื้นผิว) จากภาพทั่วไปจำนวนหลายล้านภาพไว้แล้ว การฝึกในขั้นนี้เป็นเพียงการปรับโมเดลให้เข้ากับภาพทางการแพทย์ ซึ่งเรียกว่า **การเรียนรู้แบบถ่ายโอน (transfer learning)** ค่า **ความแม่นยำ (accuracy)** เพียงค่าเดียวอาจทำให้เข้าใจผิดเมื่อมีประเภทใดประเภทหนึ่งเป็นส่วนใหญ่ จึงต้องเปรียบเทียบกับค่าอ้างอิง "ทายประเภทที่พบบ่อยที่สุดเสมอ" และพิจารณาอัตราการจำแนกถูกรายประเภทร่วมกับ **เมทริกซ์ความสับสน (confusion matrix)** ซึ่งแสดงว่าประเภทใดถูกจำแนกสลับกัน

```python
# @title 📦 เซลล์ช่วยสำหรับแนวทาง nocode: ดาวน์โหลดภาพเอกซเรย์ตัวอย่างสำหรับ Teachable Machine (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (แนวทาง nocode — ไม่บังคับ)
#   เตรียมภาพเอกซเรย์ทรวงอกตัวอย่างจากชุด PneumoniaMNIST (ขนาด 64x64 พิกเซล)
#   สำหรับนำไปฝึกโมเดลใน Teachable Machine ผ่านเว็บเบราว์เซอร์
#     - โฟลเดอร์ train/ : ภาพปกติ 40 ภาพ + ภาพปอดอักเสบ 40 ภาพ (ใช้ฝึก)
#     - โฟลเดอร์ test/  : ภาพปกติ 15 ภาพ + ภาพปอดอักเสบ 15 ภาพ (ใช้ทดสอบ ห้ามนำไปฝึก)
#   จากนั้นบีบอัดเป็นไฟล์ nocode_images.zip และดาวน์โหลดลงเครื่องคอมพิวเตอร์ของผู้เรียน
#   ผลลัพธ์ : ไฟล์ nocode_images.zip (เบราว์เซอร์จะเริ่มดาวน์โหลดอัตโนมัติเมื่อรันใน Colab)
# ============================================================================

# ตรวจว่าแนวทางที่เลือกไม่ใช่ "nocode" หรือไม่
if TRACK != "nocode":
    # หากไม่ใช่ ให้แสดงข้อความแล้วข้ามเซลล์นี้
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
# กรณีเลือกแนวทาง nocode
else:
    # นำเข้าไลบรารีมาตรฐานของ Python
    #   - subprocess : สั่งรันโปรแกรมอื่น (ใช้ติดตั้งไลบรารี)
    #   - sys        : ข้อมูลของ Python ที่กำลังทำงาน
    #   - os         : จัดการไฟล์และโฟลเดอร์
    #   - shutil     : คัดลอก/บีบอัดไฟล์และโฟลเดอร์
    import subprocess, sys, os, shutil
    # ติดตั้งไลบรารี medmnist ด้วย pip (-q = แสดงข้อความน้อย; check=True = หยุดหากติดตั้งไม่สำเร็จ)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "medmnist"], check=True)
    # นำเข้าคลาส PneumoniaMNIST (ชุดภาพเอกซเรย์ทรวงอก ปกติ/ปอดอักเสบ) จาก medmnist
    from medmnist import PneumoniaMNIST
    # สร้าง dict จับคู่หมายเลขประเภทกับชื่อโฟลเดอร์: 0 -> "normal" (ปกติ), 1 -> "pneumonia" (ปอดอักเสบ)
    names = {0: "normal", 1: "pneumonia"}
    # วนซ้ำ 2 รอบ สำหรับส่วนฝึกและส่วนทดสอบ
    #   - split     : "train" หรือ "test"
    #   - per_class : จำนวนภาพที่ต้องการต่อประเภท (ฝึก 40, ทดสอบ 15)
    for split, per_class in [("train", 40), ("test", 15)]:
        # โหลดชุดภาพส่วนนั้น
        #   - split=split   : ส่วนข้อมูล (train/test)
        #   - download=True : ดาวน์โหลดหากยังไม่มี
        #   - size=64       : ภาพขนาด 64x64 พิกเซล
        ds = PneumoniaMNIST(split=split, download=True, size=64)   # ภาพเอกซเรย์ทรวงอกขนาด 64x64 พิกเซล
        # ตัวนับจำนวนภาพที่บันทึกแล้วของแต่ละประเภท เริ่มที่ 0 ทั้งคู่
        count = {0: 0, 1: 0}
        # วนซ้ำทีละภาพในชุดข้อมูล: img = ภาพ (รูปแบบ PIL), label = ป้ายกำกับประเภท
        for img, label in ds:
            # ดึงหมายเลขประเภทออกมาเป็นจำนวนเต็ม (label เป็นอาร์เรย์ 1 ค่า จึงใช้ [0])
            lab = int(label[0])
            # ถ้าประเภทนี้ยังบันทึกไม่ครบจำนวนที่ต้องการ (< คือ น้อยกว่า)
            if count[lab] < per_class:
                # สร้างชื่อโฟลเดอร์ปลายทาง เช่น "nocode_images/train/normal"
                folder = f"nocode_images/{split}/{names[lab]}"
                # สร้างโฟลเดอร์ (รวมโฟลเดอร์ย่อยทั้งหมด)
                #   - exist_ok=True : หากมีโฟลเดอร์อยู่แล้วก็ไม่ต้องแจ้งข้อผิดพลาด
                os.makedirs(folder, exist_ok=True)
                # บันทึกภาพเป็นไฟล์ .png เช่น "train_normal_00.png"
                #   - {count[lab]:02d} : แสดงเลขลำดับเป็นจำนวนเต็ม 2 หลัก เติม 0 ด้านหน้า (00, 01, ...)
                img.save(f"{folder}/{split}_{names[lab]}_{count[lab]:02d}.png")
                # เพิ่มตัวนับของประเภทนี้ขึ้น 1
                count[lab] += 1
            # ถ้าทั้งสองประเภทบันทึกครบแล้ว (ค่าน้อยที่สุดของตัวนับ >= จำนวนที่ต้องการ)
            if min(count.values()) >= per_class:
                # break : ออกจากการวนซ้ำทันที (ไม่ต้องอ่านภาพที่เหลือ)
                break
    # บีบอัดโฟลเดอร์เป็นไฟล์ zip
    #   - "nocode_images" (ตัวแรก) : ชื่อไฟล์ผลลัพธ์ (ไม่รวมนามสกุล) ได้ nocode_images.zip
    #   - "zip"                    : รูปแบบไฟล์บีบอัด
    #   - "nocode_images" (ตัวที่สาม) : โฟลเดอร์ที่จะบีบอัด
    shutil.make_archive("nocode_images", "zip", "nocode_images")
    # แจ้งว่าสร้างไฟล์เสร็จแล้ว
    print("สร้างไฟล์ nocode_images.zip แล้ว: train/ = 40 + 40 ภาพ, test/ = 15 + 15 ภาพ")
    # ตรวจว่ากำลังรันบน Google Colab หรือไม่ (sys.modules คือรายชื่อไลบรารีที่โหลดอยู่)
    if "google.colab" in sys.modules:
        # นำเข้าเครื่องมือจัดการไฟล์ของ Colab
        from google.colab import files
        # สั่งให้เบราว์เซอร์ดาวน์โหลดไฟล์ zip ลงเครื่องคอมพิวเตอร์ของผู้เรียน
        files.download("nocode_images.zip")   # แตกไฟล์ในเครื่องคอมพิวเตอร์ แล้วอัปโหลดโฟลเดอร์ไปยัง Teachable Machine
```

```python
# @title ▶️ คำนวณผลการประเมินของแนวทาง nocode (ข้ามโดยอัตโนมัติหากเลือกแนวทางอื่น)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (แนวทาง nocode)
#   นำผลการนับ TP, FN, FP, TN จากแบบฝึกหัดที่ 2D มาคำนวณตัวชี้วัดชุดเดียวกับ
#   ที่ใช้ประเมินการตรวจคัดกรองทางคลินิก ได้แก่
#     - ความไว (sensitivity)    = TP / (TP + FN)  สัดส่วนผู้ป่วยจริงที่ตรวจพบ
#     - ความจำเพาะ (specificity) = TN / (TN + FP)  สัดส่วนกรณีปกติที่ระบุว่าปกติได้ถูกต้อง
#     - ความแม่นยำ (accuracy)   = (TP + TN) / ทั้งหมด
#       เทียบกับค่าอ้างอิง "ทายประเภทที่พบบ่อยที่สุดเสมอ"
#   และสร้างฟังก์ชัน demo() ซึ่งแสดงคำแนะนำการสาธิตสดใน Teachable Machine
#   ผลลัพธ์ : ตัวชี้วัดทั้ง 3 ค่าแสดงเป็นร้อยละ
# ============================================================================

# ตรวจว่าแนวทางที่เลือกไม่ใช่ "nocode" หรือไม่
if TRACK != "nocode":
    # หากไม่ใช่ ให้แสดงข้อความแล้วข้ามเซลล์นี้
    print(f"ข้ามเซลล์นี้: แนวทางที่เลือกคือ '{TRACK}'")
# กรณีเลือกแนวทาง nocode
else:
    # จำนวนภาพทดสอบทั้งหมด = ผลรวมทั้ง 4 ช่อง
    total = TP + FN + FP + TN
    # ความไว = TP ÷ (TP + FN)
    #   - ... if TP + FN else float("nan") : ถ้า TP + FN เป็น 0 (ไม่มีภาพผลบวกเลย) จะหารไม่ได้
    #     จึงให้ค่าเป็น NaN (Not a Number = ไม่มีค่า) แทน; float("nan") สร้างค่า NaN
    sensitivity = TP / (TP + FN) if TP + FN else float("nan")
    # ความจำเพาะ = TN ÷ (TN + FP) และป้องกันการหารด้วยศูนย์แบบเดียวกัน
    specificity = TN / (TN + FP) if TN + FP else float("nan")
    # ความแม่นยำ = จำนวนที่ทายถูก (TP + TN) ÷ ทั้งหมด
    accuracy = (TP + TN) / total
    # ค่าอ้างอิง "ทายประเภทที่พบบ่อยที่สุดเสมอ"
    #   - TP + FN : จำนวนภาพผลบวกจริงทั้งหมด; TN + FP : จำนวนภาพปกติทั้งหมด
    #   - max(...) : เลือกจำนวนที่มากกว่า แล้วหารด้วยทั้งหมด
    majority = max(TP + FN, TN + FP) / total
    # แสดงจำนวนภาพทดสอบ
    print(f"ทดสอบกับภาพที่โมเดลไม่เคยเห็น {total} ภาพ")
    # แสดงความไวเป็นร้อยละ (:.0% = ร้อยละไม่มีทศนิยม)
    print(f"ความไว (sensitivity; ตรวจพบ {POSITIVE_CLASS}): {sensitivity:.0%}")
    # แสดงความจำเพาะเป็นร้อยละ
    print(f"ความจำเพาะ (specificity; ระบุกรณีปกติได้ถูกต้อง): {specificity:.0%}")
    # แสดงความแม่นยำ เทียบกับค่าอ้างอิง
    print(f"ความแม่นยำ (accuracy): {accuracy:.0%}  (ค่าอ้างอิง 'ทายประเภทที่พบบ่อยที่สุดเสมอ': {majority:.0%})")

    # สร้างฟังก์ชัน demo สำหรับแนวทาง nocode (การสาธิตจริงทำใน Teachable Machine)
    #   - _=None : รับค่า 1 ค่า (ชื่อ _ หมายถึงไม่ได้ใช้) ค่าเริ่มต้น None
    #              เพื่อให้เรียก demo(MY_EXAMPLE) ได้เหมือนแนวทางอื่น
    def demo(_=None):
        # ส่งกลับข้อความคำแนะนำ โดยต่อข้อความ 3 ส่วนด้วยเครื่องหมาย +
        #   - ส่วนที่ 2 : ถ้ามี NOCODE_LINK จะแทรกลิงก์ในวงเล็บ มิฉะนั้นเป็นข้อความว่าง ""
        return ("การสาธิตสด: สลับการแชร์หน้าจอใน Zoom ไปยังแท็บ Teachable Machine"
                + (f" ({NOCODE_LINK})" if NOCODE_LINK else "")
                + " เปิดแผง Preview แล้วอัปโหลดภาพทดสอบ 1 ภาพที่โมเดลไม่เคยเห็น")
```

```python
# @title 🎛️ ไม่บังคับ: เว็บแอปขนาดเล็กสำหรับการสาธิต (Gradio) — กำหนด LAUNCH_WEB_DEMO = True เพื่อใช้งาน
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (ไม่บังคับ)
#   สร้างเว็บแอปขนาดเล็กด้วยไลบรารี Gradio ที่มีช่องกรอกข้อมูล 1 ช่องและช่องแสดงผล 1 ช่อง
#   เมื่อกรอกหมายเลขผู้ป่วย/หมายเลขภาพ/คำถาม แล้วกดส่ง เว็บแอปจะเรียก demo(...)
#   และแสดงผลลัพธ์ เหมาะสำหรับการสาธิตสดที่ดูเป็นมิตรกับผู้ใช้มากขึ้น
#   ค่าเริ่มต้นปิดไว้ (LAUNCH_WEB_DEMO = False) หากต้องการใช้ ให้เปลี่ยนเป็น True แล้วรันอีกครั้ง
#   ผลลัพธ์ : หน้าเว็บแอปแสดงใต้เซลล์ (หรือข้อความแจ้งว่ายังไม่ได้เปิด)
# ============================================================================

# สวิตช์เปิด/ปิดเว็บแอป: False = ไม่เปิด (ค่าเริ่มต้น), True = เปิด (True/False คือค่าความจริงของ Python)
LAUNCH_WEB_DEMO = False

# ถ้าสวิตช์เป็น True ให้สร้างเว็บแอป
if LAUNCH_WEB_DEMO:
    # นำเข้า subprocess และ sys สำหรับติดตั้งไลบรารีหากจำเป็น
    import subprocess, sys
    # ลองนำเข้า gradio ก่อน (หากติดตั้งไว้แล้ว)
    try:
        # นำเข้าไลบรารี gradio (สร้างเว็บแอปจากฟังก์ชัน Python) ตั้งชื่อย่อว่า gr
        import gradio as gr
    # หากยังไม่มีไลบรารีนี้ จะเกิด ImportError (ข้อผิดพลาดการนำเข้า)
    except ImportError:
        # ติดตั้ง gradio ด้วย pip (-q แสดงข้อความน้อย; check=True หยุดหากไม่สำเร็จ)
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "gradio"], check=True)
        # นำเข้าอีกครั้งหลังติดตั้ง
        import gradio as gr
    # เลือกข้อความกำกับช่องกรอกตามแนวทาง: dict จับคู่ แนวทาง -> ข้อความ แล้วเลือกด้วย [TRACK]
    hint = {"tabular": "หมายเลขผู้ป่วย", "image": "หมายเลขภาพ", "llm": "คำถาม (ภาษาอังกฤษ)", "nocode": "(ไม่ใช้)"}[TRACK]
    # สร้างหน้าเว็บแอปด้วย gr.Interface(...) แล้วเปิดใช้งานด้วย .launch()
    #   - fn=lambda x: str(demo(x)) : ฟังก์ชันที่ทำงานเมื่อกดส่ง
    #       * lambda คือการเขียนฟังก์ชันสั้น ๆ แบบไม่ต้องตั้งชื่อ: รับ x (ข้อความที่กรอก)
    #         แล้วส่งให้ demo(x) และแปลงผลเป็นข้อความด้วย str(...)
    #   - inputs=gr.Textbox(label=hint)                        : ช่องกรอกข้อความ 1 ช่อง มีป้ายกำกับตาม hint
    #   - outputs=gr.Textbox(label="ผลลัพธ์ของต้นแบบ")          : ช่องแสดงผลลัพธ์
    #   - title="..."                                          : ชื่อเว็บแอป (ระบุชัดว่าใช้เพื่อการเรียนการสอน ไม่ใช้ทางคลินิก)
    #   - .launch()                                            : เปิดเว็บแอป (ใน Colab จะแสดงใต้เซลล์)
    gr.Interface(fn=lambda x: str(demo(x)), inputs=gr.Textbox(label=hint), outputs=gr.Textbox(label="ผลลัพธ์ของต้นแบบ"),
                 title="ต้นแบบ AI ด้านสุขภาพดิจิทัล (เพื่อการเรียนการสอน ไม่ใช้ทางคลินิก)").launch()
# กรณีสวิตช์เป็น False
else:
    # แจ้งวิธีเปิดใช้งาน
    print("ยังไม่ได้เปิดเว็บแอปสาธิต (ไม่บังคับ) หากต้องการใช้งาน ให้กำหนด LAUNCH_WEB_DEMO = True แล้วรันเซลล์นี้อีกครั้ง")
```

โจทย์คือการทำนายว่าผู้ป่วยเบาหวานที่รับไว้ในโรงพยาบาลจะ **กลับมารักษาซ้ำภายใน 30 วัน** หรือไม่ (Diabetes 130-US hospitals, 101,766 ครั้งของการนอนโรงพยาบาล, ปี 1999–2008) ชุดทดสอบที่กันไว้แบ่ง **ตามผู้ป่วย** (20 % ของผู้ป่วย = 20,352 ครั้ง) ผู้ป่วยรายเดียวกันจึงไม่ปรากฏทั้งในข้อมูลฝึกและข้อมูลทดสอบ ซึ่งเป็นการป้องกันการรั่วไหลของข้อมูล (data leakage) ที่พบบ่อยในงานวิจัยด้วยข้อมูล EHR และทำให้ผลลัพธ์สูงเกินจริง **ตัวชี้วัด: ROC AUC** (โมเดลที่ตีพิมพ์มีค่าประมาณ 0.65–0.70 และกลุ่มผลบวกมีเพียงประมาณ 11 % ค่า accuracy จึงไม่มีความหมาย) ไฟล์ที่ส่ง: `lab10_submission.csv` มีคอลัมน์ `id,prob` (`id` = `encounter_id` จาก `readmission/test_ids.csv`, `prob` = ความน่าจะเป็นที่จะกลับมารักษาซ้ำภายใน 30 วัน)

โมเดลอ้างอิง (baseline) ด้านล่างทำงานได้ทันที โดยสร้างไฟล์ส่งที่ถูกต้อง และจะส่งผลไปยังตารางอันดับเมื่อผู้เรียนระบุ `NAME` และ `JOIN_CODE` ในเซลล์การตั้งค่าแล้ว (จากนั้นรันเซลล์การตั้งค่าอีกครั้ง)

> **หลักความซื่อสัตย์ทางวิชาการ (Honour code)** ป้ายกำกับ (label) ที่ซ่อนไว้มาจากชุดข้อมูลสาธารณะ การค้นหาป้ายกำกับเหล่านั้นจึงทำได้ง่ายแต่ไม่มีประโยชน์ใด ๆ ตารางอันดับมีไว้เพื่อการเรียนรู้ และโน้ตบุ๊กของผู้เรียนจะได้รับการประเมินด้วย ห้ามแก้ไขผลการทำนายด้วยมือ

```python
# @title 🏆 โมเดลอ้างอิง: จำนวนการใช้บริการ + gradient boosting → lab10_submission.csv
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทายบนตารางอันดับ — ไม่บังคับ)
#   สร้าง "โมเดลอ้างอิง (baseline)" สำหรับโจทย์ทำนายการกลับมารักษาซ้ำภายใน 30 วัน (30-day readmission)
#   1) ดาวน์โหลดชุดข้อมูล Diabetes 130-US Hospitals และสร้างคอลัมน์ผลลัพธ์ y (1 = กลับมาภายใน 30 วัน)
#   2) แยกข้อมูลเป็นส่วนฝึก (lb_train) และส่วนทดสอบของตารางอันดับ (lb_test ซึ่งซ่อนป้ายกำกับ)
#   3) ใช้ตัวแปรตัวเลข 8 ตัว (จำนวนวันนอน จำนวนการตรวจ จำนวนยา ประวัติการมารับบริการ ฯลฯ)
#      ฝึก gradient boosting และวัดผลด้วยการตรวจสอบไขว้แบบแบ่งกลุ่มตามผู้ป่วย (grouped cross-validation)
#      เพื่อป้องกันการรั่วไหลของข้อมูล (data leakage)
#   4) ทำนายความน่าจะเป็นของชุดทดสอบ บันทึกไฟล์ lab10_submission.csv (คอลัมน์ id, prob)
#   5) ส่งผลไปยังตารางอันดับ หากกรอก NAME และ JOIN_CODE ในเซลล์การตั้งค่าแล้ว
#   ผลลัพธ์ : ขนาดข้อมูล ค่า AUC จากการตรวจสอบไขว้ ไฟล์ส่งผล และ (ถ้ามี) คะแนนบนตารางอันดับ
# ============================================================================

# ----------------------------------------------------------------------------
# นำเข้าไลบรารี
# ----------------------------------------------------------------------------
# นำเข้าไลบรารี 4 ตัว
#   - io, zipfile : จัดการข้อมูลในหน่วยความจำและเปิดไฟล์ zip
#   - requests    : ดาวน์โหลดข้อมูลจากอินเทอร์เน็ต
#   - pandas as pd: จัดการข้อมูลแบบตาราง
import io, zipfile, requests, pandas as pd
# นำเข้าโมเดล gradient boosting จาก scikit-learn
from sklearn.ensemble import HistGradientBoostingClassifier
# นำเข้าเครื่องมือตรวจสอบไขว้ (cross-validation)
#   - GroupKFold      : แบ่งข้อมูลเป็น k ส่วนตามกลุ่ม (ผู้ป่วยรายเดียวกันอยู่ส่วนเดียวกันเสมอ)
#   - cross_val_score : ฝึกและวัดผลโมเดลวนตามแต่ละส่วน แล้วคืนคะแนนของทุกส่วน
from sklearn.model_selection import GroupKFold, cross_val_score

# ----------------------------------------------------------------------------
# ขั้นที่ 1) ดาวน์โหลดข้อมูลและสร้างคอลัมน์ผลลัพธ์
# ----------------------------------------------------------------------------
# เก็บ URL ของไฟล์ zip ชุดข้อมูลไว้ในตัวแปร URL
URL = "https://archive.ics.uci.edu/static/public/296/diabetes+130-us+hospitals+for+years+1999-2008.zip"
# ดาวน์โหลดไฟล์ zip และเปิดในหน่วยความจำ
#   - requests.get(URL, timeout=120) : ดาวน์โหลด รอนานสุด 120 วินาที
#   - .content                       : ข้อมูลไบต์ของไฟล์
#   - io.BytesIO(...)                : ห่อข้อมูลไบต์ให้ใช้เหมือนไฟล์
#   - zipfile.ZipFile(...)           : เปิดเป็นไฟล์ zip
z = zipfile.ZipFile(io.BytesIO(requests.get(URL, timeout=120).content))
# อ่านไฟล์ diabetic_data.csv จากใน zip เป็นตาราง rdf
#   - next(n for n in z.namelist() if n.endswith(...)) : หาชื่อไฟล์แรกที่ลงท้ายด้วย diabetic_data.csv
#   - z.open(...)       : เปิดไฟล์นั้น
#   - low_memory=False  : อ่านทั้งไฟล์ก่อนกำหนดชนิดข้อมูลของคอลัมน์ (ลดคำเตือนชนิดข้อมูลปนกัน)
rdf = pd.read_csv(z.open(next(n for n in z.namelist() if n.endswith("diabetic_data.csv"))), low_memory=False)
# สร้างคอลัมน์ผลลัพธ์ y: 1 ถ้า readmitted เท่ากับ "<30" (กลับมาภายใน 30 วัน) มิฉะนั้น 0
#   - .astype(int) : แปลง True/False เป็น 1/0
rdf["y"] = (rdf["readmitted"] == "<30").astype(int)

# ----------------------------------------------------------------------------
# ขั้นที่ 2) แยกส่วนฝึกและส่วนทดสอบของตารางอันดับ
# ----------------------------------------------------------------------------
# ดาวน์โหลดรายการรหัสการนอนโรงพยาบาลของชุดทดสอบ (ฟังก์ชัน lab_data จากเซลล์การตั้งค่า) แล้วเลือกคอลัมน์ "id"
test_ids = pd.read_csv(lab_data("readmission/test_ids.csv"))["id"]
# สร้างคอลัมน์จริง/เท็จ is_test: True หากรหัสการนอนโรงพยาบาล (encounter_id) อยู่ในชุดทดสอบ
#   - .isin(test_ids) : ตรวจว่าแต่ละค่าอยู่ในรายการ test_ids หรือไม่
is_test = rdf["encounter_id"].isin(test_ids)
# ส่วนฝึก = แถวที่ "ไม่ใช่" ชุดทดสอบ (~ กลับค่าจริง/เท็จ)
lb_train = rdf[~is_test]
# ส่วนทดสอบ = แถวที่อยู่ในชุดทดสอบ และลบคอลัมน์คำตอบ (readmitted, y) ออก เพื่อไม่ให้ใช้คำตอบโดยไม่ตั้งใจ
#   - .drop(columns=[...]) : ลบคอลัมน์ตามรายชื่อ
lb_test = rdf[is_test].drop(columns=["readmitted", "y"])        # ป้ายกำกับถูกซ่อน ห้ามนำมาใช้
# แสดงขนาดตาราง (.shape = (จำนวนแถว, จำนวนคอลัมน์)) และสัดส่วนผลบวกในข้อมูลฝึก
#   - lb_train["y"].mean() : ค่าเฉลี่ยของ 0/1 = สัดส่วนผู้ที่กลับมารักษาซ้ำ
#   - round(..., 3)        : ปัดเศษทศนิยม 3 ตำแหน่ง
print(lb_train.shape, lb_test.shape, "สัดส่วนผลบวกในข้อมูลฝึก:", round(lb_train["y"].mean(), 3))

# ----------------------------------------------------------------------------
# ขั้นที่ 3) เลือกตัวแปร ฝึก และวัดผลด้วยการตรวจสอบไขว้แบบแบ่งกลุ่มตามผู้ป่วย
# ----------------------------------------------------------------------------
# list ชื่อคอลัมน์ตัวเลข 8 ตัวที่ใช้เป็นตัวแปรทำนาย (features)
#   time_in_hospital = จำนวนวันนอน, num_lab_procedures = จำนวนการตรวจทางห้องปฏิบัติการ,
#   num_procedures = จำนวนหัตถการ, num_medications = จำนวนยา,
#   number_outpatient/emergency/inpatient = จำนวนครั้งที่มาแผนกผู้ป่วยนอก/ฉุกเฉิน/นอนโรงพยาบาลในปีก่อนหน้า,
#   number_diagnoses = จำนวนการวินิจฉัย
features = ["time_in_hospital", "num_lab_procedures", "num_procedures", "num_medications",
            "number_outpatient", "number_emergency", "number_inpatient", "number_diagnoses"]
# สร้างโมเดล gradient boosting (random_state=42 ให้ผลเหมือนเดิมทุกครั้งที่รัน)
clf = HistGradientBoostingClassifier(random_state=42)
# สร้างตัวแบ่งข้อมูลแบบกลุ่ม 5 ส่วน (n_splits=5): ฝึกด้วย 4 ส่วน ทดสอบด้วย 1 ส่วน วนครบ 5 รอบ
cv = GroupKFold(n_splits=5)                                      # แบ่งตามผู้ป่วย เช่นเดียวกับชุดทดสอบของตารางอันดับ
# รันการตรวจสอบไขว้ ได้ค่า AUC 5 ค่า (1 ค่าต่อรอบ) เก็บใน auc
#   - clf                     : โมเดลที่จะฝึก
#   - lb_train[features]      : ตัวแปรทำนาย (เลือกเฉพาะ 8 คอลัมน์)
#   - lb_train["y"]           : คำตอบ
#   - cv=cv                   : วิธีแบ่งข้อมูล (GroupKFold 5 ส่วน)
#   - groups=lb_train["patient_nbr"] : รหัสผู้ป่วย ใช้กำหนดกลุ่ม
#   - scoring="roc_auc"       : ใช้ AUROC เป็นคะแนน
auc = cross_val_score(clf, lb_train[features], lb_train["y"], cv=cv, groups=lb_train["patient_nbr"], scoring="roc_auc")
# แสดงค่าเฉลี่ย AUC ของ 5 รอบ (.mean()) ปัดทศนิยม 3 ตำแหน่ง (.round(3))
print("grouped CV AUC ของโมเดลอ้างอิง:", auc.mean().round(3))

# ----------------------------------------------------------------------------
# ขั้นที่ 4) ฝึกด้วยข้อมูลฝึกทั้งหมด ทำนายชุดทดสอบ และบันทึกไฟล์ส่งผล
# ----------------------------------------------------------------------------
# ฝึกโมเดลด้วยข้อมูลฝึกทั้งหมด
clf.fit(lb_train[features], lb_train["y"])
# ทำนายความน่าจะเป็นที่จะกลับมารักษาซ้ำของชุดทดสอบ ([:, 1] = คอลัมน์ความน่าจะเป็นของผลบวก)
prob = clf.predict_proba(lb_test[features])[:, 1]
# สร้างตารางส่งผล 2 คอลัมน์ แล้วบันทึกเป็นไฟล์ CSV
#   - "id"   : รหัสการนอนโรงพยาบาล (.values ดึงเป็นอาร์เรย์ค่าล้วน ไม่มีเลขแถว)
#   - "prob" : ความน่าจะเป็นที่ทำนาย
#   - .to_csv("lab10_submission.csv", index=False) : บันทึกไฟล์ โดยไม่บันทึกเลขแถว
pd.DataFrame({"id": lb_test["encounter_id"].values, "prob": prob}).to_csv("lab10_submission.csv", index=False)
# แจ้งว่าบันทึกไฟล์แล้ว พร้อมจำนวนแถว
print("บันทึกไฟล์ lab10_submission.csv แล้ว จำนวน", len(prob), "แถว")

# ----------------------------------------------------------------------------
# ขั้นที่ 5) ส่งผลไปยังตารางอันดับ (เฉพาะเมื่อกรอก NAME และ JOIN_CODE แล้ว)
# ----------------------------------------------------------------------------
# ตรวจว่ากรอกทั้ง NAME และ JOIN_CODE แล้ว (ข้อความไม่ว่าง) หรือไม่
if NAME and JOIN_CODE:
    # ส่งไฟล์ไปยังตารางอันดับด้วยฟังก์ชันจากเซลล์การตั้งค่า (จะแสดงคะแนนและอันดับ)
    submit_to_leaderboard("lab10_submission.csv")
# กรณียังไม่ได้กรอก
else:
    # แจ้งขั้นตอนที่ต้องทำเพื่อส่งผล
    print("ยังไม่ได้ส่งผล: โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่า รันเซลล์การตั้งค่าอีกครั้ง แล้วรันเซลล์นี้ใหม่")
```

**แนวทางปรับปรุงให้ดีกว่าโมเดลอ้างอิง**

- เพิ่มตัวแปรเชิงกลุ่ม (categorical features) เช่น ช่วงอายุ `age`, `admission_type_id`, `discharge_disposition_id` (เช่น จำหน่ายไปยังสถานดูแล เทียบกับ กลับบ้าน), `admission_source_id`, `A1Cresult`, `insulin`, `change`, `diabetesMed`
- จัดกลุ่มรหัส ICD-9 ใน `diag_1`/`diag_2`/`diag_3` ตามหมวด (ระบบไหลเวียนโลหิต 390–459, ระบบหายใจ 460–519, เบาหวาน 250.xx …)
- ประวัติการใช้บริการ: จำนวนครั้งรวม = outpatient + emergency + inpatient และจำนวนครั้งที่ผู้ป่วยรายเดียวกันเคยเข้ารับบริการก่อนหน้า
- ปรับไฮเปอร์พารามิเตอร์ (hyperparameter) ของโมเดล (`learning_rate`, `max_leaf_nodes`, `l2_regularization`) โดยใช้ `GroupKFold` แบ่งตาม `patient_nbr` ทุกครั้ง
- ผู้ป่วยที่เสียชีวิตหรือจำหน่ายไปยังสถานดูแลระยะสุดท้าย (hospice) (`discharge_disposition_id` 11, 13, 14, 19–21) ไม่สามารถกลับมารักษาซ้ำได้ ซึ่งเป็นลักษณะเฉพาะที่ทราบกันดีของชุดข้อมูลนี้และควรนำมาพิจารณา

```python
# 🏆 ไม่บังคับ — โมเดลที่ผู้เรียนปรับปรุง (รันได้ทันที; เพิ่มคอลัมน์ใน EXTRA_CATEGORICAL เพื่อเพิ่มคะแนน)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทายบนตารางอันดับ — ไม่บังคับ)
#   ต่อยอดจากโมเดลอ้างอิงในเซลล์ก่อนหน้า โดยเพิ่ม "ตัวแปรเชิงกลุ่ม (categorical features)"
#   เช่น ช่วงอายุ ประเภทการจำหน่าย ผล HbA1c การใช้อินซูลิน ตามที่ผู้เรียนระบุใน EXTRA_CATEGORICAL
#     - แปลงตัวแปรเชิงกลุ่มเป็นคอลัมน์ 0/1 (one-hot) โดยเก็บ 30 ค่าที่พบบ่อยที่สุด ที่เหลือรวมเป็น "other"
#     - วัดผลด้วยการตรวจสอบไขว้แบบแบ่งกลุ่มตามผู้ป่วย (GroupKFold) เทียบกับโมเดลอ้างอิง
#     - หากกำหนด SUBMIT_IMPROVED = True จะฝึกโมเดลใหม่ เขียนทับไฟล์ส่งผล และส่งไปยังตารางอันดับ
#   ต้องรันเซลล์โมเดลอ้างอิงก่อน (ใช้ตัวแปร features, lb_train, lb_test, auc จากเซลล์นั้น)
#   ผลลัพธ์ : ค่า AUC ของโมเดลที่ปรับปรุงเทียบกับโมเดลอ้างอิง และจำนวนตัวแปร
# ============================================================================

# list ชื่อคอลัมน์เชิงกลุ่มที่จะเพิ่ม (ค่าเริ่มต้นว่าง = ใช้เฉพาะตัวแปรตัวเลขเหมือนโมเดลอ้างอิง)
#   ผู้เรียนสามารถเพิ่มชื่อคอลัมน์ในเครื่องหมายคำพูดภายใน [ ] ตามตัวอย่างท้ายบรรทัด
EXTRA_CATEGORICAL = []   # เช่น ["age", "discharge_disposition_id", "admission_source_id", "A1Cresult", "insulin", "change", "diabetesMed"]
# สวิตช์ส่งผล: False = วัดผลอย่างเดียว, True = ฝึก บันทึกไฟล์ และส่งผล
SUBMIT_IMPROVED = False  # กำหนดเป็น True เพื่อเขียนทับ lab10_submission.csv ด้วยโมเดลนี้และส่งผล

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน make_X สำหรับสร้างตารางตัวแปรทำนาย
#   - frame    : ตารางข้อมูลที่ต้องการแปลง (ข้อมูลฝึกหรือข้อมูลทดสอบ)
#   - ref=None : ตารางอ้างอิงสำหรับหา "ค่าที่พบบ่อยที่สุด" (ใช้ข้อมูลฝึกเมื่อแปลงข้อมูลทดสอบ
#                เพื่อให้หมวดหมู่ตรงกันและไม่ใช้ข้อมูลทดสอบในการตัดสินใจ) ค่าเริ่มต้น None = ใช้ frame เอง
# ----------------------------------------------------------------------------
def make_X(frame, ref=None):
    """ตัวแปรตัวเลขของโมเดลอ้างอิง + one-hot ของตัวแปรเชิงกลุ่มที่เพิ่ม (30 ค่าที่พบบ่อยที่สุด ที่เหลือเป็น 'other')"""
    # เริ่มต้น list ส่วนประกอบด้วยตารางตัวแปรตัวเลข 8 ตัวเดิม (features จากเซลล์โมเดลอ้างอิง)
    parts = [frame[features]]
    # วนซ้ำทีละคอลัมน์เชิงกลุ่มที่ผู้เรียนเพิ่ม (c = ชื่อคอลัมน์)
    for c in EXTRA_CATEGORICAL:
        # หา 30 ค่าที่พบบ่อยที่สุดของคอลัมน์ c
        #   - (ref if ref is not None else frame) : ใช้ตารางอ้างอิงถ้ามี มิฉะนั้นใช้ frame
        #   - [c].astype(str)                     : เลือกคอลัมน์และแปลงเป็นข้อความ
        #   - .value_counts().index[:30]          : เรียงค่าตามความถี่ แล้วเลือก 30 ค่าแรก
        top = (ref if ref is not None else frame)[c].astype(str).value_counts().index[:30]
        # แปลงคอลัมน์เป็นข้อความ คงค่าที่อยู่ใน top ไว้ ที่เหลือแทนด้วย "other"
        #   - .isin(top)           : True หากค่าอยู่ใน top
        #   - .where(เงื่อนไข, "other") : คงค่าเดิมเมื่อจริง มิฉะนั้นเป็น "other"
        s = frame[c].astype(str).where(frame[c].astype(str).isin(top), "other")
        # แปลงเป็นคอลัมน์ 0/1 (one-hot) แล้วเพิ่มต่อท้าย list parts
        #   - prefix=c  : ตั้งชื่อคอลัมน์ใหม่ขึ้นต้นด้วยชื่อคอลัมน์เดิม เช่น insulin_Up
        #   - dtype=int : ค่าเป็นจำนวนเต็ม 0/1
        parts.append(pd.get_dummies(s, prefix=c, dtype=int))
    # นำทุกส่วนมาต่อกันด้านข้าง (axis=1 = ต่อคอลัมน์) แล้วส่งกลับ
    return pd.concat(parts, axis=1)

# ----------------------------------------------------------------------------
# สร้างตารางตัวแปร ฝึก และวัดผลเทียบกับโมเดลอ้างอิง
# ----------------------------------------------------------------------------
# สร้างตารางตัวแปรทำนายของข้อมูลฝึก
X_lb = make_X(lb_train)
# สร้างตารางตัวแปรทำนายของข้อมูลทดสอบ โดยใช้ข้อมูลฝึกเป็นตารางอ้างอิง (ref=lb_train)
#   - .reindex(columns=X_lb.columns, fill_value=0) : จัดคอลัมน์ให้ตรงกับข้อมูลฝึกทุกประการ
#     คอลัมน์ที่ข้อมูลทดสอบไม่มีจะถูกเติมด้วย 0 และคอลัมน์ที่เกินมาจะถูกตัดทิ้ง
X_lb_test = make_X(lb_test, ref=lb_train).reindex(columns=X_lb.columns, fill_value=0)
# สร้างโมเดล gradient boosting ตัวใหม่ (random_state=42 ให้ผลซ้ำได้)
clf2 = HistGradientBoostingClassifier(random_state=42)
# ตรวจสอบไขว้แบบแบ่งกลุ่มตามผู้ป่วย 5 ส่วน ได้ AUC 5 ค่า
#   - X_lb, lb_train["y"]            : ตัวแปรทำนายและคำตอบ
#   - cv=GroupKFold(n_splits=5)      : แบ่ง 5 ส่วนตามกลุ่ม
#   - groups=lb_train["patient_nbr"] : รหัสผู้ป่วย
#   - scoring="roc_auc"              : ใช้ AUROC เป็นคะแนน
auc2 = cross_val_score(clf2, X_lb, lb_train["y"], cv=GroupKFold(n_splits=5), groups=lb_train["patient_nbr"], scoring="roc_auc")
# แสดง AUC เฉลี่ยของโมเดลที่ปรับปรุง เทียบกับโมเดลอ้างอิง (:.3f = ทศนิยม 3 ตำแหน่ง)
#   และจำนวนตัวแปรทั้งหมด (X_lb.shape[1] = จำนวนคอลัมน์)
print(f"grouped CV AUC ของโมเดลที่ปรับปรุง: {auc2.mean():.3f}  (โมเดลอ้างอิง {auc.mean():.3f}) จำนวนตัวแปร {X_lb.shape[1]}")

# ----------------------------------------------------------------------------
# ส่งผล (เฉพาะเมื่อ SUBMIT_IMPROVED = True)
# ----------------------------------------------------------------------------
# ถ้าสวิตช์ส่งผลเป็น True
if SUBMIT_IMPROVED:
    # ฝึกโมเดลที่ปรับปรุงด้วยข้อมูลฝึกทั้งหมด
    clf2.fit(X_lb, lb_train["y"])
    # สร้างตารางส่งผล (id, prob) แล้วบันทึกเขียนทับ lab10_submission.csv
    #   - "id"   : รหัสการนอนโรงพยาบาลของชุดทดสอบ
    #   - "prob" : clf2.predict_proba(X_lb_test)[:, 1] ความน่าจะเป็นที่จะกลับมารักษาซ้ำ
    #   - index=False : ไม่บันทึกเลขแถว
    pd.DataFrame({"id": lb_test["encounter_id"].values,
                  "prob": clf2.predict_proba(X_lb_test)[:, 1]}).to_csv("lab10_submission.csv", index=False)
    # ถ้ากรอก NAME และ JOIN_CODE แล้ว
    if NAME and JOIN_CODE:
        # ส่งไฟล์ไปยังตารางอันดับ
        submit_to_leaderboard("lab10_submission.csv")
    # กรณียังไม่ได้กรอก
    else:
        # แจ้งว่าบันทึกไฟล์แล้ว แต่ยังส่งผลไม่ได้
        print("บันทึกไฟล์ lab10_submission.csv แล้ว โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่าเพื่อส่งผล")
```

## วัตถุประสงค์การเรียนรู้

- ใช้ผู้ช่วยเขียนโค้ดปัญญาประดิษฐ์ (AI coding assistant) สร้างไปป์ไลน์ (pipeline) การเรียนรู้ของเครื่องที่สมบูรณ์จากคำสั่งภาษาธรรมชาติ (prompt)
- ตรวจสอบ (audit) ความถูกต้องของโค้ดที่ปัญญาประดิษฐ์สร้างขึ้น ได้แก่ การรั่วไหลของข้อมูล (data leakage) การเลือกตัวชี้วัดไม่เหมาะสม และข้อผิดพลาดที่ไม่ปรากฏเป็นข้อความแจ้งเตือน (silent bug)
- พัฒนาขั้นตอนการทำงานของตนเอง: ออกคำสั่ง (prompt) → รัน → ตรวจสอบ → ปรับปรุง
- เข้าใจข้อกำหนดด้านความสามารถในการทำซ้ำ (reproducibility) ของงานวิจัยที่ใช้ปัญญาประดิษฐ์ช่วย

> - 🏆 **โจทย์ท้าทายบนตารางอันดับ — ไม่บังคับ สำหรับผู้เรียนที่เขียนโปรแกรมได้** อยู่ท้ายโน้ตบุ๊ก (notebook) ให้คะแนนอัตโนมัติบน[ตารางอันดับ (leaderboard) ของรายวิชาบน Hugging Face](https://huggingface.co/spaces/wdittaya/aidh-leaderboard) และไม่นับเป็นส่วนหนึ่งของชิ้นงานที่ต้องส่ง

> 🔒 **ข้อกำหนดด้านความเป็นส่วนตัว — โปรดอ่านก่อนเริ่ม** ห้ามนำข้อมูลผู้ป่วยจริง (ชื่อ HN บันทึกทางคลินิก ผลการตรวจทางห้องปฏิบัติการ หรือไฟล์ข้อมูลจากหน่วยงาน) ไปวางใน Gemini, ChatGPT หรือผู้ช่วยปัญญาประดิษฐ์ใด ๆ ให้ *อธิบาย* ลักษณะข้อมูลแทน (ชื่อคอลัมน์ ชนิดข้อมูล ความหมายของ label) โดยไม่วางข้อมูลรายแถว ปฏิบัติการนี้ใช้ชุดข้อมูล (dataset) โรคเบาหวาน Pima ซึ่งเป็นข้อมูลสาธารณะที่ไม่สามารถระบุตัวบุคคลได้

````python
# @title ⚙️ การตั้งค่า — โปรดรันเซลล์นี้ก่อน (NAME และ JOIN_CODE ใช้เฉพาะโจทย์ท้าทายบนตารางอันดับ ซึ่งไม่บังคับ)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้เป็นการ "เตรียมความพร้อม" ก่อนเริ่มปฏิบัติการ โดยทำ 3 อย่าง ได้แก่
#   1) ติดตั้งเครื่องมือ (ไลบรารี) เพิ่มเติมที่จำเป็นต้องใช้ใน Google Colab
#   2) กำหนดค่าพื้นฐาน เช่น รหัสปฏิบัติการ ชื่อที่จะแสดง และรหัสเข้าร่วมตารางอันดับ
#   3) สร้าง "ฟังก์ชัน" (ชุดคำสั่งสำเร็จรูปที่เรียกใช้ซ้ำได้) 2 ตัว คือ
#        - lab_data(...)              : ดาวน์โหลดไฟล์ข้อมูลของรายวิชา
#        - submit_to_leaderboard(...) : ส่งผลการทำนายไปยังตารางอันดับ (ไม่บังคับ)
#   ผู้เรียนเพียงกดรันเซลล์นี้ 1 ครั้ง ไม่จำเป็นต้องแก้ไขโค้ด
#   (ยกเว้นต้องการเข้าร่วมตารางอันดับ ให้กรอก NAME และ JOIN_CODE ด้านล่าง)
# ============================================================================

# ติดตั้งไลบรารีเพิ่มเติมลงในเครื่อง Colab
#   - เครื่องหมาย ! ด้านหน้า หมายถึงสั่งงานระบบปฏิบัติการ (ไม่ใช่คำสั่งภาษา Python)
#   - pip install : คำสั่งติดตั้งไลบรารีของ Python
#   - -q (quiet)  : ให้แสดงข้อความระหว่างติดตั้งให้น้อยที่สุด
#   - gradio_client   : ไลบรารีสำหรับติดต่อกับเว็บแอปตารางอันดับ (Hugging Face Space)
#   - huggingface_hub : ไลบรารีสำหรับดาวน์โหลดไฟล์จากเว็บไซต์ Hugging Face
!pip -q install gradio_client huggingface_hub

# สร้างตัวแปร (ช่องเก็บค่า) HF_OWNER แล้วใส่ข้อความ "wdittaya" ไว้
#   เครื่องหมาย = หมายถึง "นำค่าทางขวา ไปเก็บไว้ในชื่อทางซ้าย"
#   ข้อความในภาษา Python ต้องอยู่ในเครื่องหมายคำพูด " "
HF_OWNER  = "wdittaya"     # บัญชี Hugging Face ที่ใช้โฮสต์ตารางอันดับของรายวิชา (ผู้สอนเป็นผู้กำหนด)
# รหัสของปฏิบัติการนี้ ใช้ระบุว่าผลที่ส่งเป็นของปฏิบัติการใด (ไม่ต้องแก้ไข)
LAB_ID    = "lab02"
# ชื่อที่จะแสดงบนตารางอันดับ: พิมพ์ชื่อไว้ระหว่างเครื่องหมาย " " เช่น NAME = "หมอเอ"
NAME      = ""            # ไม่บังคับ: ชื่อที่จะแสดงบนตารางอันดับแบบสาธารณะ (สามารถใช้นามแฝงได้)
# รหัสเข้าร่วมตารางอันดับ: พิมพ์รหัสที่ผู้สอนแจ้งไว้ระหว่างเครื่องหมาย " "
JOIN_CODE = ""            # ไม่บังคับ: รหัสเข้าร่วมซึ่งผู้สอนประกาศในการเรียนการสอนผ่าน Zoom

# สร้างชื่อเว็บแอปตารางอันดับ โดยนำ HF_OWNER มาต่อกับข้อความ
#   - f"..." (f-string) คือข้อความที่แทรกค่าตัวแปรได้ โดยเขียนชื่อตัวแปรไว้ใน { }
#   - ผลลัพธ์คือ "wdittaya/aidh-leaderboard"
SPACE_ID  = f"{HF_OWNER}/aidh-leaderboard"
# สร้างชื่อคลังข้อมูล (dataset) ของรายวิชาในลักษณะเดียวกัน ผลลัพธ์คือ "wdittaya/aidh-lab-data"
DATA_REPO = f"{HF_OWNER}/aidh-lab-data"

# นำเข้า (import) ฟังก์ชัน hf_hub_download จากไลบรารี huggingface_hub
#   ฟังก์ชันนี้ใช้ดาวน์โหลดไฟล์จากเว็บไซต์ Hugging Face มาเก็บไว้ในเครื่อง
from huggingface_hub import hf_hub_download

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน lab_data สำหรับดาวน์โหลดไฟล์ข้อมูลของรายวิชา
#   - def คือคำสั่งสร้างฟังก์ชันใหม่ ตามด้วยชื่อฟังก์ชันและพารามิเตอร์ในวงเล็บ
#   - path : พารามิเตอร์ (ค่าที่ผู้ใช้ส่งเข้ามา) คือชื่อไฟล์ที่ต้องการ เช่น "pima/test_ids.csv"
#   - บรรทัดที่ย่อหน้าเข้าไปด้านล่าง คือคำสั่งที่อยู่ภายในฟังก์ชัน
# ----------------------------------------------------------------------------
def lab_data(path):
    """ดาวน์โหลดไฟล์ข้อมูลทดสอบสาธารณะของรายวิชา 1 ไฟล์ (เช่น 'pima/test_ids.csv') และคืนค่าตำแหน่งไฟล์ในเครื่อง"""
    # เรียก hf_hub_download(...) เพื่อดาวน์โหลดไฟล์ แล้วส่งตำแหน่งไฟล์ในเครื่องกลับไป (return)
    #   - DATA_REPO           : ชื่อคลังข้อมูลบน Hugging Face ที่เก็บไฟล์
    #   - path                : ชื่อไฟล์ที่ต้องการดาวน์โหลดภายในคลังข้อมูลนั้น
    #   - repo_type="dataset" : ระบุว่าคลังนี้เป็นประเภท "ชุดข้อมูล" (ไม่ใช่โมเดล)
    return hf_hub_download(DATA_REPO, path, repo_type="dataset")

# ----------------------------------------------------------------------------
# สร้างฟังก์ชัน submit_to_leaderboard สำหรับส่งไฟล์ผลการทำนายไปยังตารางอันดับ
#   - path   : ชื่อไฟล์ผลการทำนายที่ต้องการส่ง
#   - lab=LAB_ID : รหัสปฏิบัติการ หากไม่ระบุ จะใช้ค่า LAB_ID ที่กำหนดไว้ด้านบนโดยอัตโนมัติ
# ----------------------------------------------------------------------------
def submit_to_leaderboard(path, lab=LAB_ID):
    """ส่งไฟล์ผลการทำนายไปยังตารางอันดับของรายวิชา และแสดงคะแนนพร้อมอันดับ"""
    # นำเข้าเครื่องมือ 2 ตัวจากไลบรารี gradio_client
    #   - Client      : ใช้เชื่อมต่อกับเว็บแอปตารางอันดับ
    #   - handle_file : ใช้เตรียมไฟล์ในเครื่องให้อยู่ในรูปแบบที่ส่งขึ้นเว็บแอปได้
    from gradio_client import Client, handle_file
    # assert คือการตรวจสอบเงื่อนไข หากเงื่อนไขเป็นเท็จ โปรแกรมจะหยุดและแสดงข้อความเตือน
    #   - NAME and JOIN_CODE : เงื่อนไขว่าต้องกรอกทั้ง NAME และ JOIN_CODE แล้ว (ไม่เป็นข้อความว่าง)
    #   - ข้อความหลังเครื่องหมายจุลภาค คือข้อความเตือนที่จะแสดงเมื่อยังไม่ได้กรอก
    assert NAME and JOIN_CODE, "โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่าก่อน (จำเป็นเฉพาะการส่งผลไปยังตารางอันดับ)"
    # เชื่อมต่อกับเว็บแอปตารางอันดับ แล้วเก็บการเชื่อมต่อไว้ในตัวแปร client
    #   - SPACE_ID      : ชื่อเว็บแอปตารางอันดับที่กำหนดไว้ด้านบน
    #   - verbose=False : ไม่ต้องแสดงข้อความรายละเอียดระหว่างเชื่อมต่อ
    client = Client(SPACE_ID, verbose=False)
    # ส่งข้อมูลไปยังเว็บแอปด้วย client.predict(...) แล้วแสดงผลลัพธ์ (คะแนนและอันดับ) ด้วย print(...)
    #   - lab=lab                 : รหัสปฏิบัติการ
    #   - name=NAME               : ชื่อที่จะแสดงบนตารางอันดับ
    #   - code=JOIN_CODE          : รหัสเข้าร่วม
    #   - file=handle_file(path)  : ไฟล์ผลการทำนายที่เตรียมพร้อมสำหรับส่ง
    #   - api_name="/submit"      : ชื่อช่องทางรับข้อมูลของเว็บแอป (ส่วน "ส่งผล")
    print(client.predict(lab=lab, name=NAME, code=JOIN_CODE, file=handle_file(path), api_name="/submit"))

# แสดงข้อความยืนยันว่าการตั้งค่าเสร็จแล้ว ด้วยฟังก์ชัน print(...) (พิมพ์ข้อความออกทางหน้าจอ)
#   - ข้อความเป็น f-string จึงแทรกค่า SPACE_ID, LAB_ID และ NAME ลงในข้อความได้
#   - NAME or '(ยังไม่ได้ระบุ ...)' : หาก NAME ว่าง จะแสดงข้อความในวงเล็บแทน
print(f"Leaderboard: https://huggingface.co/spaces/{SPACE_ID}  |  lab={LAB_ID}  name={NAME or '(ยังไม่ได้ระบุ — ตารางอันดับไม่บังคับ)'}")
````

````python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ตัวอย่างคำตอบที่ผู้ช่วยปัญญาประดิษฐ์มักให้สำหรับคำสั่งในขั้นที่ 1 (รันได้แต่ไม่สมบูรณ์ โดยเจตนา
# เพื่อใช้ตรวจสอบในขั้นที่ 2 ในกรณีที่โค้ดของผู้เรียนรันไม่สำเร็จ)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้เป็น "ตัวอย่างโค้ดที่ผู้ช่วยปัญญาประดิษฐ์มักสร้างให้" จากคำสั่งในขั้นที่ 1
#   ใช้แทนเซลล์แบบฝึกหัดที่ 1 ในกรณีที่โค้ดของผู้เรียนรันไม่สำเร็จ โดยทำงานตามลำดับดังนี้
#   1) แทนค่าศูนย์ที่เป็นไปไม่ได้ทางสรีรวิทยา (เช่น BMI = 0) ด้วย "ค่าที่ขาดหาย" (NaN)
#   2) เติมค่าที่ขาดหาย (imputation) ด้วยค่ามัธยฐาน (median)
#   3) แบ่งข้อมูลเป็นชุดฝึก (training set) และชุดทดสอบ (test set)
#   4) ฝึกโมเดลป่าสุ่ม (random forest) แล้ววัดผลด้วย accuracy, confusion matrix และ classification report
#   - ข้อมูลนำเข้า : ตาราง df จากเซลล์โหลดข้อมูล
#   - สิ่งที่ผู้เรียนจะเห็น : ค่า accuracy, ตาราง confusion matrix และรายงานผลแยกตามคลาส 0/1
#   โค้ดนี้ "รันได้แต่ไม่สมบูรณ์โดยเจตนา" — ผู้เรียนจะใช้รายการตรวจสอบในขั้นที่ 2
#   เพื่อค้นหาจุดที่ผิดพลาดหรือขาดหายไป (ตัวเลขจึงอาจเปลี่ยนเล็กน้อยทุกครั้งที่รัน)
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: นำเข้าเครื่องมือที่ต้องใช้
# ----------------------------------------------------------------------------
# นำเข้าไลบรารี numpy โดยตั้งชื่อย่อว่า np (ใช้ค่า np.nan ซึ่งแทน "ค่าที่ขาดหาย / ไม่มีข้อมูล")
import numpy as np
# นำเข้าฟังก์ชัน train_test_split จากไลบรารี scikit-learn (ชื่อในโค้ดคือ sklearn) โมดูล model_selection
#   ฟังก์ชันนี้ใช้สุ่มแบ่งข้อมูลออกเป็นชุดฝึกและชุดทดสอบ
#   - from ... import ... : นำเข้าเฉพาะบางส่วน (ฟังก์ชัน/คลาส) จากไลบรารี
from sklearn.model_selection import train_test_split
# นำเข้าคลาส RandomForestClassifier (โมเดลป่าสุ่มสำหรับจำแนกประเภท) จากโมดูล sklearn.ensemble
#   ป่าสุ่ม (random forest) คือการสร้าง "ต้นไม้ตัดสินใจ (decision tree)" หลายต้นแล้วให้ลงคะแนนร่วมกัน
from sklearn.ensemble import RandomForestClassifier
# นำเข้าฟังก์ชันวัดผล 3 ตัวจากโมดูล sklearn.metrics
#   - accuracy_score        : ความแม่นยำโดยรวม (accuracy) = สัดส่วนผู้ป่วยที่ทำนายถูกต้องทั้งหมด
#   - classification_report : รายงาน precision, recall (= ความไว/sensitivity สำหรับคลาส 1), f1-score แยกตามคลาส
#   - confusion_matrix      : ตารางนับจำนวนผลการทำนายถูก/ผิด แยกตามคลาสจริงและคลาสที่ทำนาย
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# ----------------------------------------------------------------------------
# ส่วนที่ 2: เตรียมข้อมูล (แทนค่าศูนย์ที่เป็นไปไม่ได้ และเติมค่าที่ขาดหาย)
# ----------------------------------------------------------------------------
# สร้างสำเนาของตาราง df ด้วยเมธอด .copy() ของ pandas แล้วเก็บไว้ในตัวแปร data
#   เพื่อให้การแก้ไขต่อจากนี้ไม่กระทบตาราง df ต้นฉบับ (ตามที่คำสั่งระบุว่าห้ามเขียนทับ df)
data = df.copy()

# Replace impossible zeros with NaN
# (แทนค่าศูนย์ที่เป็นไปไม่ได้ด้วย NaN) — สร้างรายการ (list) ชื่อคอลัมน์ที่ค่า 0 เป็นไปไม่ได้ทางคลินิก
#   - list คือชุดของค่าหลายค่าเรียงกัน เขียนไว้ใน [ ] และคั่นด้วยจุลภาค
#   - คอลัมน์: ระดับน้ำตาล (Glucose), ความดันโลหิต (BloodPressure), ความหนาผิวหนัง (SkinThickness),
#     อินซูลิน (Insulin) และดัชนีมวลกาย (BMI) — ค่าเหล่านี้เป็น 0 ไม่ได้ในผู้ที่มีชีวิต จึงหมายถึง "ไม่ได้ตรวจวัด"
cols_with_zeros = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
# เลือกเฉพาะคอลัมน์ในรายการ data[cols_with_zeros] แล้วแทนค่า 0 ด้วย np.nan จากนั้นเขียนกลับลงคอลัมน์เดิม
#   - data[รายการชื่อคอลัมน์] : การเลือก (indexing) หลายคอลัมน์จากตารางพร้อมกัน
#   - .replace(0, np.nan) : เมธอดของ pandas ที่แทนค่า
#       - 0      : ค่าเดิมที่ต้องการค้นหา
#       - np.nan : ค่าใหม่ที่จะใส่แทน (NaN = Not a Number คือ "ค่าที่ขาดหาย")
data[cols_with_zeros] = data[cols_with_zeros].replace(0, np.nan)

# Impute missing values with the median
# (เติมค่าที่ขาดหายด้วยค่ามัธยฐาน) — ทำงานเป็นลำดับดังนี้
#   1) data[cols_with_zeros].median() : คำนวณค่ามัธยฐาน (median) ของแต่ละคอลัมน์ (ข้ามค่า NaN)
#   2) .fillna(...) : เมธอดของ pandas ที่เติมช่องที่เป็น NaN ด้วยค่าที่ให้ (ในที่นี้คือ median ของคอลัมน์นั้น ๆ)
#   3) นำผลที่ได้เขียนกลับลงคอลัมน์เดิมของ data
data[cols_with_zeros] = data[cols_with_zeros].fillna(data[cols_with_zeros].median())

# ----------------------------------------------------------------------------
# ส่วนที่ 3: แยกตัวแปรนำเข้า/คำตอบ และแบ่งชุดฝึก/ชุดทดสอบ
# ----------------------------------------------------------------------------
# Split into features and target
# (แยกตัวแปรนำเข้า (features) และคำตอบ (target)) — สร้าง X_ai = ตารางที่ตัดคอลัมน์ Outcome ออก
#   - .drop(columns=["Outcome"]) : เมธอดของ pandas สำหรับลบคอลัมน์ (คืนตารางใหม่ ไม่แก้ตารางเดิม)
#       - columns=["Outcome"] : รายการชื่อคอลัมน์ที่ต้องการตัดออก
#   X_ai จึงเหลือค่าที่ตรวจวัด 8 คอลัมน์ที่โมเดลใช้ทำนาย
X_ai = data.drop(columns=["Outcome"])
# เลือกคอลัมน์ Outcome (1 = เป็นโรคเบาหวาน, 0 = ไม่เป็น) เก็บไว้ใน y_ai เป็น "คำตอบจริง" (label) ที่โมเดลต้องเรียนรู้
#   - data["Outcome"] : การเลือกคอลัมน์เดียวด้วยชื่อคอลัมน์ในวงเล็บเหลี่ยม
y_ai = data["Outcome"]
# สุ่มแบ่งข้อมูลด้วย train_test_split(...) แล้วรับผลลัพธ์ 4 ชิ้นพร้อมกันทางซ้ายของเครื่องหมาย =
#   (X_train, X_test = ตัวแปรนำเข้าของชุดฝึก/ชุดทดสอบ; y_train, y_test = คำตอบของชุดฝึก/ชุดทดสอบ)
#   - X_ai          : ตารางตัวแปรนำเข้าที่จะถูกแบ่ง
#   - y_ai          : คำตอบที่จะถูกแบ่งให้ตรงกับแถวของ X_ai
#   - test_size=0.2 : กันข้อมูลร้อยละ 20 ไว้เป็นชุดทดสอบ (อีกร้อยละ 80 เป็นชุดฝึก)
X_train, X_test, y_train, y_test = train_test_split(X_ai, y_ai, test_size=0.2)

# ----------------------------------------------------------------------------
# ส่วนที่ 4: ฝึกโมเดลป่าสุ่ม
# ----------------------------------------------------------------------------
# Train a random forest
# (ฝึกโมเดลป่าสุ่ม) — สร้างโมเดล RandomForestClassifier (ยังไม่ได้เรียนรู้) เก็บไว้ในตัวแปร model
#   - n_estimators=100 : จำนวนต้นไม้ตัดสินใจในป่า = 100 ต้น
model = RandomForestClassifier(n_estimators=100)
# ฝึก (fit) โมเดลด้วยข้อมูลชุดฝึก — โมเดลจะเรียนรู้รูปแบบที่เชื่อมค่าที่ตรวจวัดกับการเป็นโรคเบาหวาน
#   - X_train : ตัวแปรนำเข้าของชุดฝึก
#   - y_train : คำตอบจริงของชุดฝึก
model.fit(X_train, y_train)

# ----------------------------------------------------------------------------
# ส่วนที่ 5: ประเมินผลบนชุดทดสอบ
# ----------------------------------------------------------------------------
# Evaluate
# (ประเมินผล) — ใช้โมเดลทำนายผู้ป่วยในชุดทดสอบด้วยเมธอด .predict(...) ได้ผลเป็น 0/1 รายบุคคล เก็บไว้ใน y_pred
#   - X_test : ตัวแปรนำเข้าของชุดทดสอบ (โมเดลไม่เคยเห็นระหว่างฝึก)
#   - .predict ใช้เกณฑ์ตัดสิน (threshold) มาตรฐานที่ความน่าจะเป็น 0.5
y_pred = model.predict(X_test)
# แสดงค่า accuracy ด้วย print(...) — print รับได้หลายค่าคั่นด้วยจุลภาค และจะพิมพ์ต่อกันโดยเว้นวรรค
#   - "Accuracy:" : ข้อความนำหน้า
#   - accuracy_score(y_test, y_pred) : คำนวณสัดส่วนที่ทำนายถูก
#       - y_test : คำตอบจริง
#       - y_pred : คำตอบที่โมเดลทำนาย
print("Accuracy:", accuracy_score(y_test, y_pred))
# แสดง confusion matrix
#   - "\n" ในข้อความ คือ "ขึ้นบรรทัดใหม่"
#   - confusion_matrix(y_test, y_pred) : ตาราง 2x2 แถว = คลาสจริง, คอลัมน์ = คลาสที่ทำนาย
#     เรียงเป็น [[TN, FP], [FN, TP]] — TN = ผลลบจริง, FP = ผลบวกลวง, FN = ผลลบลวง, TP = ผลบวกจริง
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred))
# แสดงรายงานผลการจำแนกด้วย classification_report(y_test, y_pred)
#   - ให้ค่า precision, recall, f1-score และ support (จำนวนผู้ป่วยจริง) แยกตามคลาส 0 และ 1
#   - recall ในแถวของคลาส 1 คือความไว (sensitivity); recall ในแถวของคลาส 0 คือความจำเพาะ (specificity)
print("\nClassification report:\n", classification_report(y_test, y_pred))
````

💡 **คำอธิบายผลลัพธ์** ผู้ช่วยปัญญาประดิษฐ์สร้างไปป์ไลน์ครบทุกขั้นตอนภายในเวลาไม่กี่วินาที ได้แก่ แก้ไขค่าศูนย์ที่เป็นไปไม่ได้ → เติมค่าที่ขาดหาย (imputation) → แบ่งผู้ป่วยเป็นชุดฝึก (training set) สำหรับให้โมเดลเรียนรู้ และชุดทดสอบ (test set) ที่แยกไว้สำหรับประเมินโมเดล → ฝึกโมเดล random forest (ต้นไม้ตัดสินใจหลายต้นที่ลงมติร่วมกัน) → แสดงคะแนน ขั้นตอนถัดไปคือการตรวจสอบว่าผลลัพธ์นี้ **ถูกต้อง** หรือไม่

## ขั้นที่ 2 — ตรวจสอบ (Audit) ด้วยรายการคำถามแบบ ใช่/ไม่ใช่

โค้ดการเรียนรู้ของเครื่องที่ปัญญาประดิษฐ์สร้างขึ้นมักรันได้โดยไม่เกิดข้อผิดพลาด แต่มีความผิดพลาดแฝงที่ทำให้ผลดูดีกว่าความเป็นจริง ผู้เรียน **ไม่จำเป็น** ต้องแก้ไขโค้ด เพียงพิจารณาผลลัพธ์ของเซลล์แบบฝึกหัดที่ 1 (และค้นหาคำบางคำในโค้ด) แล้วตอบว่า ใช่/ไม่ใช่

ขั้นแรก ให้รันเซลล์ช่วยตรวจสอบด้านล่าง ซึ่งจะแสดงตัวเลขสองค่าที่ใช้ในรายการตรวจสอบ

````python
# เซลล์ช่วยตรวจสอบ — แสดงข้อเท็จจริงที่ใช้ในรายการตรวจสอบ (รันเซลล์นี้เท่านั้น)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้คำนวณ "สัดส่วนของผู้ป่วยแต่ละกลุ่ม" (ไม่เป็น/เป็นโรคเบาหวาน) ในข้อมูลทั้งหมด
#   เพื่อให้ผู้เรียนเห็นว่า "โมเดล" ที่ไม่ฉลาดเลย (ทายว่าไม่เป็นโรคทุกราย) ก็ได้ accuracy ประมาณ 0.65
#   แต่มีความไว (sensitivity) เท่ากับ 0 คือพลาดผู้ป่วยทุกราย
#   - ข้อมูลนำเข้า : คอลัมน์ Outcome ของตาราง df
#   - สิ่งที่ผู้เรียนจะเห็น : ร้อยละของผู้ป่วยแต่ละกลุ่ม และคำแนะนำว่าจะหาบรรทัดใดในโค้ดแบบฝึกหัดที่ 1
#     เพื่อตอบข้อ E (การรั่วไหลของข้อมูล / data leakage) ในรายการตรวจสอบ
#   ผู้เรียนไม่ต้องแก้ไขโค้ด เพียงรันเซลล์นี้
# ============================================================================

# นับสัดส่วนของแต่ละค่าในคอลัมน์ Outcome แล้วเก็บไว้ในตัวแปร share (ทำงานเป็นลำดับ)
#   1) df["Outcome"] : เลือกคอลัมน์ Outcome (0 = ไม่เป็นโรค, 1 = เป็นโรคเบาหวาน)
#   2) .value_counts(normalize=True) : เมธอดของ pandas ที่นับจำนวนครั้งของแต่ละค่า
#       - normalize=True : แปลงจำนวนนับเป็น "สัดส่วน" (ผลรวมเท่ากับ 1) แทนจำนวนราย
#   ผลลัพธ์ share ใช้ค่า 0 หรือ 1 เป็นป้ายชื่อ เช่น share[0] = สัดส่วนผู้ที่ไม่เป็นโรค
share = df["Outcome"].value_counts(normalize=True)
# แสดงร้อยละของผู้ที่ไม่เป็นและเป็นโรคเบาหวาน
#   - f"..." (f-string) : ข้อความที่แทรกค่าได้ โดยเขียนค่าไว้ใน { }
#   - share[0], share[1] : ดึงสัดส่วนของกลุ่ม 0 และกลุ่ม 1 ออกมาจาก share
#   - :.1% : รหัสจัดรูปแบบ แสดงเป็นร้อยละ ทศนิยม 1 ตำแหน่ง (เช่น 0.651 → 65.1%)
print(f"ผู้ป่วยที่ไม่เป็นโรคเบาหวาน: {share[0]:.1%}   |   ผู้ป่วยที่เป็นโรคเบาหวาน: {share[1]:.1%}")
# แสดงค่า accuracy ของ "โมเดล" ที่ทายว่าไม่เป็นโรคทุกราย ซึ่งเท่ากับสัดส่วนผู้ที่ไม่เป็นโรค share[0]
#   - :.2f : รหัสจัดรูปแบบ แสดงเป็นตัวเลขทศนิยม 2 ตำแหน่ง (เช่น 0.65)
print(f"=> 'โมเดล' ที่ทำนายว่า 'ไม่เป็นโรคเบาหวาน' สำหรับทุกราย มีค่า accuracy {share[0]:.2f}")
# แสดงข้อความเตือนว่าโมเดลดังกล่าวตรวจไม่พบผู้ป่วยเลย (ความไว / sensitivity = 0)
#   (ขึ้นต้นด้วย f แต่ไม่มี { } จึงพิมพ์ข้อความตามที่เขียนไว้ทุกตัวอักษร)
print(f"   แต่ไม่สามารถตรวจพบผู้ป่วยโรคเบาหวานแม้แต่รายเดียว (sensitivity = 0)")
# print() ที่ไม่มีค่าในวงเล็บ จะพิมพ์บรรทัดว่าง 1 บรรทัด เพื่อเว้นระยะให้อ่านง่าย
print()
# แสดงหัวข้อคำแนะนำสำหรับข้อ E ในรายการตรวจสอบ
print("ตำแหน่งของการเติมค่าที่ขาดหายและการแบ่งข้อมูล (สำหรับข้อ E)")
# แนะนำให้หาบรรทัดที่แบ่งข้อมูล (train_test_split) ในโค้ดแบบฝึกหัดที่ 1
print("  ในโค้ดแบบฝึกหัดที่ 1 ให้หาบรรทัดที่มี  train_test_split  และบรรทัดที่มี")
# และบรรทัดที่เติมค่า/ปรับสเกลข้อมูล (fillna, SimpleImputer, fit_transform, StandardScaler)
#   เพื่อดูว่าบรรทัดใดมาก่อน — หากเติมค่าก่อนแบ่งข้อมูล ถือเป็นสัญญาณเตือนการรั่วไหลของข้อมูล
print("  fillna / SimpleImputer / fit_transform / StandardScaler")
````

💡 **คำอธิบายผลลัพธ์** ผู้ป่วยประมาณร้อยละ 65 ไม่เป็นโรคเบาหวาน ดังนั้น **การพิจารณาเฉพาะค่า accuracy อาจทำให้เข้าใจผิด** โมเดลอาจมีค่า accuracy ร้อยละ 65 โดยไม่สามารถตรวจพบผู้ป่วยโรคเบาหวานได้แม้แต่รายเดียว สำหรับการตรวจคัดกรอง (screening) ตัวชี้วัดที่สำคัญคือ **ความไว (sensitivity)** หมายถึงสัดส่วนของผู้ป่วยโรคเบาหวานที่โมเดลตรวจพบ และ **ความจำเพาะ (specificity)** หมายถึงสัดส่วนของผู้ที่ไม่เป็นโรคซึ่งโมเดลจำแนกได้ถูกต้อง ส่วน **ROC AUC** แสดงความสามารถของโมเดลในการ *จัดลำดับ* ความเสี่ยงของผู้ป่วย (0.5 = เทียบเท่าการสุ่ม, 1.0 = สมบูรณ์)

## ขั้นที่ 3 — แก้ไข: เปรียบเทียบกับไปป์ไลน์อ้างอิง

ด้านล่างเป็นการเขียนไปป์ไลน์เดียวกันโดย **ไม่มีการรั่วไหลของข้อมูล** หลักสำคัญคือ `Pipeline` ของ scikit-learn ซึ่งรวมขั้นตอน "เติมค่าที่ขาดหาย → ฝึกโมเดล" ไว้ในออบเจกต์เดียว ค่ามัธยฐาน (median) ที่ใช้เติมจึงคำนวณจาก **ผู้ป่วยในชุดฝึกเท่านั้น** ทุกครั้ง รวมถึงภายในการตรวจสอบไขว้

*การตรวจสอบไขว้ (cross-validation) แบบ 5-fold* คือการแบ่งข้อมูลเป็น 5 ส่วน ใช้ 4 ส่วนในการฝึกและอีก 1 ส่วนในการทดสอบ ทำซ้ำ 5 รอบแล้วหาค่าเฉลี่ย ทำให้ได้คะแนนที่มีเสถียรภาพมากกว่าการแบ่งข้อมูลครั้งเดียว

````python
# ⭐ Exercise 2 — แบบฝึกหัดที่ 2: ไปป์ไลน์อ้างอิงที่ไม่มี data leakage — เติมช่องว่าง 5 ช่องที่เป็น ...
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เซลล์นี้สร้าง "ไปป์ไลน์อ้างอิง (reference pipeline)" ที่แก้ปัญหาของโค้ดจาก AI ในแบบฝึกหัดที่ 1
#   - ใช้ Pipeline ของ scikit-learn รวม "เติมค่าที่ขาดหาย → ฝึกโมเดล" ไว้ด้วยกัน
#     ค่าที่ใช้เติมจึงคำนวณจากชุดฝึกเท่านั้น ไม่มีการรั่วไหลของข้อมูล (data leakage)
#   - กำหนด random_state เพื่อให้ผลเหมือนเดิมทุกครั้งที่รัน (ทำซ้ำได้ / reproducibility)
#   - วัดผลด้วยตัวชี้วัดที่เหมาะกับการตรวจคัดกรอง ได้แก่ ROC AUC จากการตรวจสอบไขว้ (cross-validation)
#     แบบ 5-fold และความไว (sensitivity) / ความจำเพาะ (specificity) บนชุดทดสอบ
#   - ข้อมูลนำเข้า : ตาราง df
#   - สิ่งที่ผู้เรียนจะเห็น : ค่า ROC AUC เฉลี่ย ± ส่วนเบี่ยงเบนมาตรฐาน, จำนวน TP/FN/TN/FP
#     และค่า sensitivity / specificity ของชุดทดสอบ
#   ผู้เรียนต้องเติมช่องว่าง 5 ช่องที่เขียนว่า ... ตามคำใบ้ท้ายบรรทัด (หากทำไม่ได้ ให้รันเซลล์ ✅ เฉลย)
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: นำเข้าเครื่องมือจาก scikit-learn (sklearn)
# ----------------------------------------------------------------------------
# นำเข้าคลาส Pipeline : ใช้ร้อยขั้นตอนประมวลผลหลายขั้นต่อกันเป็นสายเดียว แล้ว fit/predict ได้ในคำสั่งเดียว
from sklearn.pipeline import Pipeline
# นำเข้าคลาส SimpleImputer : เครื่องมือเติมค่าที่ขาดหาย (NaN) ด้วยค่าสถิติ เช่น ค่ามัธยฐาน (median)
from sklearn.impute import SimpleImputer
# นำเข้าคลาส RandomForestClassifier : โมเดลป่าสุ่ม (random forest) สำหรับจำแนกประเภท
from sklearn.ensemble import RandomForestClassifier
# นำเข้าฟังก์ชัน 2 ตัวจากโมดูล model_selection
#   - cross_val_score  : วัดผลโมเดลด้วยการตรวจสอบไขว้ (cross-validation) แล้วคืนคะแนนของแต่ละรอบ
#   - train_test_split : สุ่มแบ่งข้อมูลเป็นชุดฝึกและชุดทดสอบ
from sklearn.model_selection import cross_val_score, train_test_split
# นำเข้าฟังก์ชัน confusion_matrix : สร้างตารางนับผลการทำนายถูก/ผิด (TN, FP, FN, TP)
from sklearn.metrics import confusion_matrix

# ----------------------------------------------------------------------------
# ส่วนที่ 2: เตรียมตัวแปรนำเข้า (X) และคำตอบ (y)
# ----------------------------------------------------------------------------
# สร้างรายการ (list) ชื่อคอลัมน์ที่ "ค่า 0 หมายถึงไม่ได้ตรวจวัด" (เป็นไปไม่ได้ทางสรีรวิทยา)
#   ชื่อตัวแปรเป็นตัวพิมพ์ใหญ่ทั้งหมดตามธรรมเนียม เพื่อบอกว่าเป็นค่าคงที่ที่ไม่ตั้งใจเปลี่ยน
ZERO_MEANS_MISSING = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
# สร้างตารางตัวแปรนำเข้า X (ทำงานเป็นลำดับ)
#   1) df.drop(columns=["Outcome"]) : ตัดคอลัมน์คำตอบ Outcome ออก เหลือค่าที่ตรวจวัด 8 คอลัมน์
#   2) .replace({...}) : แทนค่าในตาราง โดยรับ dict ซ้อน dict แบบ {ชื่อคอลัมน์: {ค่าเดิม: ค่าใหม่}}
#   3) {c: {0: np.nan} for c in ZERO_MEANS_MISSING} : "dict comprehension" สร้าง dict โดยวนทีละชื่อคอลัมน์ c
#      ในรายการ ZERO_MEANS_MISSING แล้วกำหนดว่า "ในคอลัมน์ c ให้เปลี่ยน 0 เป็น np.nan (ค่าที่ขาดหาย)"
#      - dict คือชุดข้อมูลแบบ "กุญแจ: ค่า" เขียนใน { }
#   ผลคือ ค่า 0 ถูกแทนด้วย NaN เฉพาะ 5 คอลัมน์นี้ (คอลัมน์อื่น เช่น Pregnancies = 0 ยังคงเป็น 0 ตามจริง)
#   หมายเหตุ: ยังไม่เติมค่า NaN ที่นี่ — การเติมค่าจะทำภายใน Pipeline เพื่อป้องกันการรั่วไหลของข้อมูล
X = df.drop(columns=["Outcome"]).replace({c: {0: np.nan} for c in ZERO_MEANS_MISSING})
# เลือกคอลัมน์ Outcome เป็นคำตอบจริง y (1 = เป็นโรคเบาหวาน, 0 = ไม่เป็น)
y = df["Outcome"]

# ----------------------------------------------------------------------------
# ส่วนที่ 3: สร้างไปป์ไลน์ (เติมค่าที่ขาดหาย → โมเดลป่าสุ่ม)
# ----------------------------------------------------------------------------
# สร้าง Pipeline เก็บไว้ในตัวแปร pipe โดยส่ง list ของขั้นตอนเข้าไป
#   - แต่ละขั้นตอนเขียนเป็นคู่ (tuple) ในวงเล็บ ( ) แบบ ("ชื่อขั้นตอน", เครื่องมือ) เรียงตามลำดับการทำงาน
#   - เมื่อสั่ง pipe.fit(...) ทุกขั้นตอนจะเรียนรู้จากข้อมูลที่ส่งเข้ามาเท่านั้น (ในการตรวจสอบไขว้คือข้อมูลฝึกของแต่ละรอบ)
pipe = Pipeline([
    # ขั้นตอนที่ 1 ชื่อ "imputer": SimpleImputer เติมค่าที่ขาดหาย
    #   - strategy=... : วิธีคำนวณค่าที่ใช้เติม — ... คือช่องว่างที่ 1 ให้ผู้เรียนเติมตามคำใบ้ท้ายบรรทัด
    ("imputer", SimpleImputer(strategy=...)),                   # ช่องที่ 1: พิมพ์ "median" (รวมเครื่องหมายอัญประกาศ)
    # ขั้นตอนที่ 2 ชื่อ "model": RandomForestClassifier โมเดลป่าสุ่ม
    #   - n_estimators=300 : ใช้ต้นไม้ตัดสินใจ 300 ต้น
    #   - random_state=... : เลขตั้งต้นของการสุ่ม (ทำให้ผลเหมือนเดิมทุกครั้ง) — ... คือช่องว่างที่ 2 ตามคำใบ้
    ("model", RandomForestClassifier(n_estimators=300, random_state=...)),   # ช่องที่ 2: พิมพ์ 42
])

# ----------------------------------------------------------------------------
# ส่วนที่ 4: วัดผลด้วยการตรวจสอบไขว้แบบ 5-fold
# ----------------------------------------------------------------------------
# ROC AUC จากการตรวจสอบไขว้แบบ 5-fold (imputer ถูก fit ใหม่ในทุก fold จึงไม่มี data leakage)
# เรียก cross_val_score(...) แบ่งข้อมูลเป็น 5 ส่วน ฝึกด้วย 4 ส่วน ทดสอบกับอีก 1 ส่วน วนครบ 5 รอบ
#   แล้วเก็บคะแนน 5 ค่า (หนึ่งค่าต่อรอบ) ไว้ในตัวแปร auc
#   - pipe       : ไปป์ไลน์ที่ต้องการประเมิน
#   - X, y       : ตัวแปรนำเข้าและคำตอบของผู้ป่วยทั้งหมด
#   - cv=5       : จำนวนส่วน (fold) ของการตรวจสอบไขว้ = 5
#   - scoring=...: ชื่อตัวชี้วัดที่ใช้ให้คะแนน — ... คือช่องว่างที่ 3 ตามคำใบ้ท้ายบรรทัด
#     (ROC AUC คือพื้นที่ใต้กราฟ ROC บอกความสามารถแยกผู้ป่วยกับผู้ไม่ป่วย ไม่ขึ้นกับเกณฑ์ตัดสิน ค่า 0.5 = เดาสุ่ม, 1.0 = สมบูรณ์)
auc = cross_val_score(pipe, X, y, cv=5, scoring=...)          # ช่องที่ 3: พิมพ์ "roc_auc" (รวมเครื่องหมายอัญประกาศ)
# แสดงค่า ROC AUC เฉลี่ย ± ส่วนเบี่ยงเบนมาตรฐานของ 5 รอบ
#   - auc.mean() : ค่าเฉลี่ย (เมธอดของอาร์เรย์ numpy)
#   - auc.std()  : ส่วนเบี่ยงเบนมาตรฐาน (standard deviation) บอกความผันผวนระหว่างรอบ
#   - :.3f       : แสดงทศนิยม 3 ตำแหน่ง
print(f"Cross-validated ROC AUC: {auc.mean():.3f} ± {auc.std():.3f}")

# ----------------------------------------------------------------------------
# ส่วนที่ 5: คำนวณ sensitivity และ specificity บนชุดทดสอบ
# ----------------------------------------------------------------------------
# แบ่ง train/test แบบ stratified หนึ่งครั้ง เพื่อคำนวณ sensitivity และ specificity ที่เกณฑ์ตัดสินมาตรฐาน (0.5)
# เรียก train_test_split(...) แล้วรับผลลัพธ์ 4 ชิ้น: X_tr, X_te (ตัวแปรนำเข้าชุดฝึก/ทดสอบ), y_tr, y_te (คำตอบชุดฝึก/ทดสอบ)
#   - X, y            : ข้อมูลที่จะถูกแบ่ง
#   - test_size=0.2   : กันข้อมูลร้อยละ 20 ไว้เป็นชุดทดสอบ
#   - stratify=y      : แบ่งให้สัดส่วนผู้เป็น/ไม่เป็นโรคในชุดฝึกและชุดทดสอบเท่ากับในข้อมูลทั้งหมด
#   - random_state=42 : เลขตั้งต้นของการสุ่ม ทำให้แบ่งได้เหมือนเดิมทุกครั้ง
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
# ฝึกไปป์ไลน์ด้วยชุดฝึกด้วยเมธอด .fit(X_tr, y_tr) — ขั้น imputer คำนวณค่ามัธยฐานจากชุดฝึก แล้วโมเดลเรียนรู้ต่อ
#   - X_tr : ตัวแปรนำเข้าของชุดฝึก
#   - y_tr : คำตอบของชุดฝึก
pipe.fit(X_tr, y_tr)                                          # ค่า median คำนวณจากผู้ป่วยในชุดฝึกเท่านั้น
# ทำนายชุดทดสอบแล้วนับผลถูก/ผิด (ทำงานเป็นลำดับ)
#   1) pipe.predict(X_te) : ทำนาย 0/1 ของผู้ป่วยชุดทดสอบ (เกณฑ์ตัดสินมาตรฐาน 0.5)
#   2) confusion_matrix(y_te, ...) : สร้างตาราง 2x2 [[TN, FP], [FN, TP]]
#       - y_te : คำตอบจริง, อาร์กิวเมนต์ที่สอง : คำตอบที่ทำนาย
#   3) .ravel() : "คลี่" ตาราง 2x2 ให้เป็นแถวเดียว 4 ค่า เรียงเป็น TN, FP, FN, TP
#   4) tn, fp, fn, tp = ... : รับ 4 ค่าเก็บลงตัวแปร 4 ตัวตามลำดับ
#      (TN = ผลลบจริง, FP = ผลบวกลวง, FN = ผลลบลวง/ผู้ป่วยที่พลาด, TP = ผลบวกจริง/ผู้ป่วยที่ตรวจพบ)
tn, fp, fn, tp = confusion_matrix(y_te, pipe.predict(X_te)).ravel()
# คำนวณความไว (sensitivity) — ... คือช่องว่างที่ 4 ให้ผู้เรียนพิมพ์สูตรตามคำใบ้ท้ายบรรทัด
sensitivity = ...            # ช่องที่ 4: ผู้ป่วยโรคเบาหวานที่ตรวจพบ / ผู้ป่วยโรคเบาหวานทั้งหมด -> พิมพ์  tp / (tp + fn)
# คำนวณความจำเพาะ (specificity) — ... คือช่องว่างที่ 5 ให้ผู้เรียนพิมพ์สูตรตามคำใบ้ท้ายบรรทัด
specificity = ...            # ช่องที่ 5: ผู้ไม่เป็นโรคที่จำแนกถูกต้อง / ผู้ไม่เป็นโรคทั้งหมด -> พิมพ์  tn / (tn + fp)
# แสดงจำนวน TP, FN, TN, FP ของชุดทดสอบ โดยแทรกค่าตัวแปรผ่าน f-string { }
print(f"ชุดทดสอบ: TP={tp} FN={fn} TN={tn} FP={fp}")
# แสดงค่า sensitivity และ specificity ทศนิยม 3 ตำแหน่ง (:.3f)
print(f"Sensitivity: {sensitivity:.3f}, Specificity: {specificity:.3f}")
````

````python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ไปป์ไลน์อ้างอิง (ไม่มี data leakage, ทำซ้ำได้, มีตัวชี้วัดสำหรับการตรวจคัดกรอง)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เฉลยของแบบฝึกหัดที่ 2 — ไปป์ไลน์อ้างอิงที่เติมช่องว่างครบทั้ง 5 ช่อง ได้แก่
#   strategy="median", random_state=42, scoring="roc_auc",
#   sensitivity = tp / (tp + fn) และ specificity = tn / (tn + fp)
#   - แก้ปัญหาการรั่วไหลของข้อมูล (data leakage): การเติมค่าที่ขาดหายอยู่ "ภายใน" Pipeline
#     จึงคำนวณค่ามัธยฐานจากข้อมูลฝึกเท่านั้น
#   - ทำซ้ำได้ (reproducible): กำหนด random_state=42 ทุกจุดที่มีการสุ่ม
#   - ใช้ตัวชี้วัดที่เหมาะกับการตรวจคัดกรอง: ROC AUC (ตรวจสอบไขว้ 5-fold), sensitivity, specificity
#   - ข้อมูลนำเข้า : ตาราง df
#   - สิ่งที่ผู้เรียนจะเห็น : ROC AUC เฉลี่ย ± ส่วนเบี่ยงเบนมาตรฐาน, TP/FN/TN/FP และ sensitivity/specificity ของชุดทดสอบ
#   ตัวแปร pipe, X, y, X_tr, X_te, y_tr, y_te ที่สร้างในเซลล์นี้จะถูกใช้ต่อในขั้นที่ 4 และโจทย์ท้าทาย
# ============================================================================

# ----------------------------------------------------------------------------
# ส่วนที่ 1: นำเข้าเครื่องมือจาก scikit-learn (sklearn)
# ----------------------------------------------------------------------------
# นำเข้าคลาส Pipeline : ร้อยขั้นตอนประมวลผลหลายขั้นต่อกันเป็นสายเดียว แล้ว fit/predict ได้ในคำสั่งเดียว
from sklearn.pipeline import Pipeline
# นำเข้าคลาส SimpleImputer : เครื่องมือเติมค่าที่ขาดหาย (NaN) ด้วยค่าสถิติ เช่น ค่ามัธยฐาน (median)
from sklearn.impute import SimpleImputer
# นำเข้าคลาส RandomForestClassifier : โมเดลป่าสุ่ม (random forest) สำหรับจำแนกประเภท
from sklearn.ensemble import RandomForestClassifier
# นำเข้าฟังก์ชัน 2 ตัวจากโมดูล model_selection
#   - cross_val_score  : วัดผลด้วยการตรวจสอบไขว้ (cross-validation) แล้วคืนคะแนนของแต่ละรอบ
#   - train_test_split : สุ่มแบ่งข้อมูลเป็นชุดฝึกและชุดทดสอบ
from sklearn.model_selection import cross_val_score, train_test_split
# นำเข้าฟังก์ชัน confusion_matrix : สร้างตารางนับผลการทำนายถูก/ผิด (TN, FP, FN, TP)
from sklearn.metrics import confusion_matrix

# ----------------------------------------------------------------------------
# ส่วนที่ 2: เตรียมตัวแปรนำเข้า (X) และคำตอบ (y)
# ----------------------------------------------------------------------------
# รายการ (list) ชื่อคอลัมน์ที่ "ค่า 0 หมายถึงไม่ได้ตรวจวัด" (เป็นไปไม่ได้ทางสรีรวิทยา)
ZERO_MEANS_MISSING = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
# สร้างตารางตัวแปรนำเข้า X (ทำงานเป็นลำดับ)
#   1) df.drop(columns=["Outcome"]) : ตัดคอลัมน์คำตอบ Outcome ออก เหลือค่าที่ตรวจวัด 8 คอลัมน์
#   2) .replace({...}) : แทนค่าแบบ {ชื่อคอลัมน์: {ค่าเดิม: ค่าใหม่}}
#   3) {c: {0: np.nan} for c in ZERO_MEANS_MISSING} : dict comprehension — วนทีละชื่อคอลัมน์ c
#      แล้วกำหนดว่า "ในคอลัมน์ c ให้เปลี่ยน 0 เป็น np.nan (ค่าที่ขาดหาย)"
#   ยังไม่เติมค่า NaN ที่นี่ เพราะการเติมค่าจะทำภายใน Pipeline (ป้องกันการรั่วไหลของข้อมูล)
X = df.drop(columns=["Outcome"]).replace({c: {0: np.nan} for c in ZERO_MEANS_MISSING})
# เลือกคอลัมน์ Outcome เป็นคำตอบจริง y (1 = เป็นโรคเบาหวาน, 0 = ไม่เป็น)
y = df["Outcome"]

# ----------------------------------------------------------------------------
# ส่วนที่ 3: สร้างไปป์ไลน์ (เติมค่าที่ขาดหาย → โมเดลป่าสุ่ม)
# ----------------------------------------------------------------------------
# สร้าง Pipeline เก็บไว้ในตัวแปร pipe — รับ list ของขั้นตอน แต่ละขั้นเป็นคู่ (tuple) ("ชื่อขั้นตอน", เครื่องมือ)
pipe = Pipeline([
    # ขั้นตอนที่ 1 "imputer": SimpleImputer เติมค่าที่ขาดหาย
    #   - strategy="median" : เติมด้วยค่ามัธยฐานของคอลัมน์นั้น (ทนต่อค่าผิดปกติ/outlier ได้ดีกว่าค่าเฉลี่ย)
    ("imputer", SimpleImputer(strategy="median")),
    # ขั้นตอนที่ 2 "model": RandomForestClassifier
    #   - n_estimators=300 : ใช้ต้นไม้ตัดสินใจ 300 ต้น
    #   - random_state=42  : เลขตั้งต้นของการสุ่ม ทำให้ได้โมเดลเดิมทุกครั้งที่รัน (ตัวเลข 42 เป็นค่าที่นิยมใช้ ไม่มีความหมายพิเศษ)
    ("model", RandomForestClassifier(n_estimators=300, random_state=42)),
])

# ----------------------------------------------------------------------------
# ส่วนที่ 4: วัดผลด้วยการตรวจสอบไขว้แบบ 5-fold
# ----------------------------------------------------------------------------
# เรียก cross_val_score(...) แบ่งข้อมูลเป็น 5 ส่วน ฝึกด้วย 4 ส่วน ทดสอบกับอีก 1 ส่วน วนครบ 5 รอบ
#   (imputer ถูก fit ใหม่ในทุกรอบจากข้อมูลฝึกของรอบนั้น จึงไม่มีการรั่วไหลของข้อมูล) ได้คะแนน 5 ค่าเก็บใน auc
#   - pipe              : ไปป์ไลน์ที่ต้องการประเมิน
#   - X, y              : ตัวแปรนำเข้าและคำตอบของผู้ป่วยทั้งหมด
#   - cv=5              : จำนวนส่วน (fold) = 5
#   - scoring="roc_auc" : ใช้ ROC AUC (พื้นที่ใต้กราฟ ROC) เป็นคะแนน — 0.5 = เดาสุ่ม, 1.0 = แยกได้สมบูรณ์
auc = cross_val_score(pipe, X, y, cv=5, scoring="roc_auc")
# แสดง ROC AUC เฉลี่ย (auc.mean()) ± ส่วนเบี่ยงเบนมาตรฐาน (auc.std()) ทศนิยม 3 ตำแหน่ง (:.3f)
print(f"Cross-validated ROC AUC: {auc.mean():.3f} ± {auc.std():.3f}")

# ----------------------------------------------------------------------------
# ส่วนที่ 5: คำนวณ sensitivity และ specificity บนชุดทดสอบ (เกณฑ์ตัดสินมาตรฐาน 0.5)
# ----------------------------------------------------------------------------
# แบ่งข้อมูลหนึ่งครั้ง ได้ X_tr, X_te (ตัวแปรนำเข้าชุดฝึก/ทดสอบ) และ y_tr, y_te (คำตอบชุดฝึก/ทดสอบ)
#   - X, y            : ข้อมูลที่จะถูกแบ่ง
#   - test_size=0.2   : กันข้อมูลร้อยละ 20 ไว้เป็นชุดทดสอบ
#   - stratify=y      : รักษาสัดส่วนผู้เป็น/ไม่เป็นโรคในทั้งสองชุดให้เท่ากับข้อมูลทั้งหมด
#   - random_state=42 : แบ่งได้เหมือนเดิมทุกครั้ง
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
# ฝึกไปป์ไลน์ด้วยชุดฝึก — imputer คำนวณค่ามัธยฐานจากผู้ป่วยในชุดฝึกเท่านั้น แล้วโมเดลเรียนรู้ต่อ
#   - X_tr : ตัวแปรนำเข้าของชุดฝึก
#   - y_tr : คำตอบของชุดฝึก
pipe.fit(X_tr, y_tr)
# ทำนายชุดทดสอบแล้วนับผลถูก/ผิด (ทำงานเป็นลำดับ)
#   1) pipe.predict(X_te) : ทำนาย 0/1 ของผู้ป่วยชุดทดสอบ (เกณฑ์ตัดสิน 0.5)
#   2) confusion_matrix(y_te, ...) : ตาราง 2x2 [[TN, FP], [FN, TP]] (y_te = คำตอบจริง)
#   3) .ravel() : คลี่ตารางเป็น 4 ค่าเรียงกัน TN, FP, FN, TP
#   4) รับค่าทั้ง 4 ลงตัวแปร tn, fp, fn, tp ตามลำดับ
tn, fp, fn, tp = confusion_matrix(y_te, pipe.predict(X_te)).ravel()
# ความไว (sensitivity) = ผู้ป่วยโรคเบาหวานที่ตรวจพบ (tp) / ผู้ป่วยโรคเบาหวานทั้งหมด (tp + fn)
#   - วงเล็บ ( ) ทำให้บวกกันก่อนแล้วจึงหาร
sensitivity = tp / (tp + fn)
# ความจำเพาะ (specificity) = ผู้ไม่เป็นโรคที่จำแนกถูกต้อง (tn) / ผู้ไม่เป็นโรคทั้งหมด (tn + fp)
specificity = tn / (tn + fp)
# แสดงจำนวน TP, FN, TN, FP ของชุดทดสอบ โดยแทรกค่าตัวแปรผ่าน f-string { }
print(f"ชุดทดสอบ: TP={tp} FN={fn} TN={tn} FP={fp}")
# แสดงค่า sensitivity และ specificity ทศนิยม 3 ตำแหน่ง (:.3f)
print(f"Sensitivity: {sensitivity:.3f}, Specificity: {specificity:.3f}")
````

## ขั้นที่ 4 — ปรับปรุง: การเลือกเกณฑ์ตัดสินสำหรับการตรวจคัดกรอง

โมเดลให้ค่า *ความน่าจะเป็น (probability)* ของการเป็นโรคเบาหวานสำหรับผู้ป่วยแต่ละราย โดยค่าเริ่มต้น ผู้ที่มีค่าเกิน **0.5** จะถูกจำแนกว่ามีความเสี่ยง สำหรับการตรวจคัดกรอง โดยทั่วไปจะยอมรับผลบวกลวงที่เพิ่มขึ้นเพื่อลดการพลาดผู้ป่วย จึงใช้ **เกณฑ์ตัดสิน (threshold) ที่ต่ำลง**

````python
# ⭐ Exercise 3 — แบบฝึกหัดที่ 3: เลือกเกณฑ์ตัดสินสำหรับการตรวจคัดกรอง (แก้ไขตัวเลขเพียงค่าเดียว)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   โมเดลให้ "ความน่าจะเป็น (probability)" ของการเป็นโรคเบาหวานแก่ผู้ป่วยแต่ละราย
#   ผู้ที่มีค่าตั้งแต่ "เกณฑ์ตัดสิน (threshold)" ขึ้นไปจะถูกจัดว่ามีความเสี่ยง (ค่าเริ่มต้นคือ 0.5)
#   ในการตรวจคัดกรอง เรามักยอมลดเกณฑ์ลงเพื่อให้พลาดผู้ป่วยน้อยลง (ความไว / sensitivity สูงขึ้น)
#   แลกกับผลบวกลวงที่มากขึ้น (ความจำเพาะ / specificity ลดลง)
#   เซลล์นี้ทำ 2 ส่วน
#   1) สร้างตาราง sensitivity / specificity ที่เกณฑ์ 0.20–0.50 โดยใช้ "ชุดฝึกเท่านั้น"
#      (ความน่าจะเป็นแบบ out-of-fold จากการตรวจสอบไขว้) เพื่อไม่ให้แอบดูชุดทดสอบ
#   2) นำเกณฑ์ที่ผู้เรียนเลือก (THRESHOLD) ไปประเมินกับชุดทดสอบ "เพียงครั้งเดียว"
#   - ข้อมูลนำเข้า : pipe, X_tr, y_tr, X_te, y_te จากเซลล์แบบฝึกหัดที่ 2 / เฉลย
#   - สิ่งที่ผู้เรียนจะเห็น : ตารางเกณฑ์ตัดสิน และค่า sensitivity / specificity ของชุดทดสอบที่เกณฑ์ที่เลือก
#   ผู้เรียนต้องแก้ไขเพียงค่าเดียว คือ THRESHOLD (บรรทัดที่มี # TODO)
# ============================================================================

# นำเข้าฟังก์ชัน cross_val_predict จาก sklearn.model_selection
#   คล้าย cross_val_score แต่แทนที่จะคืน "คะแนน" จะคืน "ผลการทำนายของผู้ป่วยแต่ละราย"
#   โดยแต่ละรายถูกทำนายจากโมเดลที่ไม่เคยเห็นผู้ป่วยรายนั้นระหว่างฝึก (out-of-fold)
from sklearn.model_selection import cross_val_predict

# ----------------------------------------------------------------------------
# ส่วนที่ 1: สร้างตารางเกณฑ์ตัดสินจากชุดฝึก
# ----------------------------------------------------------------------------
# ความน่าจะเป็นแบบ out-of-fold ของผู้ป่วยในชุดฝึกเท่านั้น (ไม่ใช้ข้อมูลชุดทดสอบ)
# เรียก cross_val_predict(...) แล้วเลือกเฉพาะความน่าจะเป็นของคลาส 1 เก็บใน p_train_oof
#   - pipe                   : ไปป์ไลน์ (เติมค่า → ป่าสุ่ม) ที่จะถูกฝึกใหม่ในแต่ละรอบ
#   - X_tr, y_tr             : ตัวแปรนำเข้าและคำตอบของชุดฝึกเท่านั้น
#   - cv=5                   : ตรวจสอบไขว้ 5 ส่วน (fold)
#   - method="predict_proba" : ให้คืน "ความน่าจะเป็น" ของแต่ละคลาส แทนคำตอบ 0/1
#   - [:, 1] : ผลลัพธ์เป็นตาราง 2 คอลัมน์ (คลาส 0, คลาส 1) — ":" = ทุกแถว, "1" = คอลัมน์ที่สอง
#              (Python นับตำแหน่งเริ่มจาก 0) จึงได้ความน่าจะเป็นของการเป็นโรคเบาหวาน
p_train_oof = cross_val_predict(pipe, X_tr, y_tr, cv=5, method="predict_proba")[:, 1]
# สร้างตาราง (DataFrame) ด้วย pd.DataFrame([...]) จาก list ของ dict — dict 1 ตัว = 1 แถว
#   - [ {...} for t in [...] ] : "list comprehension" วนทีละเกณฑ์ t ในรายการ 0.2, 0.25, ..., 0.5
#     แล้วสร้าง dict ของแต่ละเกณฑ์ ได้ 7 แถว
#   - .round(3) : ปัดทุกตัวเลขในตารางเป็นทศนิยม 3 ตำแหน่ง
#   ผลลัพธ์เก็บไว้ในตัวแปร table (คอลัมน์ threshold, sensitivity, specificity)
table = pd.DataFrame([
    # คอลัมน์ "threshold" : ค่าเกณฑ์ตัดสิน t ของแถวนี้
    {"threshold": t,
     # คอลัมน์ "sensitivity" : (จำนวนผู้ป่วยจริงที่ความน่าจะเป็น >= t) / (จำนวนผู้ป่วยจริงทั้งหมด)
     #   - (p_train_oof >= t) : True เมื่อโมเดลจัดว่ามีความเสี่ยง
     #   - (y_tr == 1)        : True เมื่อเป็นโรคเบาหวานจริง (== คือการเปรียบเทียบว่าเท่ากันหรือไม่)
     #   - & : "และ" — True เมื่อเป็นจริงทั้งสองเงื่อนไข (= ผลบวกจริง TP)
     #   - .sum() : นับจำนวน True (True นับเป็น 1, False นับเป็น 0)
     "sensitivity": ((p_train_oof >= t) & (y_tr == 1)).sum() / (y_tr == 1).sum(),
     # คอลัมน์ "specificity" : (จำนวนผู้ไม่เป็นโรคที่ความน่าจะเป็น < t) / (จำนวนผู้ไม่เป็นโรคทั้งหมด)
     #   - (p_train_oof < t) & (y_tr == 0) : จัดว่าไม่เสี่ยง และไม่เป็นโรคจริง (= ผลลบจริง TN)
     "specificity": ((p_train_oof <  t) & (y_tr == 0)).sum() / (y_tr == 0).sum()}
    # วนค่า t จากรายการเกณฑ์ 7 ค่า แล้วปัดเศษทั้งตาราง
    for t in [0.2, 0.25, 0.3, 0.35, 0.4, 0.45, 0.5]]).round(3)
# แสดงตารางเป็นข้อความด้วยเมธอด .to_string(...) แล้วพิมพ์ด้วย print
#   - index=False : ไม่แสดงเลขลำดับแถว (0, 1, 2, ...) ทางซ้ายของตาราง
print(table.to_string(index=False))

# ----------------------------------------------------------------------------
# ส่วนที่ 2: กำหนดเกณฑ์ตัดสิน (ส่วนที่ผู้เรียนต้องแก้ไข)
# ----------------------------------------------------------------------------
# กำหนดเกณฑ์ตัดสินที่เลือกไว้ในตัวแปร THRESHOLD
#   ... คือช่องว่างให้ผู้เรียนแทนด้วยตัวเลขจากคอลัมน์ threshold ในตารางด้านบน ตามเงื่อนไขในคอมเมนต์ TODO
THRESHOLD = ...     # TODO: ระบุเกณฑ์จากตารางที่ sensitivity >= 0.80 และ specificity สูงที่สุด (คำใบ้: ประมาณ 0.3)

# ----------------------------------------------------------------------------
# ส่วนที่ 3: ประเมินเกณฑ์ที่เลือกกับชุดทดสอบ (ครั้งเดียว)
# ----------------------------------------------------------------------------
# ประเมินเกณฑ์ที่เลือกกับผู้ป่วยในชุดทดสอบที่แยกไว้ หนึ่งครั้ง
# คำนวณความน่าจะเป็นของการเป็นโรค (คลาส 1) ของผู้ป่วยชุดทดสอบ
#   - pipe.predict_proba(X_te) : เมธอดของไปป์ไลน์ที่ฝึกแล้ว คืนความน่าจะเป็นของคลาส 0 และ 1
#   - [:, 1] : เลือกทุกแถว เฉพาะคอลัมน์ความน่าจะเป็นของคลาส 1
p_test = pipe.predict_proba(X_te)[:, 1]
# สร้างผลการคัดกรอง flag : True = ความน่าจะเป็นตั้งแต่เกณฑ์ขึ้นไป (จัดว่ามีความเสี่ยง), False = ไม่เสี่ยง
flag = p_test >= THRESHOLD
# ความไว (sensitivity) ของชุดทดสอบ = (จัดว่าเสี่ยง และเป็นโรคจริง) / (ผู้ป่วยโรคเบาหวานจริงทั้งหมด)
test_sens = (flag & (y_te == 1)).sum() / (y_te == 1).sum()
# ความจำเพาะ (specificity) ของชุดทดสอบ = (จัดว่าไม่เสี่ยง และไม่เป็นโรคจริง) / (ผู้ไม่เป็นโรคทั้งหมด)
#   - ~flag : เครื่องหมาย ~ คือ "กลับค่า" (NOT) ของแต่ละราย True ↔ False จึงหมายถึง "จัดว่าไม่เสี่ยง"
test_spec = (~flag & (y_te == 0)).sum() / (y_te == 0).sum()
# แสดงเกณฑ์ที่เลือก พร้อม sensitivity และ specificity ของชุดทดสอบ ทศนิยม 3 ตำแหน่ง (:.3f)
#   - "\n" ด้านหน้า : ขึ้นบรรทัดใหม่ก่อนพิมพ์ เพื่อเว้นระยะจากตาราง
print(f"\nเกณฑ์ {THRESHOLD}: sensitivity ของชุดทดสอบ {test_sens:.3f}, specificity ของชุดทดสอบ {test_spec:.3f}")
````

````python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# เลือกเกณฑ์โดยอัตโนมัติ: เกณฑ์สูงสุดที่ sensitivity (out-of-fold ของชุดฝึก) >= 0.80
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   เฉลยของแบบฝึกหัดที่ 3 — แทนที่จะพิมพ์ THRESHOLD เอง โค้ดจะเลือกเกณฑ์ตัดสิน (threshold) ให้อัตโนมัติ
#   จากตารางที่คำนวณจาก "ชุดฝึกเท่านั้น" ดังนี้
#   - เลือกเฉพาะเกณฑ์ที่ให้ความไว (sensitivity) อย่างน้อย 0.80 (พลาดผู้ป่วยไม่เกินร้อยละ 20)
#   - ในบรรดาเกณฑ์เหล่านั้น เลือก "ค่าสูงสุด" ซึ่งให้ความจำเพาะ (specificity) สูงที่สุด (ผลบวกลวงน้อยที่สุด)
#   จากนั้นนำเกณฑ์ไปประเมินกับชุดทดสอบเพียงครั้งเดียว
#   - ข้อมูลนำเข้า : pipe, X_tr, y_tr, X_te, y_te จากเซลล์แบบฝึกหัดที่ 2 / เฉลย
#   - สิ่งที่ผู้เรียนจะเห็น : ตารางเกณฑ์ตัดสิน และ sensitivity / specificity ของชุดทดสอบที่เกณฑ์ที่เลือก
# ============================================================================

# นำเข้าฟังก์ชัน cross_val_predict จาก sklearn.model_selection
#   คืนผลการทำนายของผู้ป่วยแต่ละรายจากโมเดลที่ไม่เคยเห็นผู้ป่วยรายนั้นระหว่างฝึก (out-of-fold)
from sklearn.model_selection import cross_val_predict

# ----------------------------------------------------------------------------
# ส่วนที่ 1: สร้างตารางเกณฑ์ตัดสินจากชุดฝึก
# ----------------------------------------------------------------------------
# คำนวณความน่าจะเป็นแบบ out-of-fold ของการเป็นโรค (คลาส 1) สำหรับผู้ป่วยในชุดฝึก เก็บใน p_train_oof
#   - pipe                   : ไปป์ไลน์ที่จะถูกฝึกใหม่ในแต่ละรอบ
#   - X_tr, y_tr             : ข้อมูลชุดฝึกเท่านั้น (ไม่ใช้ชุดทดสอบ)
#   - cv=5                   : ตรวจสอบไขว้ 5 ส่วน
#   - method="predict_proba" : คืนความน่าจะเป็นของแต่ละคลาสแทนคำตอบ 0/1
#   - [:, 1]                 : ทุกแถว เฉพาะคอลัมน์ที่สอง (ความน่าจะเป็นของคลาส 1)
p_train_oof = cross_val_predict(pipe, X_tr, y_tr, cv=5, method="predict_proba")[:, 1]
# สร้างตาราง table ด้วย pd.DataFrame([...]) จาก list ของ dict (1 dict = 1 แถว)
#   - list comprehension [ {...} for t in [...] ] วนทีละเกณฑ์ t จาก 0.2 ถึง 0.5 ได้ 7 แถว
#   - .round(3) : ปัดตัวเลขเป็นทศนิยม 3 ตำแหน่ง
table = pd.DataFrame([
    # คอลัมน์ "threshold" : ค่าเกณฑ์ t ของแถวนี้
    {"threshold": t,
     # คอลัมน์ "sensitivity" : (ผู้ป่วยจริงที่ความน่าจะเป็น >= t) / (ผู้ป่วยจริงทั้งหมด)
     #   - & = "และ", == = เปรียบเทียบว่าเท่ากัน, .sum() = นับจำนวน True
     "sensitivity": ((p_train_oof >= t) & (y_tr == 1)).sum() / (y_tr == 1).sum(),
     # คอลัมน์ "specificity" : (ผู้ไม่เป็นโรคที่ความน่าจะเป็น < t) / (ผู้ไม่เป็นโรคทั้งหมด)
     "specificity": ((p_train_oof <  t) & (y_tr == 0)).sum() / (y_tr == 0).sum()}
    # วนค่า t จากรายการเกณฑ์ 7 ค่า แล้วปัดเศษทั้งตาราง
    for t in [0.2, 0.25, 0.3, 0.35, 0.4, 0.45, 0.5]]).round(3)
# พิมพ์ตารางเป็นข้อความ โดยไม่แสดงเลขลำดับแถว (index=False)
print(table.to_string(index=False))

# ----------------------------------------------------------------------------
# ส่วนที่ 2: เลือกเกณฑ์ตัดสินอัตโนมัติ
# ----------------------------------------------------------------------------
# กรองเฉพาะแถวของตารางที่ sensitivity >= 0.80 เก็บไว้ใน ok
#   - table["sensitivity"] >= 0.80 : ได้ค่า True/False ของแต่ละแถว
#   - table[ เงื่อนไข ] : เลือกเฉพาะแถวที่เงื่อนไขเป็น True (การกรองข้อมูลด้วยเงื่อนไขของ pandas)
ok = table[table["sensitivity"] >= 0.80]
# เลือกเกณฑ์ตัดสินเก็บไว้ใน THRESHOLD ด้วยนิพจน์แบบ "ค่า_ถ้าจริง if เงื่อนไข else ค่า_ถ้าเท็จ"
#   - len(ok) : จำนวนแถวที่ผ่านเงื่อนไข (หากเป็น 0 ถือว่าเป็นเท็จ)
#   - หากมีแถวผ่าน : ok["threshold"].max() = เกณฑ์ที่สูงที่สุดในกลุ่มนั้น แล้วแปลงเป็นตัวเลขทศนิยมธรรมดาด้วย float(...)
#   - หากไม่มีแถวใดผ่าน : ใช้เกณฑ์ต่ำสุดในตาราง คือ 0.2 แทน
THRESHOLD = float(ok["threshold"].max()) if len(ok) else 0.2   # เกณฑ์สูงสุด = specificity สูงที่สุด

# ----------------------------------------------------------------------------
# ส่วนที่ 3: ประเมินเกณฑ์ที่เลือกกับชุดทดสอบ (ครั้งเดียว)
# ----------------------------------------------------------------------------
# คำนวณความน่าจะเป็นของการเป็นโรค (คลาส 1) ของผู้ป่วยชุดทดสอบด้วยไปป์ไลน์ที่ฝึกแล้ว
#   - pipe.predict_proba(X_te) : คืนความน่าจะเป็นของคลาส 0 และ 1
#   - [:, 1] : เลือกเฉพาะคอลัมน์ของคลาส 1
p_test = pipe.predict_proba(X_te)[:, 1]
# flag = True เมื่อความน่าจะเป็นตั้งแต่เกณฑ์ขึ้นไป (จัดว่ามีความเสี่ยง)
flag = p_test >= THRESHOLD
# ความไว (sensitivity) ของชุดทดสอบ = (จัดว่าเสี่ยง และเป็นโรคจริง) / (ผู้ป่วยโรคเบาหวานจริงทั้งหมด)
test_sens = (flag & (y_te == 1)).sum() / (y_te == 1).sum()
# ความจำเพาะ (specificity) ของชุดทดสอบ = (จัดว่าไม่เสี่ยง และไม่เป็นโรคจริง) / (ผู้ไม่เป็นโรคทั้งหมด)
#   - ~flag : กลับค่า True ↔ False (NOT) จึงหมายถึง "จัดว่าไม่เสี่ยง"
test_spec = (~flag & (y_te == 0)).sum() / (y_te == 0).sum()
# แสดงเกณฑ์ที่เลือก พร้อม sensitivity และ specificity ของชุดทดสอบ ทศนิยม 3 ตำแหน่ง ("\n" = ขึ้นบรรทัดใหม่ก่อน)
print(f"\nเกณฑ์ {THRESHOLD}: sensitivity ของชุดทดสอบ {test_sens:.3f}, specificity ของชุดทดสอบ {test_spec:.3f}")
````

💡 **คำอธิบายผลลัพธ์** การลดเกณฑ์ตัดสินทำให้ตรวจพบผู้ป่วยโรคเบาหวานได้มากขึ้น (sensitivity เพิ่มขึ้น) แลกกับผลบวกลวงที่เพิ่มขึ้น (specificity ลดลง) โดยค่า ROC AUC ไม่เปลี่ยนแปลง มีเพียงจุดปฏิบัติงาน (operating point) ที่เปลี่ยนไป การเลือกจุดสมดุลนี้เป็น **การตัดสินใจทางคลินิก** (ความเสียหายจากการพลาดผู้ป่วยเทียบกับภาระของการตรวจเพิ่มเติม) มิใช่ประเด็นทางการเขียนโปรแกรม ตัวเลขจากชุดทดสอบมีความแปรปรวน (ชุดทดสอบมีผู้ป่วยโรคเบาหวานประมาณ 54 ราย) จึงอาจแตกต่างจากตารางเพียงเล็กน้อย

**ส่วนนี้ไม่บังคับ และไม่นับเป็นส่วนหนึ่งของชิ้นงานที่ต้องส่ง** เกณฑ์คัดกรองที่ผู้เรียนสร้างใน Lab 1 เป็นค่าอ้างอิง (baseline) แรก ในส่วนนี้ให้ส่ง **ความน่าจะเป็น (probability)** จากไปป์ไลน์ที่ *ไม่มีการรั่วไหลของข้อมูล* สำหรับ **ผู้ป่วย 154 รายชุดเดิมที่แยกไว้** **ตัวชี้วัด: ROC AUC** (ตารางอันดับแสดง sensitivity/specificity ที่เกณฑ์ 0.5 ประกอบด้วย) ไฟล์ชื่อ `lab02_submission.csv` มีส่วนหัว `id,prob` (prob = ความน่าจะเป็นของการเป็นโรคเบาหวาน) ให้ fit โมเดลกับแถว *อื่น* เท่านั้น

การส่งผล ให้ระบุ `NAME` และ `JOIN_CODE` (ผู้สอนประกาศในการเรียนการสอนผ่าน Zoom) ในเซลล์ ⚙️ การตั้งค่า แล้วรันเซลล์นั้นอีกครั้ง เซลล์ค่าอ้างอิงด้านล่างสามารถใช้งานได้ทันทีด้วยไปป์ไลน์อ้างอิงจากขั้นที่ 3

> **หลักจริยธรรมทางวิชาการ (Honour code)** label ที่ซ่อนไว้มาจากชุดข้อมูลสาธารณะ การค้นหาคำตอบจึงทำได้ง่ายแต่ไม่เกิดประโยชน์ ตารางอันดับมีไว้เพื่อสนับสนุนการเรียนรู้ และโน้ตบุ๊กของผู้เรียนจะได้รับการประเมินด้วย ห้ามแก้ไขค่าผลการทำนายด้วยตนเอง

````python
# 🏆 ส่งผลค่าอ้างอิง (baseline) — ไปป์ไลน์อ้างอิง fit กับแถวที่ไม่ใช่ชุดทดสอบเท่านั้น
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทาย — ไม่บังคับ)
#   เซลล์นี้สร้าง "ผลการทำนายค่าอ้างอิง (baseline)" สำหรับตารางอันดับ (leaderboard) ของรายวิชา
#   1) ดาวน์โหลดรายการรหัส (id) ของผู้ป่วยที่เป็น "ชุดทดสอบของตารางอันดับ" (label ถูกซ่อนไว้)
#   2) ฝึกสำเนาของไปป์ไลน์อ้างอิงจากขั้นที่ 3 ด้วยผู้ป่วยรายอื่นที่ "ไม่ใช่" ชุดทดสอบเท่านั้น
#   3) ทำนายความน่าจะเป็นของการเป็นโรคเบาหวานของผู้ป่วยชุดทดสอบ แล้วบันทึกเป็นไฟล์ lab02_submission.csv
#   4) หากกรอก NAME และ JOIN_CODE ไว้ในเซลล์การตั้งค่า จะส่งไฟล์ไปยังตารางอันดับโดยอัตโนมัติ
#   - ข้อมูลนำเข้า : X, y, pipe จากขั้นที่ 3 และฟังก์ชัน lab_data / submit_to_leaderboard จากเซลล์การตั้งค่า
#   - สิ่งที่ผู้เรียนจะเห็น : ข้อความยืนยันการบันทึกไฟล์ และคะแนน/อันดับ (หากส่งผล)
# ============================================================================

# นำเข้าฟังก์ชัน clone จาก sklearn.base
#   ใช้สร้าง "สำเนา" ของโมเดล/ไปป์ไลน์ที่มีการตั้งค่าเหมือนเดิม แต่ยังไม่ผ่านการฝึก (สิ่งที่เรียนรู้ไว้จะไม่ถูกคัดลอกมา)
from sklearn.base import clone

# ----------------------------------------------------------------------------
# ส่วนที่ 1: ระบุแถวของชุดทดสอบ (ตารางอันดับ) และชุดฝึก
# ----------------------------------------------------------------------------
# ดาวน์โหลดและอ่านรายการรหัสผู้ป่วยของชุดทดสอบ (ทำงานเป็นลำดับ)
#   1) lab_data("pima/test_ids.csv") : ฟังก์ชันจากเซลล์การตั้งค่า ดาวน์โหลดไฟล์แล้วคืนตำแหน่งไฟล์ในเครื่อง
#   2) pd.read_csv(...) : อ่านไฟล์ CSV เป็นตาราง
#   3) ["id"] : เลือกคอลัมน์ id (เลขแถวของผู้ป่วยในตาราง df)
#   เก็บผลไว้ในตัวแปร test_ids
test_ids = pd.read_csv(lab_data("pima/test_ids.csv"))["id"]
# สร้างตัวกรอง (mask) ว่าแถวใดเป็นชุดฝึก เก็บใน train_mask (ค่า True/False หนึ่งค่าต่อหนึ่งแถว)
#   - X.index : เลขแถว (index) ของตาราง X
#   - .isin(test_ids) : True เมื่อเลขแถวนั้นอยู่ในรายการ test_ids
#   - ~ : กลับค่า True ↔ False จึงได้ True เฉพาะแถวที่ "ไม่ใช่" ชุดทดสอบ
train_mask = ~X.index.isin(test_ids)

# ----------------------------------------------------------------------------
# ส่วนที่ 2: ฝึกไปป์ไลน์อ้างอิงและทำนาย
# ----------------------------------------------------------------------------
# สร้างสำเนาที่ยังไม่ผ่านการฝึกของ pipe (เติมค่า median → ป่าสุ่ม 300 ต้น, random_state=42) เก็บใน baseline_pipe
#   - pipe : ไปป์ไลน์อ้างอิงจากขั้นที่ 3 ที่ต้องการคัดลอกการตั้งค่า
baseline_pipe = clone(pipe)                                   # สำเนาไปป์ไลน์อ้างอิงจากขั้นที่ 3 (ยังไม่ผ่านการฝึก)
# ฝึกสำเนาไปป์ไลน์ด้วยเฉพาะแถวของชุดฝึก
#   - X[train_mask] : เลือกเฉพาะแถวของ X ที่ train_mask เป็น True (ตัวแปรนำเข้าของชุดฝึก)
#   - y[train_mask] : คำตอบของแถวเดียวกัน
baseline_pipe.fit(X[train_mask], y[train_mask])               # ห้าม fit กับแถวของชุดทดสอบ
# ทำนายความน่าจะเป็นของการเป็นโรคเบาหวานสำหรับผู้ป่วยชุดทดสอบ เก็บใน prob
#   - X.loc[test_ids] : เลือกแถวของ X ตามเลขแถว (index) ที่อยู่ใน test_ids โดยเรียงตามลำดับของ test_ids
#   - .predict_proba(...) : คืนความน่าจะเป็นของคลาส 0 และ 1
#   - [:, 1] : เลือกเฉพาะความน่าจะเป็นของคลาส 1 (เป็นโรค)
prob = baseline_pipe.predict_proba(X.loc[test_ids])[:, 1]

# ----------------------------------------------------------------------------
# ส่วนที่ 3: บันทึกไฟล์และส่งผล (ไม่บังคับ)
# ----------------------------------------------------------------------------
# สร้างตารางผลการทำนาย แล้วบันทึกเป็นไฟล์ CSV (ทำงานเป็นลำดับ)
#   1) pd.DataFrame({...}) : สร้างตารางจาก dict แบบ {ชื่อคอลัมน์: ค่าของคอลัมน์}
#       - "id"   : test_ids.values = รหัสผู้ป่วย (.values ดึงเฉพาะตัวเลข ไม่เอาเลขแถวของ test_ids มาด้วย)
#       - "prob" : prob = ความน่าจะเป็นที่ทำนาย
#   2) .to_csv(...) : บันทึกตารางเป็นไฟล์ CSV
#       - "lab02_submission.csv" : ชื่อไฟล์ที่จะบันทึก
#       - index=False : ไม่บันทึกเลขลำดับแถวลงในไฟล์
pd.DataFrame({"id": test_ids.values, "prob": prob}).to_csv("lab02_submission.csv", index=False)
# แสดงข้อความยืนยันการบันทึกไฟล์ พร้อมจำนวนแถว
#   - len(prob) : ฟังก์ชัน len นับจำนวนสมาชิก (จำนวนผู้ป่วยที่ทำนาย)
print("บันทึกไฟล์ lab02_submission.csv เรียบร้อย จำนวน", len(prob), "แถว")
# if ... else : ตรวจเงื่อนไข — หากกรอกทั้ง NAME และ JOIN_CODE แล้ว (ไม่ใช่ข้อความว่าง) ให้ทำคำสั่งที่ย่อหน้าใต้ if
#   - and : "และ" ต้องเป็นจริงทั้งสองค่า
if NAME and JOIN_CODE:
    # ส่งไฟล์ไปยังตารางอันดับด้วยฟังก์ชันจากเซลล์การตั้งค่า แล้วแสดงคะแนนและอันดับ
    #   - "lab02_submission.csv" : ไฟล์ผลการทำนายที่จะส่ง
    submit_to_leaderboard("lab02_submission.csv")
# else : กรณีที่เงื่อนไขของ if เป็นเท็จ (ยังไม่ได้กรอก NAME หรือ JOIN_CODE)
else:
    # แสดงข้อความแนะนำวิธีส่งผล (การส่งผลไม่บังคับ)
    print("หากต้องการส่งผล โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่า (ไม่บังคับ)")
````

**แนวทางการทำคะแนนให้สูงกว่าค่าอ้างอิง** (สามารถปรึกษาผู้ช่วยปัญญาประดิษฐ์ และต้องตรวจสอบทุกข้อเสนอด้วยรายการตรวจสอบก่อนนำไปใช้)
- เปรียบเทียบโมเดลด้วยการตรวจสอบไขว้แบบ 5-fold *ชุดเดียวกัน* บนแถวของชุดฝึก: `LogisticRegression` (ใส่ `StandardScaler` **ภายใน** Pipeline), `HistGradientBoostingClassifier` (รองรับค่า NaN โดยตรง) หรือ random forest ที่ปรับค่า (`min_samples_leaf`, `max_features`)
- เพิ่มตัวแปรบ่งชี้ค่าที่ขาดหาย (`SimpleImputer(add_indicator=True)`) เนื่องจากการ "ไม่ได้ตรวจวัด insulin" อาจเป็นข้อมูลที่มีความหมาย
- สร้างตัวแปรใหม่ (feature engineering) ภายในไปป์ไลน์ (เช่น glucose × BMI) หรือเฉลี่ยความน่าจะเป็นจากโมเดล 2–3 โมเดล
- ปรับไฮเปอร์พารามิเตอร์ด้วย `GridSearchCV` บนแถวของชุดฝึกเท่านั้น ทั้งนี้ การปรับเกณฑ์ตัดสินไม่ทำให้ค่า ROC AUC เปลี่ยนแปลง

````python
# 🏆 ไม่บังคับ — ไปป์ไลน์ที่ผู้เรียนปรับปรุง (ต้องไม่มี data leakage; หากกำหนดเป็น None จะใช้ค่าอ้างอิงต่อไป)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้ (โจทย์ท้าทาย — ไม่บังคับ สำหรับผู้เรียนที่เขียนโปรแกรมได้)
#   เซลล์นี้เปิดให้ผู้เรียนลองสร้างไปป์ไลน์ของตนเอง (my_pipe) เพื่อทำคะแนนให้สูงกว่าค่าอ้างอิง
#   - หาก my_pipe = None (ค่าเริ่มต้น) : เซลล์จะไม่ทำอะไร และใช้ไฟล์ผลการทำนายของค่าอ้างอิงต่อไป
#   - หากกำหนด my_pipe เป็น Pipeline : เซลล์จะ
#       1) วัด ROC AUC ด้วยการตรวจสอบไขว้ 5-fold บนแถวของชุดฝึกเท่านั้น
#       2) ฝึกด้วยแถวของชุดฝึก แล้วทำนายความน่าจะเป็นของผู้ป่วยชุดทดสอบ
#       3) บันทึกทับไฟล์ lab02_submission.csv และส่งผลหากกรอก NAME / JOIN_CODE แล้ว
#   - ข้อมูลนำเข้า : X, y, train_mask, test_ids จากเซลล์ค่าอ้างอิงด้านบน
#   - สิ่งที่ผู้เรียนจะเห็น : ROC AUC ของไปป์ไลน์ที่ปรับปรุง หรือข้อความว่ายังไม่ได้กำหนดไปป์ไลน์
#   ⚠️ ไปป์ไลน์ต้องไม่มีการรั่วไหลของข้อมูล (data leakage) — ทุกขั้นตอนเติมค่า/ปรับสเกลต้องอยู่ภายใน Pipeline
# ============================================================================

# กำหนดไปป์ไลน์ของผู้เรียนไว้ในตัวแปร my_pipe
#   - None : ค่าพิเศษของ Python หมายถึง "ไม่มีค่า / ยังไม่ได้กำหนด"
#   - คอมเมนต์ด้านขวาเป็นตัวอย่างไปป์ไลน์ที่ปรับปรุง: SimpleImputer(add_indicator=True) เพิ่มคอลัมน์บ่งชี้ว่าค่าใดขาดหาย
#     และ HistGradientBoostingClassifier เป็นโมเดลอีกชนิดหนึ่ง (หากนำไปใช้ ต้อง import คลาสนี้จาก sklearn.ensemble ก่อน)
my_pipe = None      # ตัวอย่าง: Pipeline([("imputer", SimpleImputer(strategy="median", add_indicator=True)),
                    #                     ("model", HistGradientBoostingClassifier(random_state=42))])

# if : ทำคำสั่งที่ย่อหน้าด้านล่าง เฉพาะเมื่อผู้เรียนกำหนด my_pipe แล้ว
#   - is not None : ตรวจว่าตัวแปร "ไม่ใช่" None
if my_pipe is not None:
    # วัด ROC AUC ด้วยการตรวจสอบไขว้ เก็บคะแนน 5 ค่าไว้ใน cv_auc
    #   - my_pipe                       : ไปป์ไลน์ของผู้เรียน
    #   - X[train_mask], y[train_mask]  : เฉพาะแถวของชุดฝึก (ไม่แตะชุดทดสอบ)
    #   - cv=5                          : 5 ส่วน (fold)
    #   - scoring="roc_auc"             : ใช้ ROC AUC เป็นคะแนน
    cv_auc = cross_val_score(my_pipe, X[train_mask], y[train_mask], cv=5, scoring="roc_auc")
    # แสดง ROC AUC เฉลี่ย ± ส่วนเบี่ยงเบนมาตรฐาน ทศนิยม 3 ตำแหน่ง (:.3f)
    print(f"ไปป์ไลน์ของผู้เรียน — CV ROC AUC บนแถวของชุดฝึก: {cv_auc.mean():.3f} ± {cv_auc.std():.3f}")
    # ฝึกไปป์ไลน์ของผู้เรียนด้วยแถวของชุดฝึกทั้งหมด (X[train_mask] = ตัวแปรนำเข้า, y[train_mask] = คำตอบ)
    my_pipe.fit(X[train_mask], y[train_mask])                 # ห้าม fit กับแถวของชุดทดสอบ
    # ทำนายความน่าจะเป็นของการเป็นโรค (คลาส 1) ของผู้ป่วยชุดทดสอบ
    #   - X.loc[test_ids] : แถวของชุดทดสอบตามรหัส test_ids
    #   - [:, 1]          : เฉพาะคอลัมน์ความน่าจะเป็นของคลาส 1
    prob = my_pipe.predict_proba(X.loc[test_ids])[:, 1]
    # สร้างตาราง id/prob แล้วบันทึกทับไฟล์ lab02_submission.csv (index=False = ไม่บันทึกเลขลำดับแถว)
    pd.DataFrame({"id": test_ids.values, "prob": prob}).to_csv("lab02_submission.csv", index=False)
    # แสดงข้อความยืนยันการบันทึกไฟล์
    print("บันทึกไฟล์ lab02_submission.csv เรียบร้อย")
    # ส่งผลเฉพาะเมื่อกรอกทั้ง NAME และ JOIN_CODE ในเซลล์การตั้งค่าแล้ว
    if NAME and JOIN_CODE:
        # ส่งไฟล์ไปยังตารางอันดับ แล้วแสดงคะแนนและอันดับ
        submit_to_leaderboard("lab02_submission.csv")
# else : กรณี my_pipe ยังเป็น None (ผู้เรียนยังไม่ได้กำหนดไปป์ไลน์)
else:
    # แสดงข้อความว่าจะใช้ไฟล์ผลการทำนายของค่าอ้างอิงต่อไป
    print("ยังไม่ได้กำหนดไปป์ไลน์ที่ปรับปรุง — ใช้ไฟล์ผลการทำนายของค่าอ้างอิงต่อไป")
````
